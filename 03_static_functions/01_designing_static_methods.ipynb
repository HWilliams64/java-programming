# Designing Static Methods

In this lesson, you will learn to define reusable static methods, supply typed inputs, and return results separately from displaying them.

CSC-239 · Module 3 · Lesson 1 of 2

You have used arithmetic, decisions, and loops to describe a program's work. Now you will give a calculation a name so several parts of a program can request the same operation with different values. You will also distinguish a caller's variable from a method's local input variable.

The examples use workshop orders, but the same design applies to many repeated calculations. Use the [Module 3 glossary](terms.md) for the terms introduced here.


## Learning Goals

- Define, call, and test a static method with typed parameters and a returned result.
- Separate a reusable calculation from console output, and distinguish returned values from void actions.
- Explain why changing a numeric parameter does not replace the caller's variable.


## Why This Matters

A larger program may need the same rule in several places. An order screen, a receipt, and a report might all calculate a subtotal. Copying that multiplication into each part leaves several places to check when the rule changes. A named method gives those callers one operation to use.

Returning the result also gives the caller a choice. One caller can display the number, another can store it, and another can use it in a later calculation. A helper that only prints its answer does not give those callers a numeric result to use.

This lesson develops that separation between an operation and the code that requests it. The next lesson will put a loop and its decisions inside a method that returns selected text. The structure will be the same even though the work inside becomes more involved.

## Concept

### Give a shared calculation a name

A workshop order needs a quantity and a unit price. Quantity counts items. Unit price is the price of one item in dollars. Multiplying them gives the order's total in dollars. That rule stays the same when the values change.

A **static method** is a named operation that belongs to a class. Calling the method asks Java to perform that operation. In Java, the kind of function we are building is written as a method inside a class.

```java
class PriceTools {
    static double orderTotal(int quantity, double unitPrice) {
        return quantity * unitPrice;
    }
}
```

The keyword **`class`** introduces `PriceTools`, the container for this calculation. Inside it, `orderTotal` names the operation. The keyword **`static`** lets us request this operation through the class name: `PriceTools.orderTotal(...)`. We do not create a separate PriceTools object for this call. Methods that act on individual objects come later in the course.

The word `double` immediately before `orderTotal` is its **return type**: the type of value this method provides. You already know double as a decimal numeric type. Here it tells a caller what type of result to expect. The inner braces enclose the method's body, while the outer braces enclose the class. Declaring the body describes the operation; it does not run the calculation by itself.

The items inside the parentheses name the inputs needed for one call. That input list lets the same body work with different orders.


### Match arguments to typed parameters

A **parameter** is an input variable declared in a method. The parameter list above declares `int quantity` and `double unitPrice`. Each parameter has its own type and name. A comma separates the two declarations.

An **argument** is a value supplied at a call. You have already supplied arguments to methods such as println. The **caller** is the code requesting the method. In the call `PriceTools.orderTotal(3, 4.5)`, the caller supplies 3 followed by 4.5.

| Argument position | Supplied value | Parameter receiving that value | Meaning |
|---|---|---|---|
| First | 3 | int quantity | 3 items |
| Second | 4.5 | double unitPrice | 4.5 dollars per item |

Position matters: Java matches these arguments to parameters in the declared order. The method then uses its parameter names inside the body. A caller can supply a literal, a variable, or an expression that produces a suitable value. Its variable names do not need to match the parameter names.

Each call creates its own parameter variables. A later call with 5 and 2.5 gives the body a new quantity and price. Parameters have local scope within the method; code outside that method cannot use them by those local names. The caller and method communicate through the supplied inputs and returned result.


### Return the calculated value

A **return value** is the result sent from a method to its caller. The keyword **`return`** introduces the statement that provides it:

```java
return quantity * unitPrice;
```

Java first evaluates the multiplication using this call's parameter values. It then ends this method call and gives the resulting double value back to the caller. A method declared to return a value must provide a compatible result whenever a normal path through its body finishes. This method has one calculation and one return, so that return covers its work.

The complete program below declares the helper and makes two calls. In each print statement, Java must finish the inner `orderTotal` call before println can print its result. The first call calculates 3 times 4.5, returning 13.5. The second calculates 5 times 2.5, returning 12.5.

Returning is not printing. There is no println inside `orderTotal`. The two statements outside the class produce the console output. We use explicit print statements so the intended output is clear; an expression displayed by a notebook is not a print statement inside the method.


````java
class PriceTools {
    static double orderTotal(int quantity, double unitPrice) {
        return quantity * unitPrice;
    }
}
System.out.println(PriceTools.orderTotal(3, 4.5));
System.out.println(PriceTools.orderTotal(5, 2.5));
````

Expected output:

```text
13.5
12.5
```

Both results are in dollars. The first line belongs to the first argument pair and the second to the next pair. Each call uses the same body with fresh parameter values. The caller's println statements display the returned numbers.

### Store the result before displaying it

A returned value can be used anywhere a suitable value is needed. The caller does not have to print it immediately:

```java
double total = PriceTools.orderTotal(3, 4.5);
System.out.println(total);
```

The method call on the right finishes first. It returns 13.5, and the assignment stores that value in the caller's variable `total`. The next statement reads total and prints it. This leaves the number available for another calculation as well.


The caller's total stores the returned double. It is a different variable from the method's quantity and unitPrice parameters. Printing the value does not remove it from total.


### Call a method for an action

Some operations perform an action without supplying a result value. A **void method** does that: the keyword **`void`** appears where a return type would otherwise appear. It means the call provides no value to store or use in another expression.

The next helper, `ReceiptDisplay.showTotal`, accepts one double parameter named total. Its body prints a label and that value. The caller first prints a message, calls showTotal as a complete statement, and then prints another message.

```java
ReceiptDisplay.showTotal(9.0);
```

This call stands alone. We do not assign it to a double variable or put it inside println, because it has no return value. When the method body reaches its end, control goes back to the caller's next statement. No word such as void appears in the output.

Printing is a **side effect**: an observable action beyond supplying a return value. In this method, the println in the body causes that effect. The total calculation used earlier leaves printing to its caller; this display method performs the output itself. Void does not mean that every method must print. It only describes the absence of a returned value.


````java
class ReceiptDisplay {
    static void showTotal(double total) {
        System.out.println("Total: " + total);
    }
}
System.out.println("Before display");
ReceiptDisplay.showTotal(9.0);
System.out.println("After display");
````

Expected output:

```text
Before display
Total: 9.0
After display
```

The first and last lines come from the caller. The middle line comes from inside showTotal, which receives total equal to 9.0. Once that body finishes, the caller continues. This is a useful output action, but it does not produce a numeric result for another calculation.

### Keep a calculation separate from outside effects

A **pure function** computes its result from its inputs without changing state outside itself. The same inputs give the same result. Our orderTotal method fits this description: it reads quantity and unitPrice, multiplies them, and returns the product.

```java
static double orderTotal(int quantity, double unitPrice) {
    return quantity * unitPrice;
}
```

Inspect the body for the reason. It does not print, read changing information elsewhere, or change another part of the program. Printing still happens in the caller, so the complete program has output even though this calculation is pure. Keeping those jobs separate makes the calculation easier to reuse and test.

The showTotal method has a different purpose. Its body prints, so that operation has a side effect. Neither design is automatically wrong; choose the operation that a caller needs. The keyword static describes how the method belongs to its class. It does not guarantee that the method is pure.

To check a calculation, choose inputs and calculate the expected result, then compare it with the actual returned value. Repeat an input and test a relevant boundary such as zero quantity. These observations provide evidence about those cases. Matching results from two runs alone cannot prove purity; the body's work must support that claim.


### Separate a numeric parameter from the caller's variable

Java uses **pass-by-value**: each parameter receives a copy of the value supplied by its argument. For an int argument, the copied value is an integer. Assigning a different number to the parameter changes that local variable, not the caller's variable.

The next method, `QuantityTools.addOne`, first adds one to its local quantity parameter and then returns it. The caller begins with `requested` equal to 3:

```java
int requested = 3;
int revised = QuantityTools.addOne(requested);
```

At the call, Java copies the value 3 into the method's quantity parameter. Inside the body, `quantity = quantity + 1;` reads that copy and stores 4 back into the local parameter. The return sends 4 to the caller. The assignment to revised stores it there.

There is no assignment to requested in this program. Requested therefore remains 3 while revised becomes 4. To replace requested deliberately, the caller would have to assign the returned result to requested instead. This is like choosing to store the result of trim rather than ignoring it.

The local reassignment does not create an outside effect in addOne. It changes only that call's parameter. This example concerns primitive numeric values; later lessons will explain what gets copied when an argument refers to an object. Do not extend this numeric example into a claim that objects are copied.


````java
class QuantityTools {
    static int addOne(int quantity) {
        quantity = quantity + 1;
        return quantity;
    }
}
int requested = 3;
int revised = QuantityTools.addOne(requested);
System.out.println(requested);
System.out.println(revised);
````

Expected output:

```text
3
4
```

Requested keeps its original value. Revised stores the returned 4. Once the method call ends, its local quantity parameter is no longer available. The caller's two variables remain available to the following print statements.

## Worked Example

A workshop needs totals for two orders: 3 items at 4.5 dollars per item and 5 items at 2.5 dollars per item. Both orders use the same multiplication. The purpose is to keep that rule in one helper and let the caller display each result.

**Subgoal 1: declare the operation and inputs.** PriceTools holds the static orderTotal method. Its typed parameters tell the caller to supply a whole-number quantity followed by a decimal unit price.

**Subgoal 2: compute and return.** The method body multiplies the parameters and returns the double result. It does not decide how that number should be displayed.

**Subgoal 3: call the helper for each order.** Each caller statement supplies one argument pair. Java completes the calculation and then gives its returned result to the enclosing println.


````java
class PriceTools {
    static double orderTotal(int quantity, double unitPrice) {
        return quantity * unitPrice;
    }
}
System.out.println(PriceTools.orderTotal(3, 4.5));
System.out.println(PriceTools.orderTotal(5, 2.5));
````

Expected output:

```text
13.5
12.5
```

For the first call, quantity is 3 and unitPrice is 4.5, so the method returns 13.5. The caller prints it. The second call gets new values, 5 and 2.5, and returns 12.5 to its caller print. Neither call needs another copy of the multiplication rule.

These decimal numbers represent dollars in our example. We are practicing methods and returned values, not special currency formatting. Another caller could store either result before using it, as the earlier total variable did.

## Summary

A static method defines a reusable operation in a class. Parameters name its typed inputs, while arguments supply values for one call. Return sends a result to the caller. The caller chooses whether to store it, print it, or use it in another calculation.

A void method supplies no result value, though it can perform an action such as printing. A pure calculation gets its result from its inputs without changing outside state. Reassigning a numeric parameter changes only that local copy; a caller variable changes only when the caller assigns a new value to it.
