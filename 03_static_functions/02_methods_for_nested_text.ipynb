## Why This Matters

Software often needs one useful part of a longer text value. A stored note may contain a marked description, while the rest of the note contains comments for a different reader. Selecting the right section lets another part of the application store or display it without copying the surrounding text.

The selection rule matters when marked sections can contain other marked sections. Stopping at the first closing symbol may cut off part of the requested content. Tracking which openings still need a closing symbol lets a program distinguish the end of an inner section from the end of the whole section.

Putting that scan in a method keeps the selection rule in one place. A clear input and result contract also tells callers how to interpret missing or empty text. These habits carry into later programs that process collections and files: separate the processing rule from its display, and test the cases where its result could be misunderstood.

## Concept

### Number the positions before reading a character

A workshop coordinator stores a color description inside square brackets in a note. Text outside the outer brackets is a general comment. Brackets inside that description belong to the description and must stay in the selected result. To identify the boundaries, the program first needs a way to examine each position in the String.

An **index** is a number identifying a position. Java String indices start at 0. In `"[blue]"`, the first position contains the opening bracket, and the last position contains the closing bracket:

| Index | 0 | 1 | 2 | 3 | 4 | 5 |
|---|---|---|---|---|---|---|
| Symbol | [ | b | l | u | e | ] |

The String has length 6, but its last valid index is 5. Length counts the positions; it does not name the last one. An empty String has length 0 and no position that `charAt` can select.

The **charAt method** selects a **character value** at the index supplied as its argument. The Java keyword **`char`** declares a variable for that value:

```java
String marked = "[blue]";
char symbol = marked.charAt(0);
```

The first statement stores the text. The second selects its opening bracket and stores that value in `symbol`. It reads from `marked`; it does not remove or change the selected symbol.

These examples use ordinary letters and delimiter symbols that each occupy one String position. More precisely, a Java `char` represents one UTF-16 code unit, a piece used to encode text. Some displayed symbols need more than one such piece. The positions in our supplied examples can be read directly from the table.

### Recognize an opening or closing value

A **delimiter pair** is an opening symbol and a closing symbol that mark a section's boundaries. We use square brackets first and parentheses in independent practice. A **character literal** writes one `char` value with single quotation marks, such as `'['` or `']'`.

```java
System.out.println(symbol == '[');
```

Here `==` compares two character values: the selected `symbol` and the literal opening bracket. Both have type `char`, so this comparison produces `true`. The double-quoted value `"["` is a String instead. Keep using `equals` when comparing String contents; a primitive `char` is a different kind of value.

The complete cell below first prints the selected symbol and then prints the comparison result. No earlier cell is needed for its setup.


````java
String marked = "[blue]";
char symbol = marked.charAt(0);
System.out.println(symbol);
System.out.println(symbol == '[');
````

Expected output:

```text
[
true
```

The first line is the character at index 0. The second line is a Boolean result from comparing that character with `'['`. Printing the character and testing its identity are separate operations.

### Select a range of text

Selecting one character helps us recognize a boundary. Once both boundaries are known, we need the whole section between them. A **substring range** identifies that section using an included start index and an excluded end index.

The **substring method** returns the selected text. Consider this complete short example:

```java
String marked = "[blue]";
System.out.println(marked.substring(1, 5));
```

The first argument, 1, selects the position holding `b`. The second argument, 5, is the first position to leave out, which holds the closing bracket. The method therefore includes indices 1, 2, 3, and 4 and returns `"blue"`. The caller prints that returned String. The original `marked` value still contains both brackets.

Equal valid bounds select no positions. For example, `marked.substring(1, 1)` starts at index 1 but stops before that same index, so it returns an empty String. It does not select the character at index 1.

Every range must start at or after 0, end at or before the String's length, and have a start no greater than its end. The scan will record the position just after an outer opening bracket and use the matching closing bracket's position as the excluded end.

The next complete cell adds labels and angle brackets around each returned value. Those extra characters belong to the caller's display. They make an empty result visible without changing what `substring` returns.


````java
String marked = "[blue]";
System.out.println("Slice: <" + marked.substring(1, 5) + ">");
System.out.println("Empty: <" + marked.substring(1, 1) + ">");
````

Expected output:

```text
Slice: <blue>
Empty: <>
```

The first range contributes `blue` between the caller's angle brackets. The equal-bounds range contributes no text, so the second line has adjacent angle brackets. This display helps distinguish an empty result from a program that printed no line.

### Count openings that still need closing symbols

One pair such as `[blue]` has an opening and a closing boundary. A longer description can contain another pair, as in `[m[n]p]`. Returning at its first closing bracket would finish too soon: that bracket closes the inner `[n]` section while the outer section still includes `p`.

**Nesting depth** is a count of opening delimiters that have not yet been closed. Begin at 0. An opening bracket increases depth by 1. A closing bracket decreases it by 1 when there is an active opening to close. Letters leave the count unchanged.

| Delimiter reached in `[m[n]p]` | Depth before | Depth after | Meaning |
|---|---|---|---|
| First `[` | 0 | 1 | The outer group begins |
| Second `[` | 1 | 2 | An inner group begins |
| First `]` | 2 | 1 | The inner group ends; the outer group remains open |
| Second `]` | 1 | 0 | The outer group is complete |

We also need to remember where the outer content begins. A variable named `start` initially holds -1 as a marker meaning that no starting position has been recorded. It is a marker, not a valid substring index.

When an opening bracket arrives with depth 0, record `start = index + 1`. Adding 1 moves past the opening bracket. When another opening arrives at a positive depth, preserve the existing start. Otherwise, the method would forget the beginning of the outer content and select only a later part of it.

After processing a closing bracket, a return to depth 0 tells us that the recorded outer group is complete. The method can then return the range from `start` up to, but excluding, that closing bracket.


### Finish the method when its answer is known

An **early return** ends a method before it reaches the end of its body. Here the required answer is known when a closing bracket completes the outer group. The following fragment belongs inside the closing-bracket branch, immediately after reducing depth:

```java
if (depth == 0) {
    return text.substring(start, index);
}
```

The condition checks the updated depth. If it is zero, `substring` constructs the selected result and `return` sends it to the caller. That return ends the whole method call. The loop does not perform another update or search a later group.

This differs from `break`, which would leave the loop and continue with statements after it. Our complete method has a different return after the entire loop:

```java
return "";
```

That final statement handles a search that reaches the end of the String without completing an outer group. It is outside the loop and outside both bracket branches. Keeping these two locations distinct explains how a method can return an answer immediately in one case and supply an empty result after a complete scan in another.


### Agree on the meaning of each result

An **input and result contract** states what a method accepts and what its returned values mean. This method follows these rules:

- The input is a String containing actual text, which may be empty. It must not be `null`, Java's value meaning that no object reference is supplied.
- Return the content of the first completed outer square-bracket group.
- Keep any inner brackets in that content, while leaving out the two outer brackets.
- Ignore a closing bracket that occurs before an opening bracket.
- Return an empty String if no outer group completes.
- Return an empty String for a completed empty group too.

The last two rules deliberately give the same result for different situations. An empty result by itself cannot say whether the input contained an empty pair, no pair, or an unfinished pair. A caller that needs those distinctions would need a different contract; it must not guess from the returned text alone.

The scan will still be a pure calculation. It reads its String parameter, changes only local counters and positions, and returns a String. It leaves the input text unchanged and does not print. The caller decides how to show the result, including the visible labels around empty text.


### Assemble the scan and trace every position

The complete method below follows one position at a time through `[m[n]p]`. Before reading that program, connect each part to the work it performs.

```java
int start = -1;
int depth = 0;
for (int index = 0; index < text.length(); index = index + 1) {
    char symbol = text.charAt(index);
```

The first two declarations prepare the unrecorded start and the count of open groups. The loop begins at the first String index, 0. Its strict `<` comparison stops before the String's length, so `charAt(index)` reads only existing positions. An empty String fails that test immediately and skips the body.

```java
if (symbol == '[') {
    if (depth == 0) {
        start = index + 1;
    }
    depth = depth + 1;
}
```

The first condition recognizes an opening bracket. The inner condition records a new start only when no group is already open. The depth increase is outside the inner condition, so it counts every opening bracket, including an inner one.

```java
} else if (symbol == ']' && depth > 0) {
    depth = depth - 1;
    if (depth == 0) {
        return text.substring(start, index);
    }
}
```

This fragment shows the alternative attached to the opening-bracket branch. Both conditions must hold: the current symbol is a closing bracket, and an opening is active. A leading closing bracket is therefore ignored instead of making depth negative. The body first decreases depth and then checks whether the outer group has finished. A letter matches neither branch and leaves `start` and `depth` unchanged.

Follow the full trace. The `start` column gives its value after that position is processed:

| Index | Symbol | Start | Depth before → after | Action |
|---|---|---|---|---|
| 0 | `[` | 1 | 0 → 1 | Record the outer content start and count its opening |
| 1 | `m` | 1 | 1 → 1 | Keep scanning; the letter changes neither value |
| 2 | `[` | 1 | 1 → 2 | Count the inner opening without replacing start |
| 3 | `n` | 1 | 2 → 2 | Keep both groups open |
| 4 | `]` | 1 | 2 → 1 | Close the inner group; the outer group is unfinished |
| 5 | `p` | 1 | 1 → 1 | Keep this letter inside the outer content |
| 6 | `]` | 1 | 1 → 0 | Complete the outer group and return `substring(1, 6)` |

The selected positions are 1 through 5, producing `m[n]p`. The original marker -1 is never used as a range boundary: reaching a permitted closing branch requires an earlier opening, and that opening recorded the real start. After the return at index 6, the loop's update is not reached.

The following cell includes the complete helper and its caller. The method declaration, all local setup, both branches, and the final empty return are present together.


````java
class TextTools {
    static String insideOuterBrackets(String text) {
        int start = -1;
        int depth = 0;
        for (int index = 0; index < text.length(); index = index + 1) {
            char symbol = text.charAt(index);
            if (symbol == '[') {
                if (depth == 0) {
                    start = index + 1;
                }
                depth = depth + 1;
            } else if (symbol == ']' && depth > 0) {
                depth = depth - 1;
                if (depth == 0) {
                    return text.substring(start, index);
                }
            }
        }
        return "";
    }
}
System.out.println(TextTools.insideOuterBrackets("[m[n]p]"));
````

Expected output:

```text
m[n]p
```

The returned range keeps the inner `[n]` pair and the letters on either side. Only the outer brackets are excluded. The caller prints the String after the method returns; no printing occurs inside the scan.

### Compare a completed group with an empty result

The same method can reach its two return statements through different inputs. Each complete cell below starts a separate call with fresh `start`, `depth`, and `index` values.

For `[a] [b]`, the closing bracket at index 2 completes the first group. The method returns `a` immediately. The space and the later `[b]` group are not scanned.

For `[]`, the opening stores start 1, and the closing at index 1 completes the group. The selected range is `substring(1, 1)`, so this early return supplies an empty String.

The input `plain` has no opening bracket. The input `[open` has an opening but no closing bracket. Both scans reach the end of the String and use the final `return "";` after the loop. Their empty result has the same text as the result for `[]`, but the path to it differs.

## Worked Example

The coordinator's complete note is `"Note [red[blue]green] end."`. The required color description is the content of the first outer group, including its inner bracketed section. The words before and after that group must stay out of the returned result.

**Step 1: prepare one search.** The method takes a String and declares a String return type. It starts with no recorded boundary and depth zero. The loop examines positions from 0 while the index remains below the length.

**Step 2: preserve the outer start.** The first opening bracket, at index 5, records start 6 and raises depth to 1. The next opening, at index 9, raises depth to 2 but leaves start at 6. This preserves the beginning of `red`.

**Step 3: finish only the matching outer group.** The closing bracket at index 14 ends the inner group, reducing depth to 1. The later closing bracket at index 20 reduces depth to 0. Only that second closing position completes the required group.

**Step 4: return the selected range.** `substring(6, 20)` includes positions 6 through 19. The caller prints that returned String. The return ends the method before it reaches the remaining ` end.` text.

Run the assembled helper and caller below. The final return after the loop remains necessary for other inputs whose search never completes a group.


````java
class TextTools {
    static String insideOuterBrackets(String text) {
        int start = -1;
        int depth = 0;
        for (int index = 0; index < text.length(); index = index + 1) {
            char symbol = text.charAt(index);
            if (symbol == '[') {
                if (depth == 0) {
                    start = index + 1;
                }
                depth = depth + 1;
            } else if (symbol == ']' && depth > 0) {
                depth = depth - 1;
                if (depth == 0) {
                    return text.substring(start, index);
                }
            }
        }
        return "";
    }
}
System.out.println(TextTools.insideOuterBrackets("Note [red[blue]green] end."));
````

Expected output:

```text
red[blue]green
```

The result keeps the inner brackets around `blue` because both lie inside the selected range. It excludes the outer brackets at indices 5 and 20, along with the surrounding comment. The method selects one continuous range; it does not remove every bracket from the note.

All processing stays inside the helper, while the final println belongs to the caller. A different caller could store the same returned description instead of displaying it.
