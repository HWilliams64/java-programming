## Why This Matters

Programs often need to repeat an action a number of times that depends on their data. A request might need several corrections before it meets a rule. A list might contain one item or many. Repeating the same source statements by hand would set a fixed limit and make later changes harder to maintain.

A loop keeps the repeated action in one place. Its condition decides whether another repetition should happen. This gives a program flexible behavior, but it also creates a responsibility: identify what changes and why repetition eventually stops.

The previous lessons chose actions with conditions or fixed commands. Here, a condition can send execution back to the same body. The next lesson will package a common counting pattern into a for loop. Understanding the test and progress steps now will make that form easier to trace.

## Concept

### Repeat a body while a condition is true

A workshop has three setup checks left. Our first program prints a countdown, one number per completed check, then announces that setup is ready. `remaining` is a whole-number count of checks still to complete.

An **iteration** is one execution of a loop's repeated body. A **while loop** checks a Boolean condition before each possible iteration. The keyword **`while`** introduces the test:

```java
int remaining = 3;
while (remaining > 0) {
```

The first line gives the count a starting value. The next line asks whether any checks remain. Parentheses surround the condition. The opening brace begins the body, the group of statements to repeat when the condition is true. This snippet is part of the complete program below.

```java
    System.out.println(remaining);
    remaining = remaining - 1;
}
```

The first body statement prints the current count. The next subtracts one and assigns the result back to `remaining`, representing one completed check. The closing brace ends the body. Execution then returns to the condition, which uses the updated value. It does not return to the declaration that originally stored three.

The sequence is **test → body → update → test again**. In this example the update is the body's last statement. When the test becomes false, Java skips the body and continues after the closing brace. The final `System.out.println("Ready.");` is outside the loop, so it runs once after repetition ends.

Run the complete countdown and follow each value through the test, print, and assignment.

````java
int remaining = 3;
while (remaining > 0) {
    System.out.println(remaining);
    remaining = remaining - 1;
}
System.out.println("Ready.");
````

Expected output:

```text
3
2
1
Ready.
```

The conditions `3 > 0`, `2 > 0`, and `1 > 0` are true. Each allows one body execution. After printing one, the assignment stores zero. The next check, `0 > 0`, is false, so zero is not printed by the body. There are three iterations and four condition checks. The last check is part of the trace even though it prints no countdown number.

### Explain progress and stopping

**Loop progress** is a change that moves execution toward the stopping condition. Here, subtracting one changes the value used by the next test. **Termination** is the end of the repetition. For the countdown, the false condition at zero causes termination.

A while loop can also run zero times. The next complete program starts with zero remaining checks. Its first condition is false, so the body is skipped immediately. Run it and compare with the three-check countdown.

An **infinite loop** keeps repeating without reaching a stopping path. If the countdown body only printed and never changed `remaining`, a starting value of three would still be three at every check. Printing does not change the variable. Later, you will diagnose a missing update in a displayed faulty program. Keep that faulty version as reading only; repair it before putting it in a runnable cell. If an edited loop repeats unexpectedly, interrupt that execution before correcting and rerunning it.

````java
int remaining = 0;
while (remaining > 0) {
    System.out.println(remaining);
    remaining = remaining - 1;
}
System.out.println("Ready.");
````

Only `Ready.` prints. There are zero body executions and one false condition check. Neither the print inside the body nor its subtraction runs. A while condition can prevent the very first iteration.

### Read another value to make progress

The workshop's kit request must be a positive integer: one or more kits. Zero and negative quantities do not meet that rule. Our supplied input text contains `0`, `-2`, and `3`, in that order. The intended result is two rejection messages followed by acceptance of three kits.

We reuse Scanner setup and numeric reads from Module 1:

```java
import java.util.Scanner;
Scanner quantities = new Scanner("0 -2 3");
int quantity = quantities.nextInt();
```

The import makes the class's short name available. The new reader starts at the supplied text's beginning. The first read returns zero and stores it in `quantity`. That read happens before the loop because the first test needs a value to examine.

```java
while (quantity <= 0) {
    System.out.println("Try a positive quantity.");
    quantity = quantities.nextInt();
}
```

The condition asks whether the stored quantity is invalid under our positive-quantity rule. If it is zero or negative, the body prints the rejection message. The next statement reads the next supplied integer and replaces `quantity`. Reading and assigning another value is this loop's progress step; there is no subtraction.

After the body, the condition checks the newly stored quantity. A positive value makes the condition false, so the program leaves the loop and prints the accepted quantity. The print after the loop does not reject anything; it reports the value that ended repetition.

These examples guarantee that the source contains integers and eventually supplies a positive value. This loop checks the numeric rule; it does not handle words where integers are expected or an input source that ends before a positive value arrives. A read needs another supplied token. Keeping those assumptions explicit lets us focus on the loop's behavior without inventing data that the reader does not have.

````java
import java.util.Scanner;
Scanner quantities = new Scanner("0 -2 3");
int quantity = quantities.nextInt();
while (quantity <= 0) {
    System.out.println("Try a positive quantity.");
    quantity = quantities.nextInt();
}
System.out.println("Accepted: " + quantity);
````

The program prints `Try a positive quantity.` twice, then `Accepted: 3`. The first body execution rejects zero and reads negative two. The second rejects negative two and reads three. The next condition is false, so the accepted print runs without another read.

The entire program recreates the reader each time you run the cell. That makes the input sequence repeatable. A positive first value would take the zero-iteration path instead.

### Move the condition after the first body execution

Sometimes an action must happen once before the program decides whether to repeat it. A **do-while loop** puts the body before its condition. The keyword **`do`** introduces that body, and `while` supplies the ending test.

The two complete programs below deliberately start `remaining` at zero so we can see the difference. Both bodies would print `Check the desk.` and subtract one. The while version tests `remaining > 0` first. Because zero is not greater than zero, it runs no body statements.

In the do-while version, execution enters the body immediately. It prints the desk reminder and stores negative one. Only then does the condition test `-1 > 0`. That false result ends repetition. The negative value here is the result of the demonstration's subtraction; it is not a claim that a workshop has a negative number of physical tasks.

Notice the ending `} while (remaining > 0);`: the semicolon after the condition is required for do-while. A regular while loop has its test before the opening brace and no semicolon between that test and its intended body.

Do-while guarantees a first body execution, not exactly one execution in every program. If its ending test is true, another iteration begins. Use while when an initial false condition should prevent the action altogether.

````java
int remaining = 0;
while (remaining > 0) {
    System.out.println("Check the desk.");
    remaining = remaining - 1;
}
````

````java
int remaining = 0;
do {
    System.out.println("Check the desk.");
    remaining = remaining - 1;
} while (remaining > 0);
````

The while program produces no output and leaves `remaining` at zero. The do-while program prints `Check the desk.` once and leaves `remaining` at negative one. Both conditions are false when evaluated; their different test positions explain the different output.

### Stop when the input carries an end marker

A second workshop task processes a list of packing quantities. The list supplies two kits, then four kits, then `-1` to mean that the list is complete. A **sentinel value** is a value chosen to signal the end of a sequence. Here, `-1` is a marker rather than a quantity to pack.

The program first creates a reader and reads one quantity, as in the validator. Its condition is different:

```java
while (quantity != -1) {
    System.out.println("Pack: " + quantity);
    quantity = quantities.nextInt();
}
```

The `!=` operator asks whether the current quantity differs from the marker. If it does, the body prints the packing instruction and reads the next integer. If the current value is `-1`, the condition is false, so the marker is not processed or printed by the body. Execution continues with `List complete.` after the loop.

Choose a marker that cannot be confused with valid data. In this example valid packing quantities are positive, and the supplied sequence ends with `-1`. The condition tests for that particular marker; it does not independently reject every other nonpositive number.

The meaning of `-1` comes from this program's rule. It is ordinary integer data in Java. In the earlier validator, `-1` is a rejected request and reading continues. In this packing program, it means stop. The selected condition gives the same value its different role.

````java
import java.util.Scanner;
Scanner quantities = new Scanner("2 4 -1");
int quantity = quantities.nextInt();
while (quantity != -1) {
    System.out.println("Pack: " + quantity);
    quantity = quantities.nextInt();
}
System.out.println("List complete.");
````

The output is `Pack: 2`, then `Pack: 4`, then `List complete.`. After the second packing instruction, the read stores `-1`. The next condition is false, so no marker line prints and no read is attempted after the marker.

A list can contain no packing requests. Supply only `"-1"` to represent that case. The first read still happens, but it returns the marker. The first loop condition is then false. Run the complete empty-list example below.

````java
import java.util.Scanner;
Scanner quantities = new Scanner("-1");
int quantity = quantities.nextInt();
while (quantity != -1) {
    System.out.println("Pack: " + quantity);
    quantity = quantities.nextInt();
}
System.out.println("List complete.");
````

## Worked Example

Return to the positive kit-request rule. We want to reject each nonpositive value from `"0 -2 3"` and report the first positive quantity. Build the program in three purpose-based steps.

**1. Supply the data and prepare the first test.** Import Scanner, create `quantities`, and read the first integer into `quantity`. These statements run once at the start. Without the first read, the condition would not have the request value it needs.

**2. Reject and advance.** While `quantity <= 0`, print the rejection message and assign the next read to `quantity`. Each body execution both handles one invalid value and prepares the following test. Keeping the read inside the body prevents the loop from checking the same rejected value forever.

**3. Report the accepted value.** After the loop, the condition has become false. Under our integer-input assumption, `quantity` is therefore positive. Print `Accepted: ` followed by that value. This statement runs once after all rejection iterations.

In a source-file demonstration, the supplied `Main` class and its `main` starting method surround these statements. Their braces provide the program frame. The notebook runs the statements directly, so this task does not require designing a class.

````java
import java.util.Scanner;
Scanner quantities = new Scanner("0 -2 3");
int quantity = quantities.nextInt();
while (quantity <= 0) {
    System.out.println("Try a positive quantity.");
    quantity = quantities.nextInt();
}
System.out.println("Accepted: " + quantity);
````

Expected output:

```text
Try a positive quantity.
Try a positive quantity.
Accepted: 3
```

Trace the stored value and reader together: read zero, reject it, read negative two, reject it, then read three. The check `3 <= 0` is false. The accepted line prints after the loop, and the reader is not asked for another value.

## Summary
