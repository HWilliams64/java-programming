## Learning Goals

- Translate stated rules into comparisons and logical expressions that produce Boolean results.
- Implement an `if`/`else if`/`else` chain and trace its selected branch, including the scope of local variables.
- Test decision boundaries and explain how short-circuit evaluation can avoid an unsafe calculation.


## Why This Matters

Applications must respond differently to different data. An order system may reject a nonpositive quantity, a booking system may show whether a seat is available, and a service desk may route a request to the correct team. Conditions turn those rules into decisions a program can carry out consistently.

If a rule is translated incorrectly, the code can run without an error and still choose the wrong action. Testing the values where a choice changes helps expose these mistakes. Combining conditions also lets you describe requirements involving more than one fact, such as availability and eligibility.

Later in this module, loops will use conditions to decide whether work should repeat. Later course projects will use them to check file records, game moves, and user input. Understanding which conditions are evaluated now will help you explain those larger programs instead of guessing from their output.

## Concept

### Give the program a precise rule

For this fictional campus event, a desk volunteer enters a visitor's age in whole years and whether the visitor has a student ID. The supplied ages are nonnegative integers, and the ID fact is a Boolean value. These are the event's example rules:

- A visitor aged 18 or older with a student ID receives a student admission price of 5 dollars.
- A visitor aged 18 or older without that ID receives a general admission price of 8 dollars.
- A younger visitor receives the instruction `Ask the event desk.`

The program's job is to print exactly one of those instructions. It does not collect a payment or check a real identification document. The age threshold and prices belong to this example's rules.

A **condition** is an expression that produces `true` or `false` for a decision. Start with the age requirement:

```java
age >= 18
```

The name `age` supplies the stored number. The **relational operator** `>=` asks whether that number is greater than or equal to 18. The expression produces a Boolean result, rather than changing the age. This fragment assumes the variable has already been declared.


````java
int age = 17;
System.out.println(age >= 18);
age = 18;
System.out.println(age >= 18);
````

The first comparison prints `false` for age 17. The assignment then changes the age to 18, and the next comparison prints `true`. Each print evaluates the comparison using the value stored at that moment.

Here are the numeric comparisons used in this lesson:

| Comparison | Meaning | Example result |
|---|---|---|
| `<` | Smaller than | `17 < 18` is `true` |
| `<=` | Smaller than or equal to | `18 <= 18` is `true` |
| `>` | Larger than | `18 > 18` is `false` |
| `>=` | Larger than or equal to | `18 >= 18` is `true` |
| `==` | Equal in value | `18 == 18` is `true` |
| `!=` | Different in value | `17 != 18` is `true` |

The first four are relational operators; `==` and `!=` are equality operators. A single `=` assigns a value. Continue using `equals` to compare String contents, as you learned in Module 1. Java requires a Boolean condition here: a number such as `1` is not treated as a true condition.

### Use the result to choose an action

Printing a Boolean helps inspect a rule, but the visitor needs an instruction. A **conditional branch** is a path through code chosen by a condition. The Java keyword **`if`** introduces a condition inside parentheses:

```java
if (age >= 18) {
    System.out.println("Adult admission");
}
```

This fragment uses an existing `age`. If the condition is true, Java executes the print statement. If it is false, Java skips that statement and continues after the closing brace.

A **block** groups the statements between `{` and `}`. The braces identify which instructions belong to this choice; indentation makes that grouping visible to a reader. Keep the braces in these lessons, including when a block contains one statement.

The keyword **`else`** supplies an alternative for a false condition. This complete example uses a younger visitor so that the alternative is visible:


````java
int age = 17;
if (age >= 18) {
    System.out.println("Adult admission");
} else {
    System.out.println("Ask the event desk.");
}
System.out.println("Decision complete.");
````

The age condition is false, so the program skips `Adult admission` and prints `Ask the event desk.`. It then prints `Decision complete.` because that last statement is after the entire choice. Selecting a branch does not end the program.

Do not put a semicolon immediately after `if (age >= 18)`. A semicolon there forms an empty statement, separating the intended block from the decision. The opening brace should follow the condition's closing parenthesis.


### Combine the age and ID requirements

The age test alone cannot distinguish the two admission prices. A **logical operator** combines or reverses Boolean values. The operator `&&` means that both parts must be true:

```java
age >= 18 && hasStudentId
```

The comparison supplies the age result, and the Boolean variable supplies the ID result. There is no need to write `hasStudentId == true`; the variable already holds a Boolean value. With age 19 and an ID, both requirements hold. With age 19 and no ID, the combined condition is false.

The operator `||` means at least one part must be true, including when both are true. The operator `!` reverses one Boolean value: `!hasStudentId` is true when the ID value is false. These symbols are operators, not Java keywords.

| Left value | Right value | Left AND right | Left OR right |
|---|---|---|---|
| false | false | false | false |
| false | true | false | true |
| true | false | false | true |
| true | true | true | true |

This table describes the combined results. Choosing `||` for the student price would change the rule: an adult without an ID or a younger visitor with an ID would meet that looser condition. Use the operator that expresses the requirement, not the one that happens to work for a single test visitor.


For a focused comparison, the next program stores a true age-test result in the Boolean variable `isAdult` and stores false for the ID. It then uses those same facts with each operator.


````java
boolean isAdult = true;
boolean hasStudentId = false;
System.out.println(isAdult && hasStudentId);
System.out.println(isAdult || hasStudentId);
System.out.println(!hasStudentId);
````

The AND result is `false` because the ID requirement fails. The OR result is `true` because the age requirement holds. Reversing the false ID value gives `true`. Parentheses can make a compound rule easier to read, as in `(age >= 18) && hasStudentId`; they do not change that rule's result.

### Put the most specific choice first

The event has three outcomes. An `else if` adds another condition after a previous condition was false. It is the keywords `else` and `if` used together, rather than one separate keyword.

Check the more specific student-price condition first. If it is false, check whether the visitor still meets the age threshold for general admission. If that is also false, use the final `else` for the younger visitor.

In one connected chain, Java executes the first matching branch and skips the remaining branches. An adult with an ID also satisfies the age-only condition, but that later test is not reached after the student branch is selected. By contrast, two separate `if` statements would make two decisions and could print two admission messages. The worked example will assemble the connected chain.

### Check a prerequisite before an unsafe calculation

Combining conditions also affects which expressions Java evaluates. **Short-circuit evaluation** means that `&&` or `||` skips its right side when the left side already determines the result. A false left side settles AND; a true left side settles OR.

For a separate calculation example, suppose a program will divide ten whole supplies by a stored divisor. Integer division by zero cannot produce the requested quotient. Check that prerequisite before evaluating the division:

```java
divisor != 0 && 10 / divisor > 2
```

Java first checks whether `divisor` differs from zero. If that check is false, both requirements cannot be true, so the right side is skipped. If it is true, the division is allowed and its quotient is compared with two. The condition does not repair the divisor or change division rules; it controls whether this particular division is attempted.


````java
int divisor = 0;
System.out.println(divisor != 0 && 10 / divisor > 2);
System.out.println(divisor == 0 || 10 / divisor > 2);
````

The AND expression prints `false` without dividing because its left side is false. The OR expression prints `true` without dividing because its left side is true. These are two different conditions with different results; both demonstrate skipping the right side.

Order matters. Writing the division on the left would attempt it before the prerequisite was checked. Use the left-to-right behavior deliberately, and do not assume that every logical operator in Java skips evaluation in the same way.

### Keep local names where they are available

A branch may need a temporary value. **Scope** is the region of source code in which a declared name can be used. A variable declared inside a branch block is local to that block:

```java
int age = 19;
if (age >= 18) {
    int price = 8;
    System.out.println(price);
}
```

The declaration introduces `price` inside the braces, and the following print statement can use it. A statement after the closing brace cannot use that local name. Whether a particular run enters the branch does not change these source-code boundaries.

If later code needs a result from either branch, introduce the shared variable before the choice and assign it on each path. This separate example does that for a descriptive message:


````java
int age = 17;
String instruction;
if (age >= 18) {
    instruction = "Adult admission";
} else {
    instruction = "Ask the event desk.";
}
System.out.println(instruction);
````

The declaration `String instruction;` introduces a name without assigning its starting text on that line. Both branches assign it before the final print. With age 17, the `else` assignment supplies the displayed instruction. Declaring a name before the choice makes it available afterward; ensuring each path supplies a value makes its later use valid.

### Test the values where a rule changes

A **boundary test** checks a value at or near a decision limit. For `age >= 18`, ages 17 and 18 test the two sides of the threshold. Age 19 alone cannot reveal an accidental `>` because both comparisons accept 19.

The admission rule also depends on the ID value. Test both ID values at those ages, and compare the full message with the stated rule. A program that prints something is not necessarily choosing the correct message.


## Worked Example

### Step 1: store the visitor's details

```java
int age = 19;
boolean hasStudentId = true;
```

These lines supply the repeatable visitor: 19 years old with a student ID. The age is a whole number and the ID fact is Boolean. The task asks for one admission message, so the following steps form one connected choice.

### Step 2: give the student price its complete condition

```java
if (age >= 18 && hasStudentId) {
    System.out.println("Student admission: 5");
}
```

The condition requires both the age threshold and the ID. The displayed `5` is the example's student price in dollars. This fragment is the first part of the larger chain, rather than the complete program.

### Step 3: handle the remaining visitors

```java
else if (age >= 18) {
    System.out.println("General admission: 8");
} else {
    System.out.println("Ask the event desk.");
}
```

This fragment attaches to the first branch. Reaching its age test means the student condition was already false. An adult on that path lacks the qualifying ID and receives the general price. The final branch handles visitors who fail the age threshold. There is no condition after the final `else` because it covers the remaining case.

### Step 4: run the complete decision

Together, the inputs and connected branches form this program:


````java
int age = 19;
boolean hasStudentId = true;
if (age >= 18 && hasStudentId) {
    System.out.println("Student admission: 5");
} else if (age >= 18) {
    System.out.println("General admission: 8");
} else {
    System.out.println("Ask the event desk.");
}
````

The program prints:

```text
Student admission: 5
```

Age 19 meets the threshold and `hasStudentId` is true, so the first branch runs. The later branches are skipped even though the visitor also meets the age-only comparison. Exactly one instruction appears, matching the desk's rule for this visitor.

## Summary

A condition produces a Boolean value. Comparisons test values, and logical operators combine or reverse Boolean results. One connected branch chain selects its first matching path. Scope determines where a name may be used, while boundary tests check the inputs where a decision changes.

Short-circuit evaluation makes the order of `&&` and `||` expressions meaningful. A left result that settles the operator prevents the right expression from being evaluated.

- [Java 21 if statements](https://docs.oracle.com/javase/specs/jls/se21/html/jls-14.html#jls-14.9) — Language rules for `if`, `else`, and Boolean conditions.
- [Java 21 conditional AND](https://docs.oracle.com/javase/specs/jls/se21/html/jls-15.html#jls-15.23) — Left-to-right evaluation and the condition for skipping the right operand of `&&`.
- [Java 21 conditional OR](https://docs.oracle.com/javase/specs/jls/se21/html/jls-15.html#jls-15.24) — When `||` can determine its result without evaluating its right operand.
- [Java 21 declaration scope](https://docs.oracle.com/javase/specs/jls/se21/html/jls-6.html#jls-6.3) — The source-code regions where declared names can be used, including block-local variables.
