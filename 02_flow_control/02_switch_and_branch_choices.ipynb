# Switch and Branch Choices

- Build and test a related command selector, including unknown input.

## Why This Matters

Programs often choose among a short list of named actions: a menu command, a delivery option, or a device mode. Keeping the alternatives together makes the supported choices easier to find and update. A clear fallback also tells users what happens when their input is outside that list.

The previous lesson used conditions to describe rules such as an age range. A switch is useful when one value should match one of several fixed choices. Its structure makes those choices visible, but you must also control where each chosen action ends. Otherwise, a correct first instruction can be followed by an incorrect second one.

Later lessons repeat work with loops. Tracing where control goes next in a switch prepares you to distinguish leaving a construct from ending the whole program.

## Concept

### Match one command to a fixed choice

The kiosk has two supported commands. `"borrow"` directs a visitor to the checkout desk. `"renew"` directs the visitor to a list of loans. Any other command gets a reminder of the supported choices. These are supplied text values, not a connection to a real library account.

A **switch statement** selects where to begin executing statements from a set of labeled choices. The keyword **`switch`** introduces that selection. The **selector** is the value being matched. In the following snippet, `command` is the selector:

```java
String command = "renew";
switch (command) {
```

The first line stores the supplied command. The next line places that variable inside parentheses after `switch`. The opening brace starts the switch body; a matching closing brace ends it. This snippet starts a larger statement, so use the complete demonstration below when running it.

The keyword **`case`** introduces a fixed value to match. The quoted text and colon form a **case label**, such as `case "renew":`. The colon marks the start of that choice's statements. For this String selector, matching compares text contents. It does not require a separate call to `equals` in each label.

```java
    case "renew":
        System.out.println("Open your loan list.");
        break;
```

When the selector matches `"renew"`, execution begins after that label. The print statement supplies the visitor's instruction. The keyword **`break`** ends this switch and continues at the next statement after its closing brace. It does not end the program. Here, each supported command needs one service message, so each case has its own `break;`.

The keyword **`default`** introduces the fallback starting point when no case label matches:

```java
    default:
        System.out.println("Choose borrow or renew.");
```

In our examples, this group is last. After printing, execution reaches the closing brace, so no final `break` is needed. Default is a fallback starting point, not an instruction that always runs after another case.

String matching is case-sensitive and includes spaces. `"Renew"` and `" renew "` do not match `"renew"`. From Module 1, `trim()` returns a String without outside spaces; you must store its returned result before matching. It does not change letter case or turn an unsupported command into a supported one. For example, storing `"  borrow  ".trim()` gives `"borrow"`, while trimming `"  search  "` still gives an unknown command.

The braces enclose one switch block. Case labels mark entry points; they do not create separate blocks or separate local scopes. As you learned with if, braces define block boundaries. These examples declare command before the switch and introduce no case-local variables.

### Compare with a familiar condition chain

The next complete program expresses the same choices using if and equals from earlier lessons. The first condition compares command with borrow. If that is false, the second compares it with renew. The else branch supplies the fallback. This chain chooses one branch automatically; it does not need break after each print.

A switch makes a fixed list of choices for one selector visible. An if chain remains useful for conditions such as age >= 18 or combinations of different facts. Choose the form that states the rule clearly; the two forms below can produce the same service instruction.

````java
String command = "renew";
if (command.equals("borrow")) {
    System.out.println("Visit the checkout desk.");
} else if (command.equals("renew")) {
    System.out.println("Open your loan list.");
} else {
    System.out.println("Choose borrow or renew.");
}
````

The if chain prints `Open your loan list.`. The borrow comparison is false, the renew comparison is true, and the fallback is skipped.

Now compare the following two switch programs. The first uses the known command renew. The second uses search to show the fallback. Each also prints `Kiosk ready.` after the switch, so you can see where execution continues.

````java
String command = "renew";
switch (command) {
    case "borrow":
        System.out.println("Visit the checkout desk.");
        break;
    case "renew":
        System.out.println("Open your loan list.");
        break;
    default:
        System.out.println("Choose borrow or renew.");
}
System.out.println("Kiosk ready.");
````

````java
String command = "search";
switch (command) {
    case "borrow":
        System.out.println("Visit the checkout desk.");
        break;
    case "renew":
        System.out.println("Open your loan list.");
        break;
    default:
        System.out.println("Choose borrow or renew.");
}
System.out.println("Kiosk ready.");
````

The known `"renew"` command prints `Open your loan list.` and then `Kiosk ready.`. Its `break` skips the later default group. The unknown `"search"` command prints `Choose borrow or renew.` and then `Kiosk ready.`.

Each program produces one service message and one separate ready message. The selector chooses a starting label. The selected statements and their exit determine what runs after that point.

### Stop before an unrelated instruction

The colon-style switch used here does not automatically stop at the next label. If execution reaches another label without encountering `break`, it continues into that group's statements. This is **fall-through**: continuing from one statement group into a later group. Java does not recheck the later label before executing those statements.

In the first program below, the `"borrow"` case is deliberately missing its `break`. This is valid Java with incorrect behavior for our one-instruction requirement. The selector matches `"borrow"`, so the checkout message prints. Execution then reaches the renew group's print statement and prints that message too. The `break` in the renew group finally exits the switch.

The second program restores the first `break;`. Now borrowing prints only the checkout instruction. Compare the complete programs and their line counts. A correct first line is not enough evidence that a branch behaves correctly.

````java
String command = "borrow";
switch (command) {
    case "borrow":
        System.out.println("Visit the checkout desk.");
    case "renew":
        System.out.println("Open your loan list.");
        break;
    default:
        System.out.println("Choose borrow or renew.");
}
````

````java
String command = "borrow";
switch (command) {
    case "borrow":
        System.out.println("Visit the checkout desk.");
        break;
    case "renew":
        System.out.println("Open your loan list.");
        break;
    default:
        System.out.println("Choose borrow or renew.");
}
````

The missing-break version prints two lines: `Visit the checkout desk.` followed by `Open your loan list.`. The repaired version prints only `Visit the checkout desk.`.

A case label identifies a possible entry point; it is not a wall that stops execution arriving from above. Test commands that enter earlier groups as well as later groups, because a later command alone may hide a missing earlier break.

## Worked Example

The complete kiosk program combines three steps. It uses the same two commands and service messages introduced above, with one message as its final result.

**1. Store the requested command.** `String command = "renew";` supplies the text to match. There is no typed input in this program; changing the literal creates a repeatable test case.

**2. Match a label and print its instruction.** `switch (command)` reads that stored value. The `"borrow"` label leads to the checkout instruction. The `"renew"` label leads to the loan-list instruction. If neither matches, `default` leads to the supported-command reminder.

**3. End the selected action.** Each named case has `break;` after its print statement. That placement prevents one supported command from printing another command's message. The last default group ends at the closing brace.

In a Java source-file demonstration, a supplied `Main` class contains a `main` method, the starting method that Java runs. The class and method braces provide the surrounding program frame. The notebook cell below runs the statements inside that frame directly; you do not need to design a class to practice this switch.

````java
String command = "renew";
switch (command) {
    case "borrow":
        System.out.println("Visit the checkout desk.");
        break;
    case "renew":
        System.out.println("Open your loan list.");
        break;
    default:
        System.out.println("Choose borrow or renew.");
}
````

Expected output:

```text
Open your loan list.
```

The stored command matches `case "renew":`. Java starts at that label, prints the loan-list instruction, and reaches `break;`. The switch ends. The earlier borrow group never runs, and the later default group is skipped.
