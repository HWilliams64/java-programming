## Why This Matters

Many programs repeat work a known number of times. A school might print one certificate per student, while a scheduling system might inspect every room and time slot. A counted loop keeps the rule for repetition in one place. It avoids maintaining a separate copy of the same statement for every item.

A report also needs a reliable record of completed work. A running total can count printed records or add the quantities processed. When work involves two dimensions, such as rooms and time slots, nested loops visit the combinations systematically. Checking only the final count could miss a duplicated or omitted combination, so the individual results matter too.

These mechanisms prepare you to process arrays and collections later in the course. In the next module, you will organize related steps into named methods. Clear repetition and meaningful totals make those larger programs easier to understand and test.

## Concept

### Group the steps of counted repetition

Our fictional workshop labels kits with whole-number identifiers starting at 1. For now, one group has three kits, so the organizer needs `Kit 1`, `Kit 2`, and `Kit 3`, in that order. Each output line represents one label; no physical printer is needed for the example.

A **for loop** groups a starting step, a condition, and an update in its header. The Java keyword **`for`** introduces that header:

```java
for (int kit = 1; kit <= 3; kit = kit + 1)
```

The first part declares `kit` and initializes it to 1. This initialization happens once when execution enters the loop. The second part tests whether the current identifier is at most 3. The third part uses a full assignment to increase the identifier by 1. The two semicolons separate these three parts; they do not end the loop.

Java checks the condition before each body execution. When it is true, Java runs the body, performs the update, and then checks the condition again. Although the update appears above the body on the page, it runs after the body. When the condition is false, Java leaves the loop without another body execution or update.

Put the body inside braces directly after the header. Do not add a semicolon between the header and opening brace; that would create an empty loop body.


````java
for (int kit = 1; kit <= 3; kit = kit + 1) {
    System.out.println("Kit " + kit);
}
````

The loop prints `Kit 1`, `Kit 2`, and `Kit 3` on separate lines. After printing kit 3, the update makes `kit` equal to 4. The comparison `4 <= 3` is false, so there is no `Kit 4` output. The inclusive condition allows the last requested number itself.

The variable declared in this header is local to the loop: its scope includes the condition, body, and update. A statement after the loop cannot use that `kit` declaration. If a value must remain available afterward, declare it in an enclosing block, as we will do with the total.

### Count completed work with a running total

The kit number identifies the current item. It is not necessarily the number of labels printed: some labels may later be skipped. The organizer therefore needs a separate total.

An **accumulator** is a variable that preserves and combines results across iterations. Here, `labelsPrinted` counts completed label lines. It starts at 0 before any label is printed:

```java
int labelsPrinted = 0;
```

After each label, update the same variable:

```java
labelsPrinted = labelsPrinted + 1;
```

The right side reads the previous total and adds 1. The assignment stores that new total. Placing the initialization before the loop preserves earlier work; placing the addition beside the label print ties the count to the event being counted. The final summary belongs after the loop so that it appears once.


````java
int labelsPrinted = 0;
for (int kit = 1; kit <= 3; kit = kit + 1) {
    System.out.println("Kit " + kit);
    labelsPrinted = labelsPrinted + 1;
}
System.out.println("Labels: " + labelsPrinted);
````

The three kit lines raise `labelsPrinted` from 0 to 1, then 2, then 3. The last statement reads that stored total and prints `Labels: 3`. The summary line is not itself a kit label, so it does not add to the total.

Declaring `labelsPrinted` before the loop also keeps its name available for the final print. A total can combine values other than 1, but the update must match what the program is measuring. Here the unit is completed labels, not kit identifier values.

### Repeat a complete inner task for every group

Now the workshop has two groups, and each group needs two kits. Kit numbers start at 1 within each group, so the label must include both identifiers. The desired pairs are group 1 with kits 1 and 2, followed by group 2 with kits 1 and 2.

A **nested loop** is a loop inside another loop. The **outer loop** chooses the group. Its body contains an **inner loop** that visits all kits for that group:

```java
for (int group = 1; group <= 2; group = group + 1) {
    for (int kit = 1; kit <= 2; kit = kit + 1) {
        System.out.println("Group " + group + ", kit " + kit);
    }
}
```

The outer header initializes `group` once. The inner header initializes `kit` each time execution enters that inner loop. While group 1 is selected, the inner loop prints both of its kits before the outer update can run. The next outer iteration enters the inner header again, creating a new local `kit` initialized to 1 for group 2.

The print statement uses the current group and current kit to form one label. Both values are available inside the inner body. To count all the labels, keep `labelsPrinted` outside both loops and add 1 immediately after this print. The same total must survive the move to a new group.


````java
int labelsPrinted = 0;
for (int group = 1; group <= 2; group = group + 1) {
    for (int kit = 1; kit <= 2; kit = kit + 1) {
        System.out.println("Group " + group + ", kit " + kit);
        labelsPrinted = labelsPrinted + 1;
    }
}
System.out.println("Labels: " + labelsPrinted);
````

The output lists `Group 1, kit 1`, `Group 1, kit 2`, `Group 2, kit 1`, and `Group 2, kit 2`, followed by `Labels: 4`. At the start of group 2, the inner kit number restarts at 1 while the total remains 2. The two new labels then raise it to 4.

Two groups times two kits per group gives four expected labels. That multiplication checks the count, but inspect the pairs and their order as well. Four lines containing a duplicate pair would still be an incorrect report. The total print is outside both loop bodies, so it appears only after every group is finished.

### Skip selected work without stopping the whole loop

Consider a separate three-kit labeling pass. Kit 2 already has a usable label, so the organizer wants new labels only for kits 1 and 3. The loop still visits the kit numbers in order, but it should skip both the print and the count for kit 2.

The Java keyword **`continue`** ends the current iteration's remaining body work. In the `for` form used here, execution then reaches the header's update and tests the condition again:

```java
if (kit == 2) {
    continue;
}
```

This fragment belongs inside the loop, before the label print and accumulator update. If the kit number is 2, neither of those later statements runs. For other kit numbers, Java skips the `if` body and reaches both statements normally. The complete version uses `keptLabels` to count only the labels actually printed.


````java
int keptLabels = 0;
for (int kit = 1; kit <= 3; kit = kit + 1) {
    if (kit == 2) {
        continue;
    }
    System.out.println("Kit " + kit);
    keptLabels = keptLabels + 1;
}
System.out.println("Kept: " + keptLabels);
````

The result is `Kit 1`, `Kit 3`, and `Kept: 2` on separate lines. At kit 2, `continue` skips the print and addition, but `kit = kit + 1` still advances to 3. After kit 3, the update makes 4 and the condition becomes false.

A **continue statement** skips the rest of the current loop body; it does not mean stop all repetition. The `continue;` form used here applies to the innermost enclosing loop. By contrast, the `break` statement you used with `switch` leaves its enclosing switch; when used directly in a loop, it leaves that loop.

The update location matters. A `continue` in a `while` loop returns to the condition, so an update later in that body could be skipped. Do not move this example's control flow into a different loop form without tracing how it still makes progress.

## Worked Example

### Step 1: start one total for the complete report

Return to the two-group workshop report, where every kit needs a label. There is no skip rule in this report. Begin with `int labelsPrinted = 0;` before both loops, because the summary must include both groups.

### Step 2: select a group, then visit its kits

The outer header is `for (int group = 1; group <= 2; group = group + 1)`. Inside its body, the inner header is `for (int kit = 1; kit <= 2; kit = kit + 1)`. Both conditions include their upper bound. Each update advances its own identifier by 1.

The inner loop completes before control can reach the outer update. When a new group begins, its inner initialization starts kit numbering at 1 again. This order gives every group the full set of kits.

### Step 3: print and count one pair at a time

Inside the inner body, `System.out.println("Group " + group + ", kit " + kit);` combines the two current identifiers into one label. The next statement, `labelsPrinted = labelsPrinted + 1;`, counts that completed label. After both closing braces, print the summary from the accumulated value.

### Step 4: run the assembled report

The complete program includes its own starting total. You can rerun it to begin a new report:


````java
int labelsPrinted = 0;
for (int group = 1; group <= 2; group = group + 1) {
    for (int kit = 1; kit <= 2; kit = kit + 1) {
        System.out.println("Group " + group + ", kit " + kit);
        labelsPrinted = labelsPrinted + 1;
    }
}
System.out.println("Labels: " + labelsPrinted);
````

Group 1 produces two labels and leaves the total at 2. Its inner loop stops when kit becomes 3. The outer update then selects group 2, whose inner loop starts at 1 and contributes two more labels. Finally, the inner condition is false at kit 3 and the outer condition is false at group 3. Only then does the program print `Labels: 4`.

This trace explains both the visible pair order and the final count. Each code cell recreates the values needed for its own demonstration; the report does not depend on the kit value left by an earlier example.
