## Concept

### Use the Desktop for a native window

A **native JavaFX window** is a window managed by the operating-system desktop. JavaFX calls this top-level window a `Stage`. Its title bar, close control, and ability to receive focus belong to the desktop window system.

The notebook and the Desktop show different parts of the same program. A notebook code cell submits Java instructions. A JavaFX window displays the resulting interface on the Workspace Desktop. `System.out.println` still sends text to the notebook output. Printing a title does not prove that a visible window has appeared; inspect the Desktop too.

If no Desktop is available, open the Workspace with its Desktop option enabled before doing this lesson. Do not replace this check with an assertion that console output proves the interface works.

### Connect the window to a scene

A **scene** is the container that connects a window to its visible content. A Stage holds a Scene. The Scene, in turn, refers to a root node: the starting point of its content.

A **scene graph** is a parent-child tree of visual objects called nodes. In this lesson, a layout node is the parent of a text control. The Stage is the window; it is not a node inside its own scene graph.

| Object | Responsibility | Connection |
|---|---|---|
| Stage | Provide the top-level window and title. | Holds the Scene. |
| Scene | Connect the window to its content. | Refers to the root node. |
| StackPane | Position its child within the available area. | Is the root and contains a Label. |
| Label | Present the message. | Is a child of the StackPane. |

### Present text and let a layout place it

A **Label control** presents text that a user does not edit directly. `new Label("Open")` creates a label with the text Open. `getText()` reads that text, and `setText("Closed")` replaces it.

A **layout container** is a parent node that places its children. A `StackPane` places this lesson’s single Label in the center. The name describes a container that can stack children; here we use just one child so there is no overlap to manage.

`new Scene(root, 420, 220)` requests a scene content area 420 pixels wide and 220 pixels high. A pixel is a small unit of screen display. The title bar and other window decorations can add to the outer window dimensions. Do not compare a decorated window’s total height with the scene height as though they must be identical.

### Start the toolkit once for this kernel

The **JavaFX runtime lifecycle** describes starting the toolkit, using its windows, and ending that runtime. A notebook kernel already has a running Java process. It needs to start the JavaFX toolkit inside that process once, then use it for later cells.

The supplied setup defines `Fx.start()`, `Fx.run(...)`, and `Fx.closeWindows()`. These are course support methods, not built-in JavaFX methods. Run the complete setup as provided. It prints `FX ready` when the toolkit is available.

| Support call | Meaning in this notebook |
|---|---|
| `Fx.start()` | Start JavaFX if needed and keep the toolkit available after the last window closes. |
| `Fx.run(() -> { ... })` | Run a short block of interface work on JavaFX’s application thread and wait for that block to finish. |
| `Fx.closeWindows()` | Close the tutorial’s currently visible JavaFX windows before building a new example. |

The setup deliberately keeps JavaFX available when the last window closes. This lets you close a window and rerun the complete example. Do not call `Platform.exit()` in this lesson: after JavaFX shuts down, the same toolkit cannot simply be started again in that Java process. Restart the notebook kernel if you need a new runtime.

### Give interface work to the application thread

An **application thread** is the path of execution that JavaFX uses for its live interface. **Application-thread ownership** means that live window and control work belongs on that thread. A notebook’s normal execution does not automatically happen there.

Put window construction and short changes inside `Fx.run`. The lambda describes the work; the supplied support arranges where it runs. Later, JavaFX event handlers will also run on that application thread. Do not put long loops, file reads, or waiting for user input in this block, because the interface needs that same thread to respond.

The supplied setup coordinates work between the notebook and JavaFX. Module 13 studies how multiple threads work together. You do not need to implement this setup machinery now. Use the three documented support operations and explain what each accomplishes.

### Build the parts in order

After the supplied setup, this complete example opens a small window:

```java
import javafx.stage.Stage;
import javafx.scene.Scene;
import javafx.scene.control.Label;
import javafx.scene.layout.StackPane;
Fx.closeWindows();
Fx.run(() -> {
    Stage stage = new Stage();
    Label message = new Label("Open");
    StackPane root = new StackPane(message);
    Scene scene = new Scene(root, 300, 180);
    stage.setTitle("Desk Status");
    stage.setScene(scene);
    stage.show();
    System.out.println("Message: " + message.getText());
});
```

The notebook prints `Message: Open`. The Desktop shows a window titled Desk Status with Open in its content area. The code first creates the objects, then connects them, and finally calls `show()` to make the window visible. The title and the Label text are separate values on separate objects.

### Prepare this kernel

This is supplied course support for running JavaFX inside IJava. Run it once after starting or restarting this notebook's Java kernel. The message `FX ready` means the support has initialized JavaFX and completed an operation on its application thread. Open the Workspace **Desktop** view to see the windows created by later cells.

`Fx.run(() -> { ... })` performs the enclosed UI work on the JavaFX Application Thread and waits for that short operation to finish. Use it for reading as well as changing a live window or its controls. `Fx.closeWindows()` hides the windows created by this kernel before another example opens its own. `Fx.start()` is safe to call again; it keeps JavaFX available after the last window closes.

The implementation below is provided runtime support. You do not need to write its thread-coordination machinery for this lesson. A thread is one sequence of execution; Module 13 studies how to coordinate more than one. Here your responsibility is to use the documented support operations and keep UI work short. The support uses a completion signal, a time limit, and an error holder so a later cell does not silently continue after unfinished or failed UI work.

Do not call `Platform.exit()` during notebook practice. That ends the toolkit for this kernel; restart the kernel and rerun setup if you do so. Closing a window is different from ending the toolkit. If setup reports a display error, check that the Workspace Desktop is running, then restart the kernel and rerun setup. A JavaFX window appears in the Desktop, not as an inline notebook control.


````java
import javafx.application.Platform;
import javafx.stage.Window;
import java.util.ArrayList;
import java.util.concurrent.CountDownLatch;
import java.util.concurrent.TimeUnit;
import java.util.concurrent.atomic.AtomicReference;
class Fx {
    static void run(Runnable action) throws InterruptedException {
        if (Platform.isFxApplicationThread()) {
            action.run();
            return;
        }
        CountDownLatch done = new CountDownLatch(1);
        AtomicReference<Throwable> failure = new AtomicReference<Throwable>();
        Platform.runLater(() -> {
            try { action.run(); }
            catch (Throwable error) { failure.set(error); }
            finally { done.countDown(); }
        });
        if (!done.await(10, TimeUnit.SECONDS)) {
            throw new IllegalStateException("FX operation timed out; restart the kernel.");
        }
        if (failure.get() != null) { throw new RuntimeException(failure.get()); }
    }
    static void start() throws InterruptedException {
        try { Platform.startup(() -> Platform.setImplicitExit(false)); }
        catch (IllegalStateException alreadyStarted) {
            // This call is also safe when this kernel already started JavaFX.
        }
        run(() -> Platform.setImplicitExit(false));
    }
    static void closeWindows() throws InterruptedException {
        run(() -> {
            for (Window window : new ArrayList<Window>(Window.getWindows())) {
                window.hide();
            }
        });
    }
}
Fx.start();
System.out.println("FX ready");
````

## Worked Example

**Subgoal 1: create the content.** Construct a Label and put it in a StackPane.

**Subgoal 2: connect the scene and window.** Construct the Scene, set the title, and attach the Scene to the Stage.

**Subgoal 3: show and verify.** Show the window, compare its title/message with the report, close it on the Desktop, and rerun the complete example.


````java
import javafx.stage.Stage;
import javafx.scene.Scene;
import javafx.scene.control.Label;
import javafx.scene.layout.StackPane;
Fx.closeWindows();
Fx.run(() -> {
    Stage stage = new Stage();
    Label message = new Label("Welcome to CSC-239");
    StackPane root = new StackPane(message);
    Scene scene = new Scene(root, 420, 220);
    stage.setTitle("Course Welcome");
    stage.setScene(scene);
    stage.show();
    System.out.println("Window: " + stage.getTitle());
    System.out.println("Message: " + message.getText());
});
````
