## Why This Matters

A console report helps a programmer inspect values. A person using a graphical application also needs a visible window whose content is connected and arranged correctly. Keeping those responsibilities separate makes it easier to find a missing display step, change a message without changing the title, and test what a user can actually see.

Earlier lessons connected objects through references and passed behavior with lambdas. Here those skills connect a window to its content and give a short interface-building action to supplied course support. These connections will support later forms and buttons. Without a clear model of ownership and window lifetime, repeated notebook runs can leave confusing old windows or reports that seem correct while the expected interface is absent.

## Concept

### Give the program a visible window

A course welcome display should show a short message inside a window that a student can inspect and close. We will set its title to `Course Welcome` and its message to `Welcome to CSC-239`. The title identifies the window; the message is the content inside it. Success means seeing both in the Workspace desktop view and being able to close and recreate the window.

A **native JavaFX window** is an operating-system window created by JavaFX. It appears on the desktop rather than inside a notebook output cell. A **Stage** represents that window and provides operations for its title and visible lifetime.

```java
    Stage stage = new Stage();
```

This statement constructs the window object. It is a fragment of the complete interface-building action shown later. Construction gives us an object to configure; it does not yet show a window. After we attach its content, the complete example calls `stage.show()` to make it visible.

Closing this window does not end the notebook kernel. The initialized toolkit can remain available for another window. We will use the native window's close button during the lesson, then run the complete construction again to create another window.


### Attach content through a Scene

A **Scene** connects the window to the content JavaFX should display. In the complete example, a layout object named `root` contains our message. These statements create a scene using that root and attach it to the stage:

```java
    Scene scene = new Scene(root, 420, 220);
```

```java
    stage.setScene(scene);
```

These fragments assume `root` and `stage` have been constructed. The first constructor argument identifies the content root. The next two arguments request a content area 420 units wide and 220 units high. These **scene content dimensions** describe the content area, not the outer window frame and title bar. On this Workspace display they correspond to the area in which our layout places the label.

The method `setScene` assigns that scene to the window. Constructing a scene without attaching it would leave the stage without that content. The connection gives each object a clear role: the stage manages the native window, and the scene holds the content structure displayed within it.


### Follow the parent-child content structure

The **scene graph** is a tree of objects that make up the scene's visual content. A **Node** is an object that participates in that tree, such as a label or layout container. A parent node can contain child nodes.

Our content tree has two levels: a `StackPane` is the **root node**, and a `Label` is its child. Root means that this node is the starting point for this scene's content tree. It does not mean that the variable must be named `root`; that is our descriptive choice.

Keep the attachment relationship separate from the content tree. The Stage has a Scene, and the Scene has a root node, but Stage and Scene are not themselves child nodes in this tree. This distinction will help when later lessons place several controls inside nested layouts.


### Put the message in a Label

A **Label** displays text that the user reads rather than edits directly. The example constructs its message with:

```java
    Label message = new Label("Welcome to CSC-239");
```

The string argument supplies the initial text. The variable `message` then refers to that Label object. Later, `message.getText()` reads the label's stored text for the console report. It does not ask the student to type anything.

The title uses a separate operation on the stage:

```java
    stage.setTitle("Course Welcome");
```

Changing the label text does not change the stage title, and changing the stage title does not change the label. These values belong to different objects and appear in different places. That separation is useful when an application keeps a stable window title while updating a message inside the window.


### Let a layout container place its child

A **layout container** is a parent node that arranges its child nodes in the available space. A **StackPane** places children in a shared area; with our single child and its default alignment, it centers the message.

```java
    StackPane root = new StackPane(message);
```

This statement assumes the Label named `message` already exists. The constructor receives that Label and makes it a child of the new container. The Scene then receives this container as its root. The result is a connected structure: window, scene, layout root, and label.

We do not calculate the label's x and y positions ourselves. The layout uses the available area to place it. This gives the example useful behavior when the window size changes and prepares us for arranging more than one control later.

Creating the label and layout still does not display the window. The complete example configures all the objects, attaches the scene, and then calls `stage.show()`. The following diagram traces these connections so you can relate the visible message to the objects that produce it.

### Send interface work to its owning thread

A **thread** is a path of execution that carries out program instructions. The **JavaFX Application Thread** is the thread that handles JavaFX interface work. Reads and changes involving live controls need to happen on that thread. The notebook cell itself should use the supplied bridge instead of assuming it already runs there.

The helper's **`Fx.run`** method receives a **Runnable**, an interface representing an action whose `run` method takes no arguments and returns no result. The empty parentheses in `() -> { ... }` mean this action takes no input arguments. The braces hold statements to run when the action is invoked; this Runnable action does not return a value. Our lambda supplies that action, using the lambda syntax learned earlier in the course. Our complete example places the window construction and inspection inside that action. The helper arranges for the action to execute on the JavaFX Application Thread and waits for the short action to complete before returning to the notebook.

That boundary includes the console report's getters: `stage.getTitle()` and `message.getText()` execute inside the same interface action as construction and showing. Printing their returned strings lets us inspect configured values, but it does not prove that the expected native window was visible. We check the desktop separately.

Keep actions passed to this helper short. A long wait inside interface work would prevent the interface thread from handling other events promptly. Later lessons teach button callbacks on this same thread; background work belongs to the multithreading module.

We now have the pieces for the full program: initialize the supplied support, close an earlier window, construct connected interface objects through `Fx.run`, show the stage, and inspect the actual desktop result.


### Prepare the supplied notebook support

This is supplied course support for running JavaFX inside IJava. Run it once after starting or restarting this notebook's Java kernel. The message `FX ready` means the support has initialized JavaFX and completed an operation on its application thread. Open the Workspace **Desktop** view to see the windows created by later cells.

Do not call `Platform.exit()` during notebook practice. That ends the toolkit for this kernel; restart the kernel and rerun setup if you do so. Closing a window is different from ending the toolkit. ### Follow a short interface action

<details class="animation-panel">
<summary>Show or hide the animation</summary>
<p><img src="media/01_javafx_windows_in_the_workspace/supplied_ui_dispatch.gif" alt="Use the supplied bridge to run short live-interface work on the application thread." width="960" style="max-width:100%;height:auto;"></p>
</details>

In this depicted notebook call, Fx.run receives the lambda outside the JavaFX Application Thread. The supplied support sends the action to that thread, including its live title and text reads. After successful completion the notebook call returns. Later Desktop interaction is a separate event.

The loop lasts 10 seconds. [View the still version of this diagram](media/01_javafx_windows_in_the_workspace/supplied_ui_dispatch_still.png).


````java
import javafx.application.Platform;
import javafx.stage.Window;
import java.util.ArrayList;
import java.util.concurrent.CountDownLatch;
import java.util.concurrent.TimeUnit;
import java.util.concurrent.atomic.AtomicReference;
class Fx {
    static void run(Runnable action) throws InterruptedException {
        if (Platform.isFxApplicationThread()) {
            action.run();
            return;
        }
        CountDownLatch done = new CountDownLatch(1);
        AtomicReference<Throwable> failure = new AtomicReference<Throwable>();
        Platform.runLater(() -> {
            try { action.run(); }
            catch (Throwable error) { failure.set(error); }
            finally { done.countDown(); }
        });
        if (!done.await(10, TimeUnit.SECONDS)) {
            throw new IllegalStateException("FX operation timed out; restart the kernel.");
        }
        if (failure.get() != null) { throw new RuntimeException(failure.get()); }
    }
    static void start() throws InterruptedException {
        try { Platform.startup(() -> Platform.setImplicitExit(false)); }
        catch (IllegalStateException alreadyStarted) {
            // This call is also safe when this kernel already started JavaFX.
        }
        run(() -> Platform.setImplicitExit(false));
    }
    static void closeWindows() throws InterruptedException {
        run(() -> {
            for (Window window : new ArrayList<Window>(Window.getWindows())) {
                window.hide();
            }
        });
    }
}
Fx.start();
System.out.println("FX ready");
````

### Assemble a small complete window

A desk status display uses the same four objects with a shorter message. The following complete program uses the supplied support you have just run. Its first four imports make the JavaFX class names available. The `javafx.stage`, `javafx.scene`, `javafx.scene.control`, and `javafx.scene.layout` package names group the window, scene, control, and layout classes.

Copy this complete example into a Java work cell if you want to try it. Keep its `Fx.closeWindows()` and `Fx.run(...)` calls: they prepare a clean window state and give interface work to its owning thread.

```java
import javafx.stage.Stage;
import javafx.scene.Scene;
import javafx.scene.control.Label;
import javafx.scene.layout.StackPane;
Fx.closeWindows();
Fx.run(() -> {
    Stage stage = new Stage();
    Label message = new Label("Open");
    StackPane root = new StackPane(message);
    Scene scene = new Scene(root, 300, 180);
    stage.setTitle("Desk Status");
    stage.setScene(scene);
    stage.show();
    System.out.println("Message: " + message.getText());
});
```

The notebook prints `Message: Open`. On the Desktop, Desk Status appears in the title bar and Open appears centered inside the content area. The scene requests 300 by 180 units of content. The imports select classes; construction creates the objects; `setScene` connects the scene to its stage; `show` makes that stage visible. Close the Desk Status window when you finish inspecting it.

## Worked Example

### Welcome students to the course

We can now use the same structure for the course welcome display. Its title is Course Welcome, its message is Welcome to CSC-239, and its scene requests a 420 by 220 content area. There is one Label, centered by one StackPane. No text entry or button action is needed.

**Create the content.** After constructing the Stage, the action creates the Label with its message and passes that Label to the StackPane constructor. Here we give the layout its child before using the layout as the scene root.

**Connect the scene and window.** The Scene receives the root and content dimensions. The Stage receives its title and then the Scene. These calls configure separate objects while connecting them into one display.

**Show and inspect.** `stage.show()` makes the configured window visible. The last two statements read the title and message within the same `Fx.run` action and print them to the notebook. Run the complete cell, then inspect the actual Desktop window.


````java
import javafx.stage.Stage;
import javafx.scene.Scene;
import javafx.scene.control.Label;
import javafx.scene.layout.StackPane;
Fx.closeWindows();
Fx.run(() -> {
    Stage stage = new Stage();
    Label message = new Label("Welcome to CSC-239");
    StackPane root = new StackPane(message);
    Scene scene = new Scene(root, 420, 220);
    stage.setTitle("Course Welcome");
    stage.setScene(scene);
    stage.show();
    System.out.println("Window: " + stage.getTitle());
    System.out.println("Message: " + message.getText());
});
````
