## Concept

### Arrange children vertically and horizontally

**Vertical layout** places child nodes in a top-to-bottom sequence. JavaFX provides `VBox` for this job. A notice can use a VBox for its heading, location row, and instruction.

**Horizontal layout** places child nodes in a left-to-right sequence. An `HBox` can keep Room: beside Lab A. The HBox itself can be a child of a VBox. This is still one scene graph: one layout contains another layout, and each places its own children.

The order of the nodes in the constructor determines their order in these examples. `new HBox(8, place, room)` puts place first and room second. `new VBox(12, title, row, note)` puts title above row, then note below it. The numeric argument has a separate purpose: it sets the gap.

### Distinguish spacing from padding

**Layout spacing** is the gap between adjacent children in a container. The HBox gap of 8 separates the two labels in the row. The VBox gap of 12 separates the heading, row, and instruction.

**Container padding** is space between a container’s boundary and its children. `root.setPadding(new Insets(20))` requests 20 pixels along each edge of the root. `Insets` is a JavaFX value object that holds those edge amounts; this constructor uses the same amount on all four edges.

| Setting | Where the space appears | What to inspect |
|---|---|---|
| HBox spacing | Between the labels within the row. | Room: and Lab A remain distinct. |
| VBox spacing | Between the three vertical children. | Heading, row, and instruction do not touch. |
| Root padding | Around the group of children. | Content stays away from window edges. |

Increasing spacing does not add an outside margin. Increasing padding does not directly change the gap between every pair of children. Compare one change at a time so you can see which setting controls which area.

### Wrap a long Label within its available width

**Text wrapping** displays long text on more than one line when its allocated width is limited. A Label can request it with `note.setWrapText(true)`. `note.isWrapText()` reads that setting.

A true setting alone does not prove that the text is readable. The parent must allocate a useful width, and the window must have enough height for the resulting lines. Our VBox gives its label the available width while respecting the child’s size limits. As that width decreases, the long note can occupy more lines.

Keep the full instruction in the Label. Do not shorten the text merely to make a failing layout test pass. Check that the title, row, and last words of the note remain visible at both tested sizes. Extremely small windows can still make content unusable; wrapping is not a promise that every possible size will fit.

### Style a heading with a small CSS declaration

A **JavaFX inline style** is a Java String containing presentation settings for one node. The settings use CSS, a notation for describing appearance. Each declaration has a property name, a colon, a value, and a semicolon.

For example, `title.setStyle("-fx-font-size: 22px; -fx-font-weight: bold;")` makes the heading use a 22-pixel font size and bold weight. The surrounding double quotes make it a Java String. The text inside the quotes uses CSS syntax, not Java assignment syntax.

The `-fx-` prefix identifies these JavaFX CSS properties. `px` means pixels. Use size and weight to make the heading easy to identify, while the heading’s actual words still communicate its purpose. This lesson uses inline settings; no extra stylesheet file is required.

### Let the layout respond to a native resize

**Resize-aware layout** means that containers recalculate child placement and available size when the native window changes size. We let the layout manage the children instead of assigning each Label a fixed screen position.

### Read a small nested layout

After the supplied setup, this complete example opens a notice with a row inside a column:

```java
import javafx.stage.Stage;
import javafx.scene.Scene;
import javafx.scene.control.Label;
import javafx.scene.layout.VBox;
import javafx.scene.layout.HBox;
import javafx.geometry.Insets;
Fx.closeWindows();
Fx.run(() -> {
    Label title = new Label("Desk Information");
    title.setStyle("-fx-font-size: 22px; -fx-font-weight: bold;");
    HBox row = new HBox(8, new Label("Place:"), new Label("Library"));
    Label note = new Label("Bring your student identification when collecting your reserved equipment.");
    note.setWrapText(true);
    VBox root = new VBox(12, title, row, note);
    root.setPadding(new Insets(20));
    Stage stage = new Stage();
    stage.setTitle("Desk Information");
    stage.setScene(new Scene(root, 420, 240));
    stage.show();
    System.out.println("Wrapping: " + note.isWrapText());
});
```

The notebook prints `Wrapping: true`. On the Desktop, the heading appears above the place row, with the wrapping instruction below. The HBox has two Label children; the VBox has three children, one of which is that HBox. Count nodes according to their actual parent-child relationships, not just the number of visible lines.


### Prepare this kernel

This is supplied course support for running JavaFX inside IJava. Run it once after starting or restarting this notebook's Java kernel. The message `FX ready` means the support has initialized JavaFX and completed an operation on its application thread. Open the Workspace **Desktop** view to see the windows created by later cells.

`Fx.run(() -> { ... })` performs the enclosed UI work on the JavaFX Application Thread and waits for that short operation to finish. Use it for reading as well as changing a live window or its controls. `Fx.closeWindows()` hides the windows created by this kernel before another example opens its own. `Fx.start()` is safe to call again; it keeps JavaFX available after the last window closes.

The implementation below is provided runtime support. You do not need to write its thread-coordination machinery for this lesson. A thread is one sequence of execution; Module 13 studies how to coordinate more than one. Here your responsibility is to use the documented support operations and keep UI work short. The support uses a completion signal, a time limit, and an error holder so a later cell does not silently continue after unfinished or failed UI work.

Do not call `Platform.exit()` during notebook practice. That ends the toolkit for this kernel; restart the kernel and rerun setup if you do so. Closing a window is different from ending the toolkit. If setup reports a display error, check that the Workspace Desktop is running, then restart the kernel and rerun setup. A JavaFX window appears in the Desktop, not as an inline notebook control.


````java
import javafx.application.Platform;
import javafx.stage.Window;
import java.util.ArrayList;
import java.util.concurrent.CountDownLatch;
import java.util.concurrent.TimeUnit;
import java.util.concurrent.atomic.AtomicReference;
class Fx {
    static void run(Runnable action) throws InterruptedException {
        if (Platform.isFxApplicationThread()) {
            action.run();
            return;
        }
        CountDownLatch done = new CountDownLatch(1);
        AtomicReference<Throwable> failure = new AtomicReference<Throwable>();
        Platform.runLater(() -> {
            try { action.run(); }
            catch (Throwable error) { failure.set(error); }
            finally { done.countDown(); }
        });
        if (!done.await(10, TimeUnit.SECONDS)) {
            throw new IllegalStateException("FX operation timed out; restart the kernel.");
        }
        if (failure.get() != null) { throw new RuntimeException(failure.get()); }
    }
    static void start() throws InterruptedException {
        try { Platform.startup(() -> Platform.setImplicitExit(false)); }
        catch (IllegalStateException alreadyStarted) {
            // This call is also safe when this kernel already started JavaFX.
        }
        run(() -> Platform.setImplicitExit(false));
    }
    static void closeWindows() throws InterruptedException {
        run(() -> {
            for (Window window : new ArrayList<Window>(Window.getWindows())) {
                window.hide();
            }
        });
    }
}
Fx.start();
System.out.println("FX ready");
````

## Worked Example

**Subgoal 1: group the information.** Put the room labels in an HBox, then place that row between the heading and note in a VBox.

**Subgoal 2: make it readable.** Set gaps, root padding, a bold heading, and wrapping for the note.

**Subgoal 3: compare real sizes.** Inspect the initial window and use the Desktop resize keys to test a narrower one.


````java
import javafx.stage.Stage;
import javafx.scene.Scene;
import javafx.scene.control.Label;
import javafx.scene.layout.VBox;
import javafx.scene.layout.HBox;
import javafx.geometry.Insets;
Fx.closeWindows();
Fx.run(() -> {
    Stage stage = new Stage();
    Label title = new Label("Study Room");
    title.setStyle("-fx-font-size: 22px; -fx-font-weight: bold;");
    Label place = new Label("Room:");
    Label room = new Label("Lab A");
    HBox row = new HBox(8, place, room);
    Label note = new Label("Bring your course notes and leave the table ready for the next group.");
    note.setWrapText(true);
    VBox root = new VBox(12, title, row, note);
    root.setPadding(new Insets(20));
    stage.setTitle("Room Notice");
    stage.setScene(new Scene(root, 440, 220));
    stage.show();
    System.out.println("Room: " + room.getText());
    System.out.println("Wrapping: " + note.isWrapText());
});
````

Expected output:

```text
Room: Lab A
Wrapping: true
```

The row reports Lab A and the note has wrapping enabled. The title uses larger bold text. The VBox arranges the heading, HBox row, and note with 12-pixel gaps; its padding leaves 20 pixels around the group. The Desktop inspection checks that the note actually wraps and fits after resizing. The printed true value checks only the wrapping setting.
