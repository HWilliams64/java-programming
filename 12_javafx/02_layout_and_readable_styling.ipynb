## Why This Matters

A window can be visible while its message is difficult to use. If related information is scattered or the last words of an instruction are hidden, a person may misunderstand the task. Layout containers let a program group information and adapt its placement without calculating a separate position for every control.

The previous lesson connected a window to its content. This lesson organizes several pieces of that content and checks their readability. Clear spacing, edge space, and text hierarchy make later forms and button-driven interfaces easier to understand. Without actual window checks, a correct printed value can hide a clipped instruction or an unsuitable arrangement.

## Concept

### Arrange the notice from top to bottom

A study-room notice needs to show a heading, a room location, and a complete reminder. The heading is `Study Room`, the location is `Lab A`, and the reminder asks visitors to bring their course notes and leave the table ready for the next group. We want those parts to remain readable when the window becomes narrower.

The previous lesson centered one Label in a StackPane. Here we need a different relationship between the content items: one part below another. A **VBox** is a layout container that arranges its children vertically, from top to bottom.

```java
    VBox root = new VBox(12, title, row, note);
```

This fragment assumes `title`, `row`, and `note` have already been constructed. Their order in the constructor gives the visible order: heading first, location row second, reminder last. The first argument controls the gap between these children; we will examine that value after building the row.

The VBox becomes the Scene's root, just as the StackPane did in the earlier lesson. We change the layout container because the desired arrangement has changed. The Stage and Scene still manage the same window-to-content connection.


### Keep the location's label and value together

The location has two pieces: a label saying `Room:` and its value `Lab A`. These belong side by side on one line, rather than on separate vertical rows. An **HBox** is a layout container that arranges children horizontally.

```java
    Label place = new Label("Room:");
    Label room = new Label("Lab A");
    HBox row = new HBox(8, place, room);
```

The first two statements construct separate Label objects. `place` describes what the value means; `room` contains that value. The third constructs the HBox and supplies those labels in left-to-right order. Its first argument specifies the gap between them.

The HBox named `row` then appears as one child in the outer VBox. This is a **nested layout**: a layout container holds another layout container. The outer container decides where the entire row sits among the other sections. The inner container decides how the row's two labels sit beside each other. Keeping those responsibilities separate avoids manually calculating each label's coordinates.


### Use spacing between neighboring children

**Layout spacing** is the gap a container inserts between adjacent children. Our HBox uses `8` between `Room:` and `Lab A`. The VBox uses `12` between the heading and location row, then another `12` between the row and reminder.

Spacing belongs to the container that arranges those children. Changing the VBox's value changes the vertical gaps between its three sections. It does not change the HBox's gap between its two labels. Likewise, changing the HBox's value does not move every section farther apart vertically.

These numbers describe layout distances, not a number of text spaces inserted into a Label. The strings remain `Room:` and `Lab A`. Separating text from placement lets us revise the layout without editing the information the notice communicates.

Spacing also does not create a border around the entire notice. For that outer breathing room, the container needs padding.


### Reserve space inside the container's edges

**Container padding** is space between the container's edge and the area used for its children. It keeps the notice from touching the window's content edges. JavaFX represents these edge distances with an **Insets** object:

```java
    root.setPadding(new Insets(20));
```

This statement assumes the VBox named `root` exists and Insets has been imported. The one-argument constructor uses the same distance on all four sides: top, right, bottom, and left. The method `setPadding` assigns those distances to the root container.

Padding and spacing solve different placement problems. Padding reserves the outside edges of this container's content area; spacing separates its neighboring children. Our notice uses both: an outer distance of twenty and vertical gaps of twelve.

Padding also consumes available content space. At the initial scene width of 440, twenty units on each horizontal side leave 400 for the content area before any other layout constraints. Increasing padding in a fixed-width window leaves less width for the reminder, which can affect how its text wraps.

### Allow the reminder to use several lines

The reminder is longer than the heading and location. A single unbroken display line may not fit when the window narrows. **Text wrapping** allows a label to continue its text on additional display lines when its allocated width is limited.

```java
    note.setWrapText(true);
```

This statement enables wrapping on the Label named `note`. The Boolean `true` is the requested setting. The label's text remains the same complete reminder; wrapping changes its presentation rather than adding newline characters to the stored string.

The method `note.isWrapText()` later reports whether this setting is enabled. A report of `true` does not prove that every word is currently visible. The actual layout must provide enough width and height for the text at the tested window size.

We therefore inspect the native window as well as the reported property. The important result is a complete, readable reminder. At a narrower width it may need more lines and more height, even though its stored text has not changed.


### Give the heading a clear visual role

The heading should stand out from the location and instructions. A **JavaFX inline style** attaches styling declarations directly to one object. Our example styles only the heading Label:

```java
    title.setStyle("-fx-font-size: 22px; -fx-font-weight: bold;");
```

The argument is a Java String containing two **CSS declarations**, property-and-value instructions used by JavaFX's styling system. In the first declaration, `-fx-font-size` is the property, the colon separates it from its value `22px`, and the semicolon ends the declaration. The `px` unit requests a size in pixels.

The second declaration sets `-fx-font-weight` to `bold`. The hyphenated property names belong to JavaFX CSS; they are not Java variable names. The surrounding quotation marks make the complete styling text a string argument to `setStyle`.

This changes how the heading appears without changing its text. It also leaves the other labels' styles alone. Font size affects the space text needs, so styling and layout should be inspected together. This lesson uses a short inline style; it does not require a separate stylesheet or website-building workflow.

### Prepare the supplied notebook support

This is supplied course support for running JavaFX inside IJava. Run it once after starting or restarting this notebook's Java kernel. The message `FX ready` means the support has initialized JavaFX and completed an operation on its application thread. Open the Workspace **Desktop** view to see the windows created by later cells.

`Fx.run(() -> { ... })` performs the enclosed UI work on the JavaFX Application Thread and waits for that short operation to finish. Use it for reading as well as changing a live window or its controls. `Fx.closeWindows()` hides the windows created by this kernel before another example opens its own. `Fx.start()` is safe to call again; it keeps JavaFX available after the last window closes.

The implementation below is provided runtime support. You do not need to write its thread-coordination machinery for this lesson. A thread is one sequence of execution; Module 13 studies how to coordinate more than one. Here your responsibility is to use the documented support operations and keep UI work short. The support uses a completion signal, a time limit, and an error holder so a later cell does not silently continue after unfinished or failed UI work.

Do not call `Platform.exit()` during notebook practice. That ends the toolkit for this kernel; restart the kernel and rerun setup if you do so. Closing a window is different from ending the toolkit. If setup reports a display error, check that the Workspace Desktop is running, then restart the kernel and rerun setup. A JavaFX window appears in the Desktop, not as an inline notebook control.


````java
import javafx.application.Platform;
import javafx.stage.Window;
import java.util.ArrayList;
import java.util.concurrent.CountDownLatch;
import java.util.concurrent.TimeUnit;
import java.util.concurrent.atomic.AtomicReference;
class Fx {
    static void run(Runnable action) throws InterruptedException {
        if (Platform.isFxApplicationThread()) {
            action.run();
            return;
        }
        CountDownLatch done = new CountDownLatch(1);
        AtomicReference<Throwable> failure = new AtomicReference<Throwable>();
        Platform.runLater(() -> {
            try { action.run(); }
            catch (Throwable error) { failure.set(error); }
            finally { done.countDown(); }
        });
        if (!done.await(10, TimeUnit.SECONDS)) {
            throw new IllegalStateException("FX operation timed out; restart the kernel.");
        }
        if (failure.get() != null) { throw new RuntimeException(failure.get()); }
    }
    static void start() throws InterruptedException {
        try { Platform.startup(() -> Platform.setImplicitExit(false)); }
        catch (IllegalStateException alreadyStarted) {
            // This call is also safe when this kernel already started JavaFX.
        }
        run(() -> Platform.setImplicitExit(false));
    }
    static void closeWindows() throws InterruptedException {
        run(() -> {
            for (Window window : new ArrayList<Window>(Window.getWindows())) {
                window.hide();
            }
        });
    }
}
Fx.start();
System.out.println("FX ready");
````

### Read a complete nested layout

A library desk can use the same arrangement for equipment collection: a heading, a place row, and a complete instruction. This small example uses the supplied support above. Its imports make the window, scene, Label, layout, and Insets types available by their short names.

The two `new Label(...)` expressions inside the HBox constructor each create a child directly. Their references are supplied to the row without first assigning separate variable names. The row is then one of the VBox's three children. Copy this complete example into a Java work cell if you want to inspect it.

```java
import javafx.stage.Stage;
import javafx.scene.Scene;
import javafx.scene.control.Label;
import javafx.scene.layout.VBox;
import javafx.scene.layout.HBox;
import javafx.geometry.Insets;
Fx.closeWindows();
Fx.run(() -> {
    Label title = new Label("Desk Information");
    title.setStyle("-fx-font-size: 22px; -fx-font-weight: bold;");
    HBox row = new HBox(8, new Label("Place:"), new Label("Library"));
    Label note = new Label("Bring your student identification when collecting your reserved equipment.");
    note.setWrapText(true);
    VBox root = new VBox(12, title, row, note);
    root.setPadding(new Insets(20));
    Stage stage = new Stage();
    stage.setTitle("Desk Information");
    stage.setScene(new Scene(root, 420, 240));
    stage.show();
    System.out.println("Wrapping: " + note.isWrapText());
});
```

Expected output:

```text
Wrapping: true
```

The Desktop shows Desk Information as the title and heading, a Place: Library row, and the longer instruction below. ### Make a study-room notice readable

The study-room notice brings the seven concepts together. Its heading is Study Room, its row says Room: Lab A, and its instruction asks visitors to bring their course notes and leave the table ready for the next group. The native title bar reads Room Notice. Keep the heading and window title separate when checking the display.

**Group related content.** Construct the heading and give it a larger bold style. Construct the two room labels and place them in an HBox with a gap of 8. The row keeps its label and value together.

**Arrange the sections.** Enable wrapping on the long instruction. Put the heading, row, and instruction in a VBox with a gap of 12 between its children. Give the VBox padding of 20 along its edges. These are separate settings for separate regions of space.

**Connect, show, and inspect.** Attach the VBox to a Scene with an initial content area of 440 by 220, give the Stage that scene, and show it. The two prints read the room value and wrapping setting within `Fx.run`. Run the complete cell, then compare the actual Desktop view at its initial and narrower sizes.


````java
import javafx.stage.Stage;
import javafx.scene.Scene;
import javafx.scene.control.Label;
import javafx.scene.layout.VBox;
import javafx.scene.layout.HBox;
import javafx.geometry.Insets;
Fx.closeWindows();
Fx.run(() -> {
    Stage stage = new Stage();
    Label title = new Label("Study Room");
    title.setStyle("-fx-font-size: 22px; -fx-font-weight: bold;");
    Label place = new Label("Room:");
    Label room = new Label("Lab A");
    HBox row = new HBox(8, place, room);
    Label note = new Label("Bring your course notes and leave the table ready for the next group.");
    note.setWrapText(true);
    VBox root = new VBox(12, title, row, note);
    root.setPadding(new Insets(20));
    stage.setTitle("Room Notice");
    stage.setScene(new Scene(root, 440, 220));
    stage.show();
    System.out.println("Room: " + room.getText());
    System.out.println("Wrapping: " + note.isWrapText());
});
````

Expected output:

```text
Room: Lab A
Wrapping: true
```

The row reports Lab A and the note has wrapping enabled. The title uses larger bold text. The VBox arranges the heading, HBox row, and note with 12-pixel gaps; its padding leaves 20 pixels around the group. The Desktop inspection checks that the note actually wraps and fits after resizing. The printed true value checks only the wrapping setting.

## Summary

A VBox places its children vertically, while an HBox places its children horizontally. Nesting an HBox inside a VBox lets each container manage one relationship. Constructor order determines the order of these children.

Spacing separates neighboring children. Padding reserves space inside a container's edges. The Label's wrapping setting permits its text to occupy additional lines when the allocated width requires them. A short instruction can still fit on one line.

Inline JavaFX styles are Java Strings containing CSS declarations. The example changes the heading's font size and weight while keeping its meaning in its actual words. Native initial and narrow views test whether the complete notice remains readable; a getter reporting true cannot replace that check.
