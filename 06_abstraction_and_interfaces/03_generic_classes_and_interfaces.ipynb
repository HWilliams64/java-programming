# Generic Classes and Interfaces

You will write reusable classes and interfaces that keep their input, stored-value, and result types consistent.

You have already used collections such as `ArrayList<String>` and `ArrayList<Integer>`. The type written inside the angle brackets tells the caller what kind of element the collection accepts. In this lesson, you will move from using that design to writing a small version of your own.

You will begin with a holder for one value, then connect a generic class to a generic interface. The complete example uses separate shelves for names and whole-number scores. One class definition supports both, while each object keeps its own storage and each caller uses the appropriate types.

Use the [Module 6 glossary: abstraction, interfaces, and generics](terms.md) to review the terms introduced in this lesson.


## Learning Goals

By the end of this lesson, you will be able to:

- Define a generic class and interface with consistent types, distinguishing type parameters, type arguments, and values.
- Trace separate generic objects that store String and Integer values, including boxing and unboxing during storage and arithmetic.
- Test generic code to distinguish incompatible-type errors from incorrect values or behavior.


## Why This Matters

Storage logic often stays the same even when the stored data changes. An application may need to hold a customer name, a whole-number count, or a record object. **Generics** let a type definition describe those relationships using a type placeholder. A caller supplies the particular type it needs.

This can avoid separate text-only and number-only versions of the same class. The compiler can also detect operations that supply an incompatible type, such as adding text to a collection of Integer values. Those checks catch one category of mistakes before execution, but they cannot tell whether a correctly typed value is the value the task requires. You still need behavior tests.

You will combine the private state, public methods, and interface relationships from earlier lessons with the type arguments you used in Module 4. That combination helps you understand reusable library APIs and implement a class that follows a supplied generic interface.

## Concept

### Connect stored values and returned values with a type parameter

Suppose an application needs a small object that holds one value and returns it when asked. A String-only holder would work for a message, but a count would need another version. The storage and retrieval steps are otherwise the same.

A **type parameter** names a type placeholder in a class or interface declaration. In `Holder<T>`, `T` is that placeholder. A **generic class** uses the parameter to describe related types throughout its definition. Here the field, constructor argument, and `get()` result all use T. This says that the stored value and returned value must agree with the selected type.

A **type argument** is the particular type chosen when the generic definition is used. `Holder<String>` selects String. The word `Open` is different: it is the value passed to the constructor, not a type. The constructor is named `Holder`, without `<T>` after its name, because the surrounding class declaration already introduces T.

Run the complete example. It constructs a holder for the message `Open`, retrieves that message into a String variable, and prints it.

````java
class Holder<T> {
    private T value;
    public Holder(T value) {
        this.value = value;
    }
    public T get() {
        return value;
    }
}
Holder<String> message = new Holder<String>("Open");
String text = message.get();
System.out.println(text);
````

The output is `Open`. `new Holder<String>("Open")` constructs one object and supplies the initial message. In the constructor, `this.value = value` stores the argument in that object's field. The `get()` method returns the stored value. Because this use selected String, the caller can assign the result to `String text` and print it.

Keep the three roles separate: T is a type parameter in the definition, String is a type argument at the use site, and `Open` is a stored value. T is not a variable that holds the text, and it does not change into a new type whenever a different value arrives. It expresses a relationship the compiler checks.

The field could not simply be changed to String while keeping a general T constructor input and result. That would break the agreement needed for other type arguments. Next, we will use the unchanged definition with Integer as well as String.

### Use the same definition with another type

The same Holder definition can support a place name and a whole-number count. Each object has its own field, even though both objects come from the same class definition. Sharing the definition does not mean sharing the stored value.

The next example selects String for `place` and Integer for `count`. Generic type arguments are reference types, so `Holder<int>` is not a valid alternative to `Holder<Integer>`. Integer is the wrapper type you used with collections in Module 4.

The constructor call with 4 uses boxing to supply an Integer value. Later, `count.get()` provides an Integer result, and arithmetic unboxes its value to an int before adding one. This example uses a nonnull value, so unboxing has a number to use.

Run the complete example and follow the place value separately from the count. The program prints the place, then the result of adding one to the retrieved count.

````java
class Holder<T> {
    private T value;
    public Holder(T value) {
        this.value = value;
    }
    public T get() {
        return value;
    }
}
Holder<String> place = new Holder<String>("Hall");
Holder<Integer> count = new Holder<Integer>(4);
int nextCount = count.get() + 1;
System.out.println(place.get());
System.out.println(nextCount);
````

The first line is `Hall`, and the second is `5`. The `place` object stores the String `Hall`. The separate `count` object stores an Integer representing 4. Calling `place.get()` returns the place name. Calling `count.get()` supplies the value used by the addition, producing the int value 5 in `nextCount`.

That addition does not replace the value stored inside `count`. This Holder example only has a constructor and a getter; the caller computes a separate result. Likewise, constructing the Integer holder does not change the existing String holder.

The selected type controls compatible operations, but it does not prove that the value is correct for the task. A count of 8 would be a valid Integer and could still be the wrong input. Type checks and behavior tests answer different questions. We will preserve that distinction when the generic relationship passes through an interface.

### Carry the type relationship through an interface

A caller may need the ability to retrieve a value without depending on the class that stores it. We can express that promise with a **generic interface**, then connect an implementing class to it.

In `Source<T>`, the required `get()` operation returns T. `FixedSource<T>` declares its own type parameter and uses it in `implements Source<T>`. That connection makes the interface's promised result agree with the class's parameter. The private field, constructor input, and public `get()` result use that same class parameter.

The caller selects String in both `Source<String>` and `new FixedSource<String>("Map")`. Through the interface-typed reference, `source.get()` therefore supplies a String and can initialize `String item`. The constructor's input value is `Map`; the type selected for the relationship is String.

Run the complete example to trace that value from the constructor, through the field and interface operation, to the printed result.

````java
interface Source<T> {
    T get();
}
class FixedSource<T> implements Source<T> {
    private T value;
    public FixedSource(T value) {
        this.value = value;
    }
    @Override
    public T get() {
        return value;
    }
}
Source<String> source = new FixedSource<String>("Map");
String item = source.get();
System.out.println(item);
````

The output is `Map`. The actual object is a FixedSource, and the caller's declared reference type is `Source<String>`. The constructor stores `Map` in the object's private field. The interface exposes `get()`, and FixedSource's implementation returns the stored value.

The caller receives the result as a String because its interface type selected String. The class also selected String for this construction, so its stored value and method result agree with that promise. The `@Override` annotation checks that the class supplies the interface operation; it does not replace the type relationship written in the declarations.

This example holds one value directly. The worked example will apply the same relationship to an internal ArrayList: the interface methods, implementing methods, and list element type will agree. Each new shelf will still create its own list. We will use one shelf for text names and another for Integer scores.

### Distinguish Type Errors from Wrong Values

**Compile-time generic type checking** detects operations that do not fit the chosen type relationship. For a `Holder<String>`, supplying the number 3 to its constructor does not satisfy the String input type. The expression `new Holder<String>(3)` is therefore an invalid example for discussion only. Likewise, `Holder<int>` is invalid because int is a primitive type, not a permitted generic type argument.

Keep these invalid expressions out of runnable examples. To diagnose a mismatch, compare the operation's required type with the supplied type. Repair the inconsistent input while preserving the intended interface and class relationship.

A value can have the correct type and still be wrong for the task. A holder that should store `Open` could receive the String `Closed`; both are valid Strings. An expected-value test is needed to catch that behavior error. The next example combines type checks with actual output checks for two separate collections.

### Carry the Type Through a Collection Wrapper

The worked program expands the one-value idea into a small collection interface. `Sequence<T>` promises add, get, and size. `Shelf<T>` implements those operations using a private `ArrayList<T>`. The same type relationship must hold at every step.

| Location | Relationship |
|---|---|
| `Sequence<T>` | add accepts T, get returns T, and size returns an int count. |
| `Shelf<T> implements Sequence<T>` | The class supplies operations using its own T consistently. |
| `ArrayList<T> values` | The private list stores that same element type. |
| `Sequence<String> names` | The caller adds String values and receives String results from get. |
| `Sequence<Integer> scores` | The caller adds Integer values and receives Integer results from get. |

The chosen element type does not change indexing rules. Valid get indexes run from 0 through size() minus 1, and an empty sequence has no valid element index. Each constructor creates a new internal list, so independently constructed shelves keep independent contents.

## Worked Example

### Keep Names and Scores in Separate Typed Shelves

Staff need a small report containing the first stored name, the number of names, and the total of two whole-number scores. The names are `Maya` followed by `Luis`. The scores are 4 and 7 points. These are two independent collections; this example does not pair a particular name with a particular score. All requested indexes will refer to existing entries.

`Sequence<T>` describes three operations: add accepts a T value, get returns a T value at an index, and size returns an int count. `Shelf<T>` implements those operations using a private `ArrayList<T>`. Its constructor creates the list. Each public method forwards the request to that object's list, while the type parameter keeps the accepted and returned element types consistent.

The caller first constructs a `Shelf<String>` and uses it through `Sequence<String> names`. Adding Maya and Luis stores two entries in that order. Index 0 supplies the first name, and size reports how many entries were added. The caller then constructs a separate `Shelf<Integer>`, adds 4 and 7, and retrieves both values for addition. Boxing and unboxing provide the familiar conversions between Integer values and int arithmetic.

This tutorial wrapper deliberately uses a library list internally so we can focus on generic interface relationships. Your graded StudentArrayList task has its own supplied interface and storage requirements; this example does not replace those requirements. Run the complete program and follow the two shelves separately.

````java
import java.util.ArrayList;
interface Sequence<T> {
    void add(T value);
    T get(int index);
    int size();
}
class Shelf<T> implements Sequence<T> {
    private ArrayList<T> values;
    public Shelf() {
        values = new ArrayList<T>();
    }
    @Override
    public void add(T value) {
        values.add(value);
    }
    @Override
    public T get(int index) {
        return values.get(index);
    }
    @Override
    public int size() {
        return values.size();
    }
}
Sequence<String> names = new Shelf<String>();
names.add("Maya");
names.add("Luis");
System.out.println("First: " + names.get(0));
System.out.println("Names: " + names.size());
Sequence<Integer> scores = new Shelf<Integer>();
scores.add(4);
scores.add(7);
int total = scores.get(0) + scores.get(1);
System.out.println("Total: " + total);
````

The program prints:

```text
First: Maya
Names: 2
Total: 11
```

The names shelf has its own private list containing Maya at index 0 and Luis at index 1. `names.get(0)` returns a String, so the first report includes `Maya`. `names.size()` returns the number of stored entries, 2. It does not report the list's unused backing capacity.

The scores shelf owns a different list containing Integer values for 4 and 7. Its get operations supply those two values, which are unboxed for addition. The result is the int value 11 in `total`. Computing that total does not replace either stored score, and adding scores does not alter the names list.

The same Shelf definition supports both uses because T connects the interface, methods, and internal element type. Through names, add expects String and get supplies String. Through scores, those operations use Integer. A quoted value such as `"7"` would not satisfy the scores input type; an unquoted but incorrect value such as 8 would satisfy the type and still fail a test expecting the intended total.

In `Slot<String>`, route starts as desk and replace stores studio. In `Slot<Integer>`, quantity starts as 2; get supplies an Integer that is unboxed for adding 3, and the result is boxed for replace, leaving 5.

Both calls use their declared generic interfaces and matching public implementations. The replacement updates the current object’s private field using this.value. Each new ValueSlot constructs a separate instance.

## Summary

A type parameter names a relationship in a generic definition. A type argument selects a particular type when that definition is used. A stored value is data accepted by the chosen type, not the type itself. Keeping these roles separate helps you trace a value from a constructor or method input into a field and back to the caller.

A generic class can implement a generic interface while keeping its input, storage, and result types consistent. Selecting String makes the generic input and result types String. Selecting Integer makes those types Integer, with boxing and unboxing where primitive arithmetic is involved. Separately constructed objects still own separate state.

The compiler rejects incompatible types, but a compatible value can still be wrong for the task. Good tests check expected values, valid indexes, boundary cases, and whether independent objects remain independent.
