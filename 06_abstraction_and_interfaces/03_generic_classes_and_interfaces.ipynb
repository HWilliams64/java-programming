## Concept

### Declare a placeholder for a type

A **type parameter** is a type placeholder declared by a class or interface. T is a common name for a general type parameter. It is a name in the type definition, not a String value and not a variable that stores arbitrary objects.

A **generic class** uses a type parameter consistently for values it stores or exchanges. Here T connects the field, constructor parameter, and method result:

```java
class Holder<T> {
    private T value;
    public Holder(T value) {
        this.value = value;
    }
    public T get() {
        return value;
    }
}
Holder<String> message = new Holder<String>("Open");
String text = message.get();
System.out.println(text);
```

This prints `Open`. In the definition, T is a type parameter. At the use site, String is the type argument. For this use, the constructor accepts a String and get returns a String. The constructor name is Holder, not `Holder<T>`; the class declaration already supplies its type parameter.

Keep the same type relationship everywhere it is required. A String field would no longer match a general T constructor parameter; a text-only design would need matching String operations instead. Returning a different type would break the promised connection between the stored value and result.

### Use the same definition with another type argument

The same class definition also supports Integer values:

```java
class Holder<T> {
    private T value;
    public Holder(T value) {
        this.value = value;
    }
    public T get() {
        return value;
    }
}
Holder<String> place = new Holder<String>("Hall");
Holder<Integer> count = new Holder<Integer>(4);
int nextCount = count.get() + 1;
System.out.println(place.get());
System.out.println(nextCount);
```

The output is `Hall` and then 5. Boxing converts the whole-number value for storage as an Integer reference; unboxing provides an int for the addition. These are the conversions you practiced with HashMap. The examples store nonnull numbers, so this unboxing has a value to use.

You choose a reference type such as Integer as a generic type argument. `Holder<int>` is not a valid declaration. Each object still has its own field; using the same generic class definition does not make the objects share stored values.

### State a generic operation contract

A **generic interface** uses type parameters to relate its method inputs and results. An implementation carries the same type parameter through the contract:

```java
interface Source<T> {
    T get();
}
class FixedSource<T> implements Source<T> {
    private T value;
    public FixedSource(T value) {
        this.value = value;
    }
    @Override
    public T get() {
        return value;
    }
}
Source<String> source = new FixedSource<String>("Map");
String item = source.get();
System.out.println(item);
```

This prints `Map`. `Source<T>` requires a result of its element type. `FixedSource<T>` implements that same promise. At the use site, both interface and implementation use String. The interface reference exposes get while the object holds its own value.

The interface and class each declare their own type parameter. Writing implements `Source<T>` connects the class’s parameter to the interface’s parameter for this implementation. You may use a different parameter name in another definition, but consistency within each relationship matters.

### Let the compiler check the agreed types

**Compile-time generic type checking** rejects an incompatible operation before that operation runs. For example, after declaring `Holder<String> message`, a constructor call that tries to supply a numeric value for that String use is invalid. IJava processes a cell as several snippets, so valid statements earlier in a cell can run before a later snippet is rejected. A rejected cell does not undo those earlier effects.

These are intentionally incorrect expressions for discussion only: `new Holder<String>(3)` and `Holder<int>`. Do not place them in the runnable examples. State the required type, identify the supplied type, and fix the mismatch before execution.

Type checking does not prove that a program’s behavior is correct. A String slot could still receive the wrong room name. You must still test state changes, values, and boundaries. Keep the declared interface and implementation type arguments consistent in this lesson rather than trying to widen or convert them.

### Carry the type through a collection wrapper

The worked example uses `Sequence<T>` to promise add, get, and size. `Shelf<T>` keeps a private `ArrayList<T>` and forwards those operations to it. The type parameter connects the interface method, class method, and stored element type.

| Location | Relationship |
|---|---|
| `Sequence<T>` | Declares that add accepts T and get returns T. |
| `Shelf<T>` implements `Sequence<T>` | Promises those same operations for the class’s type parameter. |
| `ArrayList<T>` values | Stores the same element type inside the implementation. |
| `Sequence<String>` names | Gives the caller String inputs and results. |
| `Sequence<Integer>` scores | Gives the caller Integer inputs and results, with familiar boxing and unboxing for arithmetic. |

Use valid get indexes from 0 through size() - 1. Empty sequences have no valid element index. The type parameter does not change those collection rules.

Shelf is a tutorial example that uses a library list internally. Your graded StudentArrayList assignment follows the instructor-provided SimpleArrayList interface and storage requirements. Do not replace those requirements with this tutorial wrapper. Read the supplied interface’s actual method names, parameter types, result types, and behavior before implementing it.

## Worked Example

**Subgoal 1: define the typed operations.** `Sequence<T>` relates add inputs and get results to the same type parameter.

**Subgoal 2: implement the contract.** `Shelf<T>` uses a private `ArrayList<T>` and matching public methods.

**Subgoal 3: use two element types.** Construct separate String and Integer shelves through their corresponding interface references.


````java
import java.util.ArrayList;
interface Sequence<T> {
    void add(T value);
    T get(int index);
    int size();
}
class Shelf<T> implements Sequence<T> {
    private ArrayList<T> values;
    public Shelf() {
        values = new ArrayList<T>();
    }
    @Override
    public void add(T value) {
        values.add(value);
    }
    @Override
    public T get(int index) {
        return values.get(index);
    }
    @Override
    public int size() {
        return values.size();
    }
}
Sequence<String> names = new Shelf<String>();
names.add("Maya");
names.add("Luis");
System.out.println("First: " + names.get(0));
System.out.println("Names: " + names.size());
Sequence<Integer> scores = new Shelf<Integer>();
scores.add(4);
scores.add(7);
int total = scores.get(0) + scores.get(1);
System.out.println("Total: " + total);
````

Expected output:

```text
First: Maya
Names: 2
Total: 11
```

The names sequence stores two String values and returns Maya from index 0. Its size is 2. The scores sequence stores 4 and 7 as Integer values; retrieval and unboxing let the program compute 11. The class logic is reused, while each object has its own list and each declared interface use has its own agreed element type.

`Slot<T>` connects get results and replace inputs to one type parameter. `ValueSlot<T>` implements that contract and stores a private T value initialized by its constructor. In `Slot<String>`, route starts as desk and replace stores studio. In `Slot<Integer>`, quantity starts as 2; get supplies an Integer that is unboxed for adding 3, and the result is boxed for replace, leaving 5. Both calls use their declared generic interfaces and matching public implementations. The replacement updates the current object’s private field using this.value. Each new ValueSlot constructs a separate instance. A numeric replacement cannot satisfy the String slot’s promised input type, even though the same generic definition also supports a separate Integer use. Starting quantity at 0 while preserving the +3 operation leaves Quantity: 3. Starting at 2 and explicitly replacing it with 0 leaves Quantity: 0. Both are valid Integer-slot uses, but they test different state transitions. The separate String slot is constructed with lab before route is changed to studio; its own private field remains lab, while the baseline quantity still becomes 5. Sharing `ValueSlot<String>` as a class use does not make two new instances share a field. Conversely, assigning another reference to route would create an alias and would not establish independent state. The compiler rejects numeric input for `Slot<String>`; it still permits a wrong String value, which manual output checks must catch.

```java
interface Slot<T> {
    T get();
    void replace(T value);
}
class ValueSlot<T> implements Slot<T> {
    private T value;
    public ValueSlot(T value) {
        this.value = value;
    }
    @Override
    public T get() {
        return value;
    }
    @Override
    public void replace(T value) {
        this.value = value;
    }
}
Slot<String> route = new ValueSlot<String>("desk");
Slot<Integer> quantity = new ValueSlot<Integer>(2);
route.replace("studio");
quantity.replace(quantity.get() + 3);
System.out.println("Route: " + route.get());
System.out.println("Quantity: " + quantity.get());
```

Expected output:

```text
Route: studio
Quantity: 5
```

## Summary

A type parameter such as T lets a class or interface describe a consistent relationship between stored values, method inputs, and results. A type argument such as String selects that relationship at a use site. A generic class can implement a generic interface and use generic storage. The compiler rejects incompatible types, while tests still check behavior and boundaries.
