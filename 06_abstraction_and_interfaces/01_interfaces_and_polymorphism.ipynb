# Interfaces and Polymorphism

You will define a shared set of operations and use it to work with objects from different classes.

So far, each class has described its own data and methods. Now we need a way to ask different objects for the same kind of result. Campus staff, for example, may want one program to announce both room locations and course identifiers. The objects store different information, but each can supply a short text label.

You will first describe that shared operation with an interface. Then you will build classes that provide it, trace which method runs, and add a shared announcement method. The complete example prints one announcement for each object in an array. You will use the same ideas to design a different set of objects in practice.

[Review the module vocabulary](terms.md) as you meet new terms.


## Learning Goals

By the end of this lesson, you will be able to:

- Define a shared interface, implement it in two classes, and use compiler feedback and output tests to check their behavior.
- Distinguish a reference variable's declared type from the actual object's class, and trace which instance method runs as a loop visits different objects.
- Explain how a default method combines shared behavior with the receiving object's required operation.


## Why This Matters

Programs often need to perform the same operation on objects that work differently. A reporting tool may request a description from several kinds of records. A payment system may request a total from several pricing rules. An **interface** names a set of operations that a caller can use. Each implementing class supplies the behavior those operations promise.

This gives the caller a stable way to request a result. Without a shared type, the caller may need separate code for each concrete class. That repetition makes new cases harder to add and existing cases easier to handle inconsistently. Interfaces do not guarantee that a result is useful or correct; the classes must also follow the operation's stated meaning.

This lesson connects the classes and objects from Module 5 to reusable collection designs. You will learn how one loop can call the same operation while each object supplies its own implementation. Later, that separation will help you use and implement collections through their published interfaces.

## Concept

### Describe the operation before its implementation

Campus staff need labels they can place in announcements. A room label might contain a location, while a course label might contain a course identifier. The caller needs a text result from either kind of object. It does not need to know which field stores the information.

**Abstraction** means describing what an operation promises while leaving the steps that fulfill the promise elsewhere. Here the promise is simple: return a useful text label. An **interface** gives that shared capability a name in Java. The `interface` keyword introduces the type, and `Labelled` is the name we choose for it.

Inside the declaration, `String label();` describes the required operation. `String` is the result type, `label` is the method name, and the empty parentheses mean the caller supplies no arguments. This declaration has no method body, so it ends with a semicolon. In this interface, the required method is public even though we have not written `public`.

Run the declaration below. It introduces the shared type; it does not yet create a sign or print a label.

````java
interface Labelled {
    String label();
}
````

The declaration produces no printed output. Defining `Labelled` tells Java about a type and its required operation; it does not call `label()`. You also cannot construct the interface directly with `new Labelled()`.

The compiler can check the operation's name, parameter types, and compatible result type. It cannot decide whether returned text is useful to campus staff. That behavioral meaning belongs in the interface's explanation and in tests. Next, we will write a class that fulfills both the Java declaration and the intended purpose.

### Supply the promised operation

The interface describes the request, but a class must explain how to answer it. Our first class will hold a room location and put the word `Room` before that location. For a location named `Gate`, the useful label will be `Room Gate`.

**Interface implementation** connects a class to an interface and provides the required behavior. The keyword `implements` declares that connection in `class RoomLabel implements Labelled`. The private field and constructor use the patterns you practiced in Module 5: the object stores its own location, and `this.room = room` copies the constructor argument into that object's field.

The public `label` method supplies the missing behavior. It returns a String, takes no arguments, and uses the required method name. The implementation must preserve the public access promised by the interface. Making this method private would prevent it from fulfilling that promise.

The **`@Override` annotation** asks the compiler to check the intended implementation or overriding relationship. An annotation is information for Java tools; it is not a method call or a language keyword. Here it helps catch a misspelled method name. The annotation does not supply the method or replace the `implements` declaration.

The following cell includes the interface again so the example contains every type it needs. It constructs one room label and prints the result of calling its method.

````java
interface Labelled {
    String label();
}
class RoomLabel implements Labelled {
    private String room;
    public RoomLabel(String room) {
        this.room = room;
    }
    @Override
    public String label() {
        return "Room " + room;
    }
}
RoomLabel room = new RoomLabel("Gate");
System.out.println(room.label());
````

The output is `Room Gate`. Construction stores `Gate` in the new object's private `room` field. The caller then requests `room.label()`. Inside that method, the String expression combines the prefix `Room ` with the stored location and returns the new text to the caller. `println` displays that returned value.

The caller never reads the private field directly. It uses the public operation that the class supplies. Notice that `room` is also used as the caller's variable name: `room.label()` names the object through that variable, while the `room` inside the method refers to the field of the receiving object.

A class that merely has a similarly named method does not automatically establish the interface relationship. The declaration, the public implementation, and the promised meaning all matter. The animation follows those connections before we narrow a caller's view to an interface type.

### Separate the reference type from the object's class

Suppose a caller only needs text. It should be able to request that text without depending on every additional method the object happens to provide. An **interface-typed reference** uses an interface as the variable's declared type and refers to an object that implements that interface.

This example introduces `TextProvider`, whose operation is `text()`. The `FixedText` class stores a supplied String and returns it from that method. It also offers `textLength()`, which returns the number of characters in the stored value. That second method belongs to the class, but it is not part of the TextProvider promise.

In `TextProvider message = new FixedText("Open")`, the left side determines the operations available through `message`. The right side constructs the actual FixedText object. These two types describe different parts of the same arrangement; assigning the reference does not turn the object into a different class.

Run the complete example and follow the `text()` call through `message`.

````java
interface TextProvider {
    String text();
}
class FixedText implements TextProvider {
    private String value;
    public FixedText(String value) {
        this.value = value;
    }
    @Override
    public String text() {
        return value;
    }
    public int textLength() {
        return value.length();
    }
}
TextProvider message = new FixedText("Open");
System.out.println(message.text());
````

The output is `Open`. There is one `new` expression, so this example constructs one FixedText object. Its field stores `Open`, and `message` refers to it. The declared type of `message` remains TextProvider; the object's class remains FixedText.

TextProvider declares `text()`, so `message.text()` is an available call. The receiving object's FixedText method returns the stored value. By contrast, `message.textLength()` is unavailable through this declared type because TextProvider does not declare that operation. This is a compile-time restriction on the caller's expression, not evidence that the actual object has lost a method.

If a task specifically needs the extra class operation, a FixedText reference can expose it. For a caller that only requests text, the interface is enough. Next, we will use that common view with more than one implementation.

### Let each object provide its implementation

Now imagine a display that needs two greetings in a chosen order: a morning greeting followed by an evening greeting. Both objects can provide words, but their methods return different text. We can place their references in one array whose element type is the shared interface.

A **polymorphic method call** selects an instance-method implementation using the actual receiving object. The caller uses the same operation even when successive objects provide different implementations. Java's selection of that implementation is also called **method dispatch**.

`Greeting` declares the `words()` operation. MorningGreeting and EveningGreeting each implement it. Their explicit no-argument constructors have empty bodies because these examples have no fields to initialize. The array initializer constructs the morning object first and the evening object second.

The enhanced-for loop follows the array order you practiced earlier. Its variable has the declared type Greeting on every iteration. The reference stored in that variable changes, so the same `greeting.words()` expression can reach a different method body. Run the example and compare the two output lines.

````java
interface Greeting {
    String words();
}
class MorningGreeting implements Greeting {
    public MorningGreeting() {
    }
    @Override
    public String words() {
        return "Good morning";
    }
}
class EveningGreeting implements Greeting {
    public EveningGreeting() {
    }
    @Override
    public String words() {
        return "Good evening";
    }
}
Greeting[] greetings = {new MorningGreeting(), new EveningGreeting()};
for (Greeting greeting : greetings) {
    System.out.println(greeting.words());
}
````

The first line is `Good morning`, and the second is `Good evening`. On the first iteration, `greeting` refers to the MorningGreeting object. Its `words()` method returns `Good morning`, which the loop prints. On the second iteration, the loop stores the next reference in `greeting`; EveningGreeting's method returns `Good evening`.

The declared variable type and the written loop body stay the same. The receiving object changes. There is no separate condition that asks which class the object belongs to and chooses a matching print statement. Each object answers the operation it promised to provide.

Another class could participate if it implements Greeting and honors the meaning of `words()`. The caller can then use that object through the same interface. Our next example adds a shared method body while preserving this selection of object-specific behavior.

### Share the common steps in a default method

Our room labels are now useful on their own. For an announcement, staff also want the same `Ready: ` prefix before every label. Writing that prefix separately in every implementing class would repeat the same formatting step.

A **default interface method** supplies a method body that implementing classes can inherit or override. The `default` keyword distinguishes this method from the required, body-free `label()` declaration. In the following interface, `announcement()` returns the shared prefix followed by the result of calling `label()`.

That call still needs a receiving object. When a RoomLabel object uses the default announcement method, the inner `label()` call uses that same object and reaches RoomLabel's implementation. The interface supplies the common formatting; the object supplies its particular label.

The example uses the location `West`. Read the complete definition, then run it to follow the shared method and the class method to the final printed announcement.

````java
interface Labelled {
    String label();
    default String announcement() {
        return "Ready: " + label();
    }
}
class RoomLabel implements Labelled {
    private String room;
    public RoomLabel(String room) {
        this.room = room;
    }
    @Override
    public String label() {
        return "Room " + room;
    }
}
Labelled sign = new RoomLabel("West");
System.out.println(sign.announcement());
````

The output is `Ready: Room West`. `sign` refers to the RoomLabel object initialized with `West`. RoomLabel does not replace `announcement()`, so the call enters the inherited default body. That body asks the same object for its label. RoomLabel's `label()` returns `Room West`, and the default body adds `Ready: ` before returning the complete String.

The caller's `println` displays that final result. The prefix is defined once, while the location still comes from the particular object. A class may override the default when a different implementation fits the same promise; you will try that in guided practice.

A default is useful when common behavior makes sense for all participating implementations. It does not make every future interface change safe. Adding a new required operation can leave existing classes incomplete, and multiple inherited defaults can require an explicit choice. This lesson uses one interface per class so we can focus on the call sequence.

### Keep the Contract and the Call Separate

Before combining the examples, separate the questions that the type declarations and the running program answer.

| Part of the design | Question it answers |
|---|---|
| Interface declaration | Which operations does the shared type promise? |
| Implementing class | Which compatible public methods fulfill that promise? |
| Declared reference type | Which operations may the caller request through this variable? |
| Actual receiving object | Which implementation runs for an ordinary instance call? |
| Default method | Which common steps can use the receiving object's required operation? |

Use these questions in that order when you trace a program. Establish what the caller may request, identify the receiving object, and then follow the method bodies and returned values. The next example combines two label classes in one announcement loop.

## Worked Example

### Announce Different Kinds of Signs

Campus staff are preparing announcements for a room and a course. The room's location is the text `Lab A`, and the course identifier is the text `CSC-239`. These are labels, not quantities to calculate. Staff want one line for each sign, in room-first order. Every line must begin with `Ready: ` and identify what the remaining text describes.

We will keep the shared request in Labelled. Its required `label()` method returns the object-specific text, and its default `announcement()` method adds the common prefix. RoomLabel stores its location in `room`; CourseLabel stores its identifier in `code`. Each constructor initializes its own object's field, and each public label method combines a descriptive prefix with that field.

The caller constructs the two objects and places their references in a Labelled array. The array preserves the requested room-first order. In each loop iteration, `sign.announcement()` requests the same operation through the same declared type. The default method then calls `label()` on the current object, allowing that object's class to provide the appropriate wording.

This arrangement separates three decisions: the array determines order, each label method determines the sign's text, and the default method determines the shared announcement prefix. Run the complete program and follow those decisions through both iterations.

````java
interface Labelled {
    String label();
    default String announcement() {
        return "Ready: " + label();
    }
}
class RoomLabel implements Labelled {
    private String room;
    public RoomLabel(String room) {
        this.room = room;
    }
    @Override
    public String label() {
        return "Room " + room;
    }
}
class CourseLabel implements Labelled {
    private String code;
    public CourseLabel(String code) {
        this.code = code;
    }
    @Override
    public String label() {
        return "Course " + code;
    }
}
Labelled[] signs = {new RoomLabel("Lab A"), new CourseLabel("CSC-239")};
for (Labelled sign : signs) {
    System.out.println(sign.announcement());
}
````

The program prints:

```text
Ready: Room Lab A
Ready: Course CSC-239
```

During the first iteration, `sign` refers to the RoomLabel constructed with `Lab A`. RoomLabel inherits the default announcement method, so that body runs first. Its `label()` call reaches RoomLabel's public method, which returns `Room Lab A`. The default body adds `Ready: `, and the caller prints the first line.

During the second iteration, `sign` refers to the CourseLabel constructed with `CSC-239`. The same default body now calls CourseLabel's label implementation. That method returns `Course CSC-239`, producing the second announcement. The declared type of `sign` is still Labelled; the reference value and receiving class have changed.

Neither iteration reads a private field from the caller. Each object supplies its label through the interface operation. The program also does not need a separate branch for rooms and courses. ## Summary

An interface names operations that different classes can provide. A required method declaration states the operation's shape, while each implementing class supplies a compatible public method and honors its meaning. The `implements` declaration establishes the relationship, and `@Override` checks an intended implementing or overriding method.

A reference's declared type determines which operations the caller can request. The actual receiving object determines the implementation selected for an ordinary polymorphic instance call. A default method can share common steps while calling a required operation on that same object.

These ideas let one caller work with different implementations. They still require careful behavior tests: a method can return a value of the required type and still produce the wrong label or total.
