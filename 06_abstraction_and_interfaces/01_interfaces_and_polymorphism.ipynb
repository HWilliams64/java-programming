## Concept

### Describe the promise before the implementation

**Abstraction** means describing what an operation promises while leaving its implementation details elsewhere. For a sign, the promise can be “supply a text label.” The caller does not need to know which field supplies the text.

An **interface** is a named Java type that states operations its implementations provide. This declaration defines an interface:

```java
interface Labelled {
    String label();
}
```

The required label operation returns a String and takes no arguments. Its declaration ends with a semicolon because it has no method body here. This required interface operation is public even though the declaration omits the word public. An interface is not an object that you construct directly.

The method declaration gives the compiler a shape to check. Prose gives the operation its meaning: label should return a useful text description. The compiler can check types and method names; it cannot decide whether the text is useful to a person.

### Supply the promised operation

**Interface implementation** means that a class declares implements and supplies the required operations. Here the class stores a room name and uses it to produce a label:

```java
interface Labelled {
    String label();
}
class RoomLabel implements Labelled {
    private String room;
    public RoomLabel(String room) {
        this.room = room;
    }
    @Override
    public String label() {
        return "Room " + room;
    }
}
RoomLabel room = new RoomLabel("Gate");
System.out.println(room.label());
```

The output is `Room Gate`. The constructor and private field follow the patterns from Module 5. The new part is implements Labelled and the matching public label method. An implementation cannot reduce this required public access.

The **@Override annotation** asks the compiler to check that this method implements or overrides an inherited operation. An annotation is information written for Java tools. Here it catches mistakes such as naming the method labels instead of label. It does not create the relationship by itself; implements declares the interface relationship, and the method must match the required name and parameter types with a compatible result type.

### Use the interface as a variable type

An **interface-typed reference** has the interface as its declared type and refers to an object that implements it. Consider the complete example:

```java
interface TextProvider {
    String text();
}
class FixedText implements TextProvider {
    private String value;
    public FixedText(String value) {
        this.value = value;
    }
    @Override
    public String text() {
        return value;
    }
    public int textLength() {
        return value.length();
    }
}
TextProvider message = new FixedText("Open");
System.out.println(message.text());
```

This prints `Open`. The variable’s declared type is TextProvider. The actual object’s class is FixedText. Assigning the reference does not change the object’s class or create another object.

The caller can use the interface’s text operation through message. The concrete-only textLength method is not available through that declared type. This keeps the caller focused on the agreed capability. You may still use a separate FixedText reference when a task specifically needs FixedText operations.

### Let the actual object supply the behavior

A **polymorphic method call** uses the actual object to choose the instance-method implementation. One call in a loop can work with different implementing classes:

```java
interface Greeting {
    String words();
}
class MorningGreeting implements Greeting {
    public MorningGreeting() {
    }
    @Override
    public String words() {
        return "Good morning";
    }
}
class EveningGreeting implements Greeting {
    public EveningGreeting() {
    }
    @Override
    public String words() {
        return "Good evening";
    }
}
Greeting[] greetings = {new MorningGreeting(), new EveningGreeting()};
for (Greeting greeting : greetings) {
    System.out.println(greeting.words());
}
```

The output is `Good morning` followed by `Good evening`. Each loop iteration uses the same declared type and the same method call. The first reference reaches a MorningGreeting object; the second reaches an EveningGreeting object. Each object supplies its own words implementation. Each class explicitly declares a no-argument constructor with an empty body because this example needs no initial field values.

The program does not ask for a class name and then choose a separate branch. It asks each object to perform the operation that both classes promise. This lets a new implementation work with the existing caller when it honors the same contract.

### Share behavior in a default method

A **default interface method** has a body that implementing classes may inherit or replace. It can call a required operation on the same object:

```java
interface Labelled {
    String label();
    default String announcement() {
        return "Ready: " + label();
    }
}
class RoomLabel implements Labelled {
    private String room;
    public RoomLabel(String room) {
        this.room = room;
    }
    @Override
    public String label() {
        return "Room " + room;
    }
}
Labelled sign = new RoomLabel("West");
System.out.println(sign.announcement());
```

This prints `Ready: Room West`. RoomLabel supplies label and inherits announcement. The inherited body calls label on the RoomLabel object. The shared prefix appears in one place, while the object still decides its own label.

Adding a new required operation to an interface can make existing implementations incomplete. A default can provide shared behavior when that behavior makes sense for the contract. Defaults do not make every interface change safe; conflicting inherited defaults still require a design decision. This lesson uses one interface per class so that the operation choices stay clear.

## Worked Example

**Subgoal 1: declare the shared contract.** Labelled requires label and supplies the default announcement.

**Subgoal 2: implement the changing detail.** RoomLabel and CourseLabel keep their own private text and produce different labels.

**Subgoal 3: use one caller.** Traverse the Labelled array and request an announcement from each object.


````java
interface Labelled {
    String label();
    default String announcement() {
        return "Ready: " + label();
    }
}
class RoomLabel implements Labelled {
    private String room;
    public RoomLabel(String room) {
        this.room = room;
    }
    @Override
    public String label() {
        return "Room " + room;
    }
}
class CourseLabel implements Labelled {
    private String code;
    public CourseLabel(String code) {
        this.code = code;
    }
    @Override
    public String label() {
        return "Course " + code;
    }
}
Labelled[] signs = {new RoomLabel("Lab A"), new CourseLabel("CSC-239")};
for (Labelled sign : signs) {
    System.out.println(sign.announcement());
}
````

Expected output:

```text
Ready: Room Lab A
Ready: Course CSC-239
```

The first array element refers to a RoomLabel and the second to a CourseLabel. Both inherit the default announcement body. Its label call reaches the corresponding class implementation. The loop depends on Labelled, while each object keeps its own state and behavior.

## Summary

An interface states an operation contract. A class uses implements and matching public methods to fulfill it. An interface reference exposes the promised operations while still referring to one concrete object. Polymorphic calls use that object’s implementation. @Override checks the intended relationship, and a default method can share behavior that calls the required operation.
