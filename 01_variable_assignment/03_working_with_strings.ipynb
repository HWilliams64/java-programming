## Why This Matters

Applications use text for names, search requests, filenames, messages, and many other values. Entered text often needs preparation before it can be displayed or compared. Extra spaces can make a name look wrong on a badge or prevent an exact comparison from finding a match.

String operations let software prepare that text deliberately. Keeping the original while storing a cleaned result is useful when a program needs both the person's entry and a consistent form for later processing. Without that distinction, it is easy to call an operation and then accidentally keep using the unprocessed value.

Later modules will use strings in conditions, file records, and dictionary searches. These operations are building blocks for that work. They help you answer precise questions about text; they do not decide a person's identity or replace the application's rules.

## Concept

### Store the participant's entered name

The volunteer receives the sample text `"  Maya  "`: the name Maya with two ordinary spaces before it and two after it. We supply that text directly so everyone can repeat the same example. The program will display `Hello, Maya!`, report the cleaned name's length, and check whether its contents exactly match `Maya`. It does not read a live form or confirm anyone's identity.

A **String** is Java's ready-made type for text. The capital `S` is part of its name. It is a class, rather than one of the primitive types such as `int`. A String **object** holds text and provides named operations that work with it; later modules will teach how to design objects of your own.

```java
String enteredName = "  Maya  ";
```

This declaration introduces the variable `enteredName` and gives it a text value. All four surrounding spaces are inside the literal's quotation marks, so they belong to that value. Declaring the variable does not clean them away.


### Ask the text to perform an operation

A **method call** requests a named operation. To remove the ordinary spaces around the name, use:

```java
enteredName.trim()
```

The variable before the dot identifies the text to use. **`trim()`** names the operation, and the parentheses belong to the call. The empty parentheses mean that this call needs no additional input. Other methods may take an **argument**, a value supplied inside the parentheses; we will use one when comparing names below.

The call produces a **return value**, the result that the rest of the program can use. Here, that result is text without the surrounding spaces. Store it in another variable:

```java
String cleanName = enteredName.trim();
```

Java evaluates the method call on the right and stores its result in `cleanName`. This follows the same evaluate-then-store pattern as the arithmetic assignments in the previous lesson. The original variable and the cleaned result now have separate roles.


````java
String enteredName = "  Maya  ";
String cleanName = enteredName.trim();
System.out.println(enteredName);
System.out.println(cleanName);
````

The first output line still has two spaces before and after `Maya`; the second is `Maya` without them. Trailing spaces can be hard to see in console output, so the diagram marks the spaces explicitly.

Strings are **immutable**: an existing String object's text cannot be changed in place. `trim()` returns a result; it does not edit the original text or automatically replace `enteredName`. If no trimming is needed, Java may return the same String object. You can rely on the returned contents, without assuming that every call creates a different object.

Calling `enteredName.trim();` without storing or otherwise using its result leaves later reads of `enteredName` unchanged. You can instead write `enteredName = enteredName.trim();` to make the variable refer to the cleaned result. That assignment changes what the variable holds, rather than editing the original String text.

`trim()` removes leading and trailing characters whose values are at or below ordinary space; ordinary spaces are what matter in these examples. It leaves spaces between words in place, so trimming `"  Maya Chen  "` preserves the space between Maya and Chen. It is not a general cleaner for every spacing character or a rule for deciding how a person's name should be written.


### Join the name to a message

**Concatenation** means joining text values. In a string expression, Java's `+` operator can join a greeting, a stored name, and punctuation:

```java
String greeting = "Hello, " + cleanName + "!";
```

The first literal supplies `Hello, `, including the space after the comma. The unquoted name `cleanName` supplies its stored text. The final literal supplies the exclamation mark. The result is `Hello, Maya!`.

Java adds neither spaces nor punctuation on its own. `"Hello," + cleanName` would join the name directly to the comma. Putting `"cleanName"` in quotes would insert that word instead of the variable's value.


````java
String cleanName = "Maya";
String greeting = "Hello, " + cleanName + "!";
System.out.println(greeting);
````

The output is `Hello, Maya!`. The greeting combines the three supplied pieces in order; `cleanName` itself remains `Maya` afterward.

### Group a calculation before joining it to text

The same `+` symbol also adds numbers, so its operands matter. Imagine a volunteer counting two blue badges and three red badges. Compare these two displays:

```java
System.out.println("Total: " + 2 + 3);
System.out.println("Total: " + (2 + 3));
```

In the first expression, `"Total: " + 2` already makes text: `Total: 2`. The next `+` joins the displayed form of three to that text, producing `Total: 23`. It has not added the badge counts.

The parentheses in the second expression make the numeric addition happen first. Java adds two and three to get five, then joins that result to the label. It displays `Total: 5`, which represents the intended count. This reuses expression grouping from the arithmetic lesson.


````java
System.out.println("Total: " + 2 + 3);
System.out.println("Total: " + (2 + 3));
````

The two lines are `Total: 23` and `Total: 5`. Choose grouping according to whether you want to calculate a number first or join text pieces in sequence.

### Measure the cleaned name

The volunteer also wants to inspect the length of the text being displayed. **`length()`** returns a whole-number count of the String's storage positions:

```java
cleanName.length()
```

For the name `Maya`, each of its four letters occupies one position, so the result is `4`. Before trimming, the two spaces on each side also count, giving `8`. The return value is a number, not a shortened name, and the call does not change the text.

For names made from these English letters, the count matches the number of letters. Some symbols, such as certain emoji, occupy more than one Java text position, so `length()` is not a general count of the symbols a person sees.


````java
String cleanName = "Maya";
System.out.println(cleanName.length());
````

The output `4` reports the four positions occupied by `Maya`. It does not give the number of words or the last position number; we will distinguish position numbers from length below.

### Compare the contents with an expected name

A **string comparison** can check whether two text values have the same contents. Use **`equals(...)`** for this job:

```java
cleanName.equals("Maya")
```

Here, `"Maya"` is the argument: the expected text supplied inside the parentheses. The method returns the Boolean value `true` for an exact match and `false` otherwise. The spelling and letter case matter; `Maya` and `maya` are different text values.

Use `equals` to compare String contents. A **reference** is a value that identifies an object. The operator `==` checks whether two String references identify the same object, which is a different question. A small literal example can make `==` seem to work for text comparison, but that is not a reliable rule for the contents you want to compare.


````java
String cleanName = "Maya";
System.out.println(cleanName.equals("Maya"));
System.out.println(cleanName.equals("maya"));
````

The first output is `true` because every character matches. The second is `false` because the initial lowercase `m` differs from uppercase `M`. These checks inspect the text; neither one changes its capitalization.

### Select a numbered character position

A **character index** is the number of a position in Java text. Indexing starts at zero:

| Position number, or index | 0 | 1 | 2 | 3 |
| --- | --- | --- | --- | --- |
| Character in `Maya` | M | a | y | a |

**`charAt(index)`** selects the character at the supplied index. In `cleanName.charAt(0)`, the argument `0` chooses the first character. The method returns a `char`, Java's type for one text storage unit. We will print that result directly here.

The index must be at least zero and less than the string's length. A four-position string has indexes zero through three; index four would be outside it. `length()` and `charAt(...)` answer different questions: how many positions exist, and what is at one of those positions.


````java
String cleanName = "Maya";
System.out.println(cleanName.charAt(0));
System.out.println(cleanName.charAt(3));
````

The output is `M` and then `a`, from the first and last positions. The diagram uses both numbers and letter labels to show which position is selected. For an empty string, no character index is valid.

## Worked Example

### Step 1: preserve the entered name

```java
String enteredName = "  Maya  ";
```

The original text includes two spaces before and after the name. This variable gives us a place to keep that entry instead of immediately losing its original form.

### Step 2: store the cleaned result

```java
String cleanName = enteredName.trim();
```

The call produces text without those boundary spaces, and the assignment stores it in `cleanName`. The original `enteredName` still holds the padded text.

### Step 3: build the greeting

```java
String greeting = "Hello, " + cleanName + "!";
```

The greeting combines the supplied label, the cleaned name, and the exclamation mark. Each space and punctuation mark comes from one of those pieces.

### Step 4: display and inspect the result

```java
System.out.println(greeting);
System.out.println(cleanName.length());
System.out.println(cleanName.equals("Maya"));
```

The first statement displays the greeting. The second reports the cleaned length, and the third checks its exact contents. They provide three different pieces of information about the result. Run the complete example below.


````java
String enteredName = "  Maya  ";
String cleanName = enteredName.trim();
String greeting = "Hello, " + cleanName + "!";
System.out.println(greeting);
System.out.println(cleanName.length());
System.out.println(cleanName.equals("Maya"));
````

The result is:

```text
Hello, Maya!
4
true
```

The visitor's greeting has one intended space after the comma and no extra spaces around Maya. The cleaned name occupies four positions. The comparison is `true` because that cleaned text exactly matches the expected name. The program keeps the original entry separately; printing the new result does not erase it.

## Summary

String methods return results that a program can use or store. **Immutability** means the original text is not edited in place. A new assignment can change what a variable holds, and a separate variable can preserve the original while keeping the processed result.

**Concatenation** joins text pieces. **`length()`** reports size, **`equals(...)`** checks exact contents, and **`charAt(...)`** selects a valid numbered position. Parentheses help distinguish numeric addition from joining numbers to text.
