## Why This Matters

Programs need ways to communicate results. Text output can show an order total, give a user an instruction, or reveal an intermediate result while a programmer checks a calculation. A command-line interface lets people interact with software through text; printing messages is one part of building that kind of interface.

Without a useful display, a calculation can finish while the person using the program still cannot see its result. Other programs communicate through windows, files, or network messages. Here, text output gives us a direct way to see what our Java instructions do.

Later in this course, you will calculate with variables, choose messages with conditions, and process collections of data. You will keep using output to inspect those results. The wording and order matter now because a program can run successfully and still communicate the wrong thing.

## Concept

### A notice people can use

The campus help-desk staff want visitors to know that they have reached the right place and should have their student ID ready. We will supply the notice's wording directly in the code. The program does not read an ID, check a visitor's identity, or wait for a typed response.

Success means displaying exactly two lines: a welcome first, followed by the student-ID instruction. Capital letters, spaces, and periods belong to the required messages. These inputs are pieces of text, so there are no quantities or measurement units to calculate.

A **program** is a set of instructions a computer carries out. In our program, an output **statement** gives Java one instruction to display text. We choose the wording and the order; Java carries out those instructions when they run.


### Tell Java what to display

The first statement asks Java to display the welcome:

```java
System.out.println("Welcome to the help desk.");
```

The **string literal** is `"Welcome to the help desk."`: text written directly inside double quotation marks. Those marks tell Java where the text begins and ends. The space between words and the final period are inside the marks, so they belong to the message. The quotation marks themselves mark the text's boundaries and are not displayed by this statement.

**`System.out.println`** is a ready-made Java operation that displays a value and then ends the output line. Here, the value is our text. The opening and closing parentheses surround the value we want to display. The final semicolon ends the statement. Copy `System` with a capital `S`; `out` and `println` use lowercase letters.

This punctuation follows Java's **syntax**, its rules for writing code. `println` is a method name, meaning the name of an operation we can request; it is not a Java keyword. We will use methods in more detail later. For now, the important distinction is between the operation outside the quotes and the message inside them.


````java
System.out.println("Welcome to the help desk.");
````

The cell displays:

```text
Welcome to the help desk.
```

That displayed text is **console output**: text a running program sends to its output area. In this notebook, the output appears below the code cell. It contains the message, not the instruction `System.out.println` or the punctuation around the message.

Ending the line matters when another message follows. The next `println` starts its text on the next output line. We will use that behavior to keep the welcome separate from the action the visitor should take.

### Step 3: run the complete notice

The two statements now form our complete notebook program. For this straight sequence of statements, Java runs the first one before the second one. Run the cell and compare its result with the help desk's requirements.


````java
System.out.println("Welcome to the help desk.");
System.out.println("Please have your student ID ready.");
````

The result is:


The first statement produced the first line. Because `println` ended that line, the second statement placed its message below it. The two messages meet the staff's requirements: identify the desk, then tell the visitor what to prepare.

Running the same unchanged statements again produces the same messages. Java does not improve their wording or decide that a welcome belongs first. If we reverse the statements, their output order reverses too. To **trace** this program, follow each statement in order and record the line it produces.

### Recognize the same instructions in a source file

The video uses `Main.java`, a **source file**: a text file containing Java instructions. Here is the same notice inside the supplied structure for that file:

```java
public class Main {
    public static void main(String[] args) {
        System.out.println("Welcome to the help desk.");
        System.out.println("Please have your student ID ready.");
    }
}
```

We call the surrounding starter structure a **frame** here. That is a teaching description, not a special Java feature or a window. The brace pairs `{` and `}` mark where each part starts and ends. The indentation helps us see which instructions are inside which part.

The **`class`** keyword introduces a **class**, a named container for related Java code. Its name here is `Main`. The **main method**, named `main`, is the starting point for this application's instructions. Our two statements sit inside that method. `Main` and `main` are names with different capitalization; they are not the same name.

Three other keywords belong to this supplied frame. **`public`** makes the class and starting method accessible. **`static`** allows the starting method to run without first creating an object of `Main`. **`void`** says that the method does not return a value to the code that called it; it can still print text. `String[] args` provides a place for text supplied when an application starts. We do not use that input in this notice. You will study methods, objects, and arrays in later modules; keep this starter structure as supplied for now.

The notebook accepts our two output statements directly, so you do not need to copy the frame into its practice cells. A notebook **kernel** is the running language process that handles code sent from those cells.

**Compilation** checks and translates Java source into a form the Java runtime can execute. **Execution** is carrying out the instructions. A missing quotation mark can prevent the code from compiling, while correctly formed code can still print a message in the wrong order. That is why we check both the code and the actual output.

## Summary

A **program** gives a computer instructions. An output **statement** uses **`System.out.println`** to display a message and end its line. A **string literal** supplies text directly inside double quotation marks. The message appears as **console output**; the surrounding source punctuation tells Java how to interpret the instruction.

For the straight sequence you practiced, statement order determines output order. **Tracing** connects each instruction with its effect. **Debugging** repairs problems, and checking the output against the task catches mistakes that a successful run alone cannot detect.
