## Learning Goals

- Declare variables with types suited to counts, fractional values, and true-or-false facts.
- Evaluate arithmetic expressions, including the effect of grouping and operand types on division.
- Change an input with assignment, recalculate a dependent result, and explain both stored values.


## Why This Matters

Applications work with information that changes: available seats, ordered quantities, temperatures, and whether a service is open. Giving each value a name makes its purpose visible and lets several instructions use it. Choosing a matching type also helps prevent mistakes, such as trying to store half a ticket in a whole-number count.

Arithmetic turns those facts into useful results. An order system can multiply a quantity by a price, while a scheduling tool can divide people into complete groups and count those left over. Without named values, repeated numbers become hard to interpret and update consistently.

Later, conditions and loops will inspect and change variables. First, you need to understand when each value is stored. A calculated total is a result from a particular moment, so changing its input does not automatically keep that result up to date.

## Concept

### Give each fact a name and a suitable type

The campus club's sample order starts with three tickets at $4.50 per ticket, with sales open. Each ticket admits one person. We will calculate the cost of that initial order, then record a count of four to examine a later update. These values are supplied in the program; it does not sell tickets, take payment, or wait for keyboard input.

A **variable** is a named place for a value. A variable's **type** describes the kind of value it can hold. Java's built-in **primitive types** include the three needed for this record:

| Java type keyword | What it stores | Meaning in the club record |
| --- | --- | --- |
| **`int`** | A whole number | `3` tickets |
| **`double`** | A floating-point number, which can include a fractional part | `4.5` dollars per ticket |
| **`boolean`** | A true-or-false fact | `true` means sales are open |

`int`, `double`, and `boolean` are Java keywords. The values **`true`** and **`false`** are Boolean literals; they are not keywords or text. Numbers and Boolean values are unquoted here. The text `"3"` would not be the integer `3`.

### Introduce a variable and its starting value

```java
int tickets = 3;
```

The **declaration** introduces the type `int` and the name `tickets`. **Initialization** gives it the starting value `3`. The equals sign directs Java to store the value on the right in the variable on the left, and the semicolon ends the statement.

The name tells us what the number counts. Use descriptive names without spaces, such as `ticketPrice`, and preserve capitalization: `tickets` and `Tickets` are different names. An `int` cannot directly store `4.5`; its type is a whole-number type.

The price and sales status follow the same pattern. Run the cell to see the stored values.


````java
int tickets = 3;
double ticketPrice = 4.5;
boolean salesOpen = true;
System.out.println(tickets);
System.out.println(ticketPrice);
System.out.println(salesOpen);
````

The output is `3`, `4.5`, and `true`, on separate lines. In `System.out.println(tickets);`, the unquoted name tells Java to read the value stored in `tickets`. Printing `"tickets"` instead would display the word, as in the starting-point question.

### Calculate with the values stored now

An **arithmetic expression** is a calculation that produces a value. To find the cost of three tickets, multiply the count by the price of one ticket:

```java
double total = tickets * ticketPrice;
```

The operator `*` means multiplication. Its **operands**, the values the operator uses, come from `tickets` and `ticketPrice`. Java reads their current values, multiplies `3` by `4.5`, and stores `13.5` in `total`. The result is a cost in dollars because the calculation combines tickets with dollars per ticket.

The variable stores the calculated number. It does not store a live rule that repeats the multiplication whenever `tickets` changes. The diagram separates reading the operands, evaluating the expression, and storing its result.

### Update an input, then recalculate when needed

Suppose the count becomes four. **Assignment** can replace the value of the existing variable:

```java
tickets = 4;
```

We use the existing name without declaring its type again. Only `tickets` changes. The earlier value of `total` is still `13.5`. To calculate the cost for four tickets, run the multiplication again and assign its result:

```java
total = tickets * ticketPrice;
```

Now the operands are `4` and `4.5`, so the stored total becomes `18.0`. The next complete example prints `total` after changing the count and again after recalculating. That placement lets us observe the difference directly.


````java
int tickets = 3;
double ticketPrice = 4.5;
double total = tickets * ticketPrice;
tickets = 4;
System.out.println(total);
total = tickets * ticketPrice;
System.out.println(total);
````

The first printed line is `13.5`, even though `tickets` has already become four. The stored total has not been recalculated yet. The next assignment performs the multiplication again, so the second line is `18.0`.

This is also why rerunning just one cell can leave older values in a notebook. The example above recreates all of its starting values so its entire sequence can run together.

### Control the order of a calculation

Java uses `+` for addition, `-` for subtraction, `*` for multiplication, `/` for division, and `%` for remainder. **Operator precedence** determines which operation happens first when an expression uses several operators. Multiplication, division, and remainder take place before addition and subtraction. Parentheses let us group an operation that should happen first.

Consider two ways to calculate a supply count. One staff member combines two loose items with three packs of four items. Another needs four packs, each containing two standard items and three extra items:

```java
System.out.println(2 + 3 * 4);
System.out.println((2 + 3) * 4);
```

The first expression multiplies `3 * 4` before adding two, producing `14`. The second adds the five items per pack inside parentheses first, then multiplies by four packs, producing `20`. The grouping changes the meaning of the calculation, not just its appearance.


````java
System.out.println(2 + 3 * 4);
System.out.println((2 + 3) * 4);
````

The output is `14` and then `20`. Both expressions use the same numbers. Their grouping determines which intermediate result Java calculates first.

### Divide whole groups and find what remains

A volunteer has five supplies to divide equally between two tables. If each table must receive a whole number of supplies, Java's **integer division** gives the whole-number part of the quotient, or division result. With integer operands, `5 / 2` produces `2`. For positive quantities, that means two complete supplies per table.

The remainder operator `%` gives what is left after those complete groups: `5 % 2` produces `1`. Division and remainder answer different questions about the same grouping.

For a quantity that can be divided into fractions, such as five liters shared between two containers, `5.0 / 2` produces `2.5`. The `double` operand makes Java perform floating-point division. Assigning the result of `5 / 2` to a `double` afterward cannot restore the discarded fraction: it stores `2.0`.

Run these four checks and connect each result with the operation that produced it.


````java
System.out.println(5 / 2);
System.out.println(5.0 / 2);
double quotient = 5 / 2;
System.out.println(quotient);
System.out.println(5 % 2);
````

The results are `2`, `2.5`, `2.0`, and `1`. The operand types control the division before assignment stores the result. For negative integer operands, Java discards the fractional part toward zero; for example, `-5 / 2` produces `-2`.

### Recognize a limit of decimal arithmetic

A `double` has limited precision. Some decimal values are stored as nearby values rather than exactly; this is **floating-point approximation**. For example:

```java
System.out.println(0.1 + 0.2);
```

It displays `0.30000000000000004`, not exactly `0.3`. The difference comes from how these decimal fractions are represented and added. Our ticket prices `4.5` and `5.0` can be represented exactly, but that does not make `double` an exact choice for every financial calculation. Later software may need a deliberate representation for money. For now, recognize the limitation instead of deleting digits from an unexpected result without understanding it.


````java
System.out.println(0.1 + 0.2);
````

The output makes the approximation visible. This example is a limit to remember when interpreting a numeric result; it does not change the arithmetic rule or the chosen values in our ticket demonstration.

## Summary

**Variables** give values descriptive names, and **types** describe what those values can represent. Declaration and initialization introduce a variable with its starting value. **Assignment** replaces an existing value.

An **arithmetic expression** reads current operands and produces a result. Grouping and operand types affect the calculation. Integer division gives a whole-number quotient, remainder gives what is left, and `double` allows fractional values with limited precision.
