- Connect Scanner to standard input, display prompts, and read responses while a program runs.

## Why This Matters

A program becomes more useful when its users can supply information for each run. A workshop assistant can enter a new order without finding and editing values inside Java source code. The same calculation can then serve many attendees.

Typed input begins as text, even when its characters look like numbers. Reading a count as a number lets the program multiply it by a price. Reading a full name as a line keeps the space between its words. Choosing the right method protects the meaning of what the user enters.

This lesson connects input to the variables, String operations, calculations, and output you already know. Later, conditions can check values and loops can repeat questions. First, you need a clear agreement between each prompt and the response its reading method accepts.

## Concept

### Connect a reader to keyboard input

For the first workshop order, Maya Chen requests three `art` kits at $4.50 per kit and will collect them at the Student Center. The program needs all five details before it can display the receipt. It calculates the cost; it does not take payment.

An **input source** is the place a program reads information from. An **input stream** supplies data over time. Java's **standard input**, written `System.in`, is the stream we will use for typed responses. In a terminal you type at the running program; a notebook passes your response from its input control to the kernel.

A **library object** provides ready-made operations. **Scanner** is a Java library class whose objects read text as lines or smaller pieces and interpret numbers. As with String, we are using a class Java already provides.

Run the setup cell once before the other Java cells in this lesson.

````java
import java.util.Scanner;
Scanner keyboard = new Scanner(System.in);
````

The **`import`** keyword makes the short class name `Scanner` available to this code. It does not create a reader or read a response.

The next statement declares a Scanner variable named `keyboard`. The **`new`** keyword creates an object, and `System.in` supplies that object's input source. The assignment stores a reference to the reader in `keyboard`. Creating the reader does not ask for input yet, so this setup has no printed output.

The following cells share this one reader. Run them in order. If you restart the kernel, run this setup again before any reading cell. Do not create a separate Scanner for each notebook cell or close this reader during the lesson: multiple readers can compete for the same input, and closing it can close `System.in`.

### Ask for a response while the program runs

A **prompt** is a message telling the user what to enter. Printing a prompt and reading its response are separate steps:

```java
System.out.println("Full name:");
String customer = keyboard.nextLine();
```

The first statement displays the question. The second waits if a response is needed, reads it, and stores the returned text. These are illustrative lines; the runnable examples below use the same pattern.

For each runnable input cell, select Run, read its printed prompt, then enter your response in the notebook's input control and press Enter. Its input label may be blank; use the printed prompt to identify the requested value. Continue answering until the cell finishes before running another cell. In the terminal shown in the video, type the response directly after the prompt.

Enter the requested values as ordinary text. Do not type Java quotation marks around names, and do not replace values in the source code to supply a response. Enter submits a line ending, also called a **newline**, along with the typed text.

### Choose a whole line, one word, or a number

A **token** is one piece of input separated from others by a **delimiter**, a separator. Scanner uses **whitespace** as its default token delimiters: spaces, tabs, and line endings. Its token methods skip whitespace before a token, then stop at the token's end.

**Parsing** means interpreting text as a value of a chosen type. For example, `nextInt()` parses the typed characters `3` into the integer value three. The program can then calculate with that returned number.

| Call | Returned type | What it reads | Spaces and line endings |
| --- | --- | --- | --- |
| `keyboard.nextLine()` | `String` | The rest of the current line | Preserves its spaces and moves past its ending; the ending is not returned |
| `keyboard.next()` | `String` | The next token, such as `art` | Skips leading whitespace; stops before the whitespace after that token |
| `keyboard.nextInt()` | `int` | The next token interpreted as a whole number, such as `3` | Uses the same token boundaries; does not consume the following line ending |
| `keyboard.nextDouble()` | `double` | The next token interpreted as a floating-point number, such as `4.5` | Uses the same token boundaries; does not consume the following line ending |

A full name such as `Maya Chen` needs `nextLine()` to keep both words. A kit code such as `art` is one word, so `next()` matches it. If you enter `art extra`, `next()` returns only `art`; the space and `extra` remain for a later read. Both methods return a String, so the variable type alone cannot choose between them.

Use `nextInt()` for a whole-number quantity and `nextDouble()` for a price that may have a fractional part. Enter prices with a decimal point, such as `4.5`, and omit the dollar sign. A **locale** is a setting for conventions such as decimal separators. The course runtime used here accepts these dot-decimal inputs; a differently configured reader may use another convention.

### Handle the line ending after a token

A token read stops before the whitespace that follows the token. Another token read can skip that whitespace. A line read behaves differently: it returns the remaining text on the current line, even when that text is empty.

Run the following cell. At the first prompt, type `3` and press Enter. At the next prompt, type `Maya` and press Enter. The variable `exampleInput` gives the existing `keyboard` reader another name; it does not create a second reader.

After `nextInt()` reads `3`, the first `nextLine()` reads the **line remainder**, the text left before that line's ending. It returns empty text here and passes the ending without asking you for another response. The next `nextLine()` then reads the name you enter.

````java
Scanner exampleInput = keyboard;
System.out.println("Count:");
int count = exampleInput.nextInt();
String remainder = exampleInput.nextLine();
System.out.println("Name:");
String name = exampleInput.nextLine();
System.out.println(count);
System.out.println("[" + remainder + "]");
System.out.println(name);
````

After the prompts, the final three output lines are:

```text
3
[]
Maya
```

The brackets make an empty String visible as `[]`. They come from the print statement, not from your input. `remainder` means leftover text here; it is unrelated to the arithmetic remainder operator.

If you had typed `3 Library Desk` on the first line, its remainder would instead be ` Library Desk`, including the first space. Do not discard a line automatically after every numeric read: useful text may remain. In the workshop program, each response occupies its own line, so we can deliberately finish the price line before asking for the pickup location.

### Read responses in the requested order

**Input order** is the agreement between the values the program asks for and its sequence of reads. To focus on parsing before assembling the full workshop form, this smaller example asks for just a name, quantity, and price. Type `Maya`, `3`, and `4.5`, pressing Enter after each prompt.

The name read moves past its line ending. The quantity read stops after `3`; the price read skips that remaining newline before reading `4.5`. After displaying the result, the final `nextLine()` finishes the price line so the next notebook example starts on a new line.

````java
Scanner orderInput = keyboard;
System.out.println("Name:");
String customer = orderInput.nextLine();
System.out.println("Number of kits:");
int quantity = orderInput.nextInt();
System.out.println("Price per kit (use a decimal point):");
double price = orderInput.nextDouble();
double orderTotal = quantity * price;
System.out.println(customer + " owes " + orderTotal);
orderInput.nextLine();
````

The receipt line is `Maya owes 13.5`. The read methods returned a String, an int, and a double. Multiplying three kits by 4.5 dollars per kit gives 13.5 dollars. Each read advances through the responses; it does not reread an earlier answer. As in the boundary example, `orderInput` is another name for the shared reader.

### Match the input format to the method

`nextInt()` cannot parse the decimal token `4.5` as an integer. It reports an input mismatch error; it does not silently discard `.5`. Declaring the receiving variable as `double` does not change what `nextInt()` tries to read. Parsing must succeed before assignment can store the result.

Use the requested response format in this lesson: one response per prompt, whole-number counts, and dot-decimal prices. Words where numbers are expected can prevent the program from reaching its receipt. Handling invalid input is later work.

If an accidental invalid response stops a cell, restart the kernel, run the Scanner setup cell again, and retry the reading cell with the requested format. Restarting clears earlier variables as well as the interrupted input state.


## Worked Example

### Step 1: ask for the attendee and kit code

We can now build the complete workshop receipt. Maya Chen orders three `art` kits at $4.50 each, with pickup at the Student Center. The notebook setup already connected `keyboard` to `System.in`. These displayed snippets explain the program; run the complete cell after Step 4.

```java
System.out.println("Full name:");
String customer = keyboard.nextLine().trim();
System.out.println("Kit code (one word):");
String kitCode = keyboard.next();
```

Each print statement asks for the value read immediately afterward. `nextLine()` returns the whole name, including its internal space. Then `trim()` returns text with outside spaces removed, and the assignment stores that cleaned String in `customer`. The trim call belongs to the returned String, not to the Scanner. It does not alter the original String.

`next()` stores the next single-word token in `kitCode`. Use `art` as one word. Its line ending remains, but the next token read can skip it.

### Step 2: read the quantity and price as numbers

```java
System.out.println("Number of kits:");
int quantity = keyboard.nextInt();
System.out.println("Price per kit in dollars (use a decimal point):");
double price = keyboard.nextDouble();
```

The first prompt asks for a whole-number kit count. `nextInt()` skips preceding whitespace, parses that response, and returns the int stored in `quantity`. The next prompt states the price's unit and format. `nextDouble()` reads the price as a number that can include a fraction. Here the returned values are three kits and 4.5 dollars per kit.

### Step 3: finish the price line, then read the pickup location

```java
keyboard.nextLine();
System.out.println("Pickup location:");
String pickup = keyboard.nextLine();
```

The first statement passes the remainder of the price line and its ending. We do not store that result because our form requests only the price on that line. Without this step, the later line read would take the empty price-line remainder instead of waiting for a location.

The prompt then asks for the location, and the second `nextLine()` stores the whole response in `pickup`. `Student Center` contains a space, so `next()` would not capture the complete location.

### Step 4: calculate and display the receipt

```java
double total = quantity * price;
System.out.println(customer + ": " + quantity + " " + kitCode + " kits");
System.out.println("Total: $" + total);
System.out.println("Pickup: " + pickup);
```

The multiplication produces dollars: number of kits times dollars per kit. The first output line joins the name, quantity, and kit code. The next two display the cost and pickup location. The quoted text supplies the labels, spaces, punctuation, and dollar sign.

This cell contains all the receipt steps and uses the reader from the setup cell. In the video's standalone `Main.java`, the same steps are inside the supplied main method, after `Scanner keyboard = new Scanner(System.in);`.

````java
System.out.println("Full name:");
String customer = keyboard.nextLine().trim();
System.out.println("Kit code (one word):");
String kitCode = keyboard.next();
System.out.println("Number of kits:");
int quantity = keyboard.nextInt();
System.out.println("Price per kit in dollars (use a decimal point):");
double price = keyboard.nextDouble();
keyboard.nextLine();
System.out.println("Pickup location:");
String pickup = keyboard.nextLine();
double total = quantity * price;
System.out.println(customer + ": " + quantity + " " + kitCode + " kits");
System.out.println("Total: $" + total);
System.out.println("Pickup: " + pickup);
````
