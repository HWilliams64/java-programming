## Why This Matters

Programs need information that varies between uses: a visitor's name, an order quantity, or a measurement. Reading that information lets the same calculation serve different cases. A clear input format keeps a program from treating a name as a number or confusing a quantity with a price.

Input begins as text, even when its characters look like numbers. Interpreting that text correctly is what lets a program calculate with it. Without that step, a receipt might join the characters `3` and `4.5` instead of multiplying the values they represent.

Supplied text also makes testing repeatable: everyone can check the same record, then change one value and compare results. Later, conditions can check accepted values and loops can process repeated records. First, you need to know what each read consumes and returns.

## Concept

### Give the reader an input source

For our first workshop record, Maya requests three supply kits at $4.50 per kit. The source provides the name first, the quantity second, and the price third. A successful program will read those facts and display the amount Maya owes in dollars. These examples calculate the cost only; they do not take payment.

An **input source** is the place a program reads information from. Here it is text supplied inside the program. A **library object** provides ready-made operations. **Scanner** is a Java library class whose objects can read text and interpret numbers. As with String, we use a class Java already provides; designing our own classes comes later.

We need two setup statements:

```java
import java.util.Scanner;
Scanner orderInput = new Scanner("Maya\n3\n4.5\n");
```

An **import** lets this source code use a class's short name. The keyword **`import`** makes `Scanner` available instead of requiring its full name, `java.util.Scanner`, each time. This statement neither creates a reader nor reads data.

The second statement declares `orderInput` with type `Scanner`. The keyword **`new`** creates an object. In `new Scanner(...)`, the text inside the parentheses supplies the new reader's input source. Assigning the result to `orderInput` lets later calls use that reader. Keep the two setup statements together in each complete example.

### Put separate lines inside one literal

An **escape sequence** is source notation beginning with a backslash that represents a character such as a line break. Inside this Java string literal, **`\n`** represents a **newline**. Java turns that notation into a line break in the text value; the reader does not receive a backslash followed by the letter n.

The source literal `"Maya\n3\n4.5\n"` therefore provides three lines. Its last `\n` also ends the price line. The quotation marks delimit the literal in the source; they are not part of the input data.

| Input line | Text on that line | Meaning |
| --- | --- | --- |
| 1 | `Maya` | Attendee name |
| 2 | `3` | Kit quantity, still represented as text |
| 3 | `4.5` | Dollars per kit, still represented as text |

Run the setup cell below. It creates a reader but has no output statement and makes no read call, so it displays nothing. The following diagram shows the text given to that reader before any read occurs.

````java
import java.util.Scanner;
Scanner orderInput = new Scanner("Maya\n3\n4.5\n");
````

The cell has prepared `orderInput` at the beginning of its source. No name or number has been read yet. A blank output area here means that the program did not request printed output.

### Read a line or interpret a numeric token

You already used method calls on String values. A call such as `orderInput.nextLine()` requests an operation on this reader. Empty parentheses mean that this call takes no extra arguments.

A **delimiter** is a separator between pieces of input. A **token** is one piece between those separators. By default, Scanner's numeric reading methods use **whitespace** as delimiters: spacing characters such as spaces, tabs, and line endings. They skip whitespace before the next token.

**Parsing** means interpreting text as a value of a chosen type. The characters `3` become an `int` value when the reader successfully parses them. That gives multiplication a number it can use.

| Call | What it does here | Returned type |
| --- | --- | --- |
| `orderInput.nextLine()` | Reads the rest of the current line and moves past its ending; returns the text without that ending | `String` |
| `orderInput.nextInt()` | Skips leading whitespace and parses the next whole-number token | `int` |
| `orderInput.nextDouble()` | Skips leading whitespace and parses the next numeric token as a floating-point value | `double` |

Line reading preserves spaces within and around a name. A full name such as `Maya Chen` fits on one line even though a space divides its words. Numeric reading follows tokens, so it does not mean “read the entire next line.”

Use the supplied dot-decimal values, such as `4.5`. A **locale** is a setting for conventions such as decimal separators. These fixtures are checked with the course Workspace's reader settings; other settings can interpret numeric text differently. Locale configuration is outside this lesson.

### Read in the same order as the record

**Input order** is the agreement about which value comes next. Our source provides name, quantity, and price, so the calls must follow that order:

```java
String customer = orderInput.nextLine();
int quantity = orderInput.nextInt();
double price = orderInput.nextDouble();
```

This displayed snippet needs the reader setup shown above. The first line stores the returned name in `customer`. That call also moves past the name line's ending. The second parses `3` and stores an integer in `quantity`. It stops before the following newline. The third skips that newline, parses `4.5`, and stores a double in `price`. The final newline is still unread.

The reader remembers its position. A later call continues from there; it does not start the source again. Creating a new Scanner in a complete rerun starts a new reading at the beginning.

The complete cell below then multiplies quantity by dollars per kit to calculate dollars owed. Its final statement joins the name, the literal `" owes "` with spaces, and the numeric result. The worked section will revisit those last two lines step by step.

````java
import java.util.Scanner;
Scanner orderInput = new Scanner("Maya\n3\n4.5\n");
String customer = orderInput.nextLine();
int quantity = orderInput.nextInt();
double price = orderInput.nextDouble();
double orderTotal = quantity * price;
System.out.println(customer + " owes " + orderTotal);
````

The output is `Maya owes 13.5`. The reader returned the name, the integer `3`, and the double `4.5`. Multiplying those numeric values gives a cost of `13.5` dollars. The diagram now separates the reading steps from the calculation and output.

### Notice what remains after a numeric read

Now consider a different record: a count of three comes before the name Maya. The first read is `nextInt()`. After reading `3`, the reader is still before that line's ending. A following `nextLine()` reads the rest of this same line, which is empty in this particular source, and moves past its ending. One more `nextLine()` then returns Maya.

The variable `remainder` below stores that returned text, not an arithmetic remainder. We print square brackets around it so an empty result is visible as `[]`. Those brackets come from our output literals; they are not in the source. Run the complete example and compare its three results.

````java
import java.util.Scanner;
Scanner exampleInput = new Scanner("3\nMaya\n");
int count = exampleInput.nextInt();
String remainder = exampleInput.nextLine();
String name = exampleInput.nextLine();
System.out.println(count);
System.out.println("[" + remainder + "]");
System.out.println(name);
````

The output is `3`, `[]`, and `Maya`, on separate lines. The first line read after the number returned empty text because there were no characters between `3` and its newline. If other text followed the number on that same line, `nextLine()` would return that text instead. This behavior follows the source layout; it is not a rule that every numeric read always needs an extra line read.

### Use the text returned by a read

Reading a name does not trim it. We can apply the previous lesson's cleanup operation to the returned String. The two-step form makes that sequence explicit:

```java
String enteredName = nameInput.nextLine();
String cleanName = enteredName.trim();
```

This snippet needs a Scanner named `nameInput` with a name waiting in its source. The first statement reads the full line, including its outside spaces. The second stores cleaned text in a different variable. The original String's text stays unchanged.

Run the complete example. Brackets make the original outside spaces visible.

````java
import java.util.Scanner;
Scanner nameInput = new Scanner("  Maya Chen  \n");
String enteredName = nameInput.nextLine();
String cleanName = enteredName.trim();
System.out.println("[" + enteredName + "]");
System.out.println("[" + cleanName + "]");
````

The output is `[  Maya Chen  ]` and then `[Maya Chen]`. The space inside the name remains. Only its outside spaces were removed in the cleaned result.

When you only need the cleaned name, you can write the same read-then-trim sequence in one declaration:

```java
String cleanName = nameInput.nextLine().trim();
```

Use a newly created `nameInput` for this alternative. First `nextLine()` returns a String; then `trim()` operates on that String; finally the declaration stores the result. The call to `trim()` belongs to the returned text, not to the Scanner. Calling `nextLine()` again on the already consumed one-line reader would not repeat the earlier read.

### Match the input format to the read

`nextInt()` cannot parse the decimal token `4.5` as an integer. It reports an input mismatch error; it does not silently discard `.5`. Declaring the receiving variable as `double` does not change what `nextInt()` tries to read. The method must succeed before assignment can store its returned result.

Our examples supply enough values in a known format. Missing data, words where numbers are expected, and an unexpected read order can prevent a read from succeeding. Handling such errors is later work. For this lesson, inspect the next input value and choose the matching method.

### Recognize another input source

An **input stream** supplies data for a program to read over time. Java's **standard input** stream is `System.in`; an interactive terminal commonly connects it to typed responses. The source can change while the Scanner reading methods stay the same:

| Source supplied to Scanner | Where its data comes from | What to expect |
| --- | --- | --- |
| `"Maya\n3\n4.5\n"` | Fixed text already supplied by the program | These examples do not wait for typing |
| `System.in` | The program's standard input stream | A read can wait for an entered response |

For example, `Scanner keyboardInput = new Scanner(System.in);` connects a reader to standard input. A later `keyboardInput.nextLine()` would read a line from that source. This is a display-only alternative, not an instruction to run a waiting read in this notebook. Notebook input behavior depends on the kernel and frontend; keep the supplied-text source for these exercises.

Later, file reading will use the same separation between a source and reading operations. Do not close a Scanner around `System.in` during a shared session: closing that reader can also close standard input. Resource management will be taught later.
