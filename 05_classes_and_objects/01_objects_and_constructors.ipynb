## Why This Matters

Applications often manage many things of the same kind: customer accounts, equipment loans, or items in an order. Each needs its own values, but many follow the same rules. A class lets you keep a rule in one definition and apply it to the right object's state.

Without that organization, adding another account or changing an update rule can mean maintaining more unrelated variables and copied calculations. Objects give larger programs a consistent way to group related data and operations. This prepares you to protect those values in the next lesson and later build collection classes and interactive applications.

## Concept

### Keep each visitor's information together

A campus welcome desk issues a pass to each visitor. For this fictional example, a pass stores the visitor's name and the number of recorded visits. Each new pass starts at zero. Recording one visit adds one to that pass's count. The program should report an owner's name with the count belonging to that owner.

Names are String values; visit counts are whole numbers stored as `int`. We use small nonnegative counts that fit in `int`. We are modeling the records in code, rather than scanning a physical pass or checking a real person's identity.


### Define the fields and their starting values

A **class** defines the stored values and operations for a kind of object. The Java keyword **`class`** starts its definition. The braces group the class body:

```java
class WelcomePass {
    String owner;
    int visits;
}
```

`WelcomePass` is the class name. The two declarations inside the body describe what each pass stores. An **instance field** is a variable stored separately in each object. `owner` holds that pass's name, while `visits` holds that pass's count. Field declarations appear inside the class but outside any method or constructor body.

An **object instance** is one object created from a class. Its **state** is the set of values it currently holds; its **behavior** is what its operations can do. The class definition describes the design. It does not, by itself, create a pass for a visitor.

We also need a reliable starting state. An explicit **constructor** supplies the initialization steps that run when an object is created. Add this constructor inside the class:

```java
WelcomePass(String owner) {
    this.owner = owner;
    this.visits = 0;
}
```

The constructor's name matches the class name. Its parentheses declare a String parameter named `owner`. Unlike a method header, a constructor header has no return type. Writing `void WelcomePass(...)` would define a method, rather than this constructor.

The Java keyword **`this`** refers to the current object inside its constructor or instance method. In `this.owner = owner`, the left side names that object's field. The right side names the temporary parameter received by this constructor call. The assignment stores the supplied name in the object. The next line explicitly sets that object's visit count to zero.

The matching names do not mean one shared variable: a constructor parameter supplies an input for this call, while a field retains a value in the object afterward. A **reference** is a value that lets the program reach an object. String values are reached through references too; this assignment does not create a new copy of the name's characters.

Now create a pass and keep a way to use it:

```java
WelcomePass pass = new WelcomePass("Zara");
```

The left side declares `pass` as a variable that can refer to a WelcomePass. The **`new`** keyword starts the expression that creates a new object and invokes its constructor with `"Zara"`. During that call, `this` refers to the new pass. After the constructor finishes, the new expression yields the reference assigned to `pass`. The constructor itself does not return a value.

Finally, `pass.owner` and `pass.visits` read fields from the object reached through `pass`. The dot selects a member of that object. The complete example includes a `visit` operation too; the next section explains how its body updates state.

These introductory fields are accessible to the surrounding lesson code so the reports can read them directly. We will introduce `private` fields and controlled public operations in the next lesson. For now, use the demonstrated methods to change counts.


````java
class WelcomePass {
    String owner;
    int visits;
    WelcomePass(String owner) {
        this.owner = owner;
        this.visits = 0;
    }
    void visit() { this.visits = this.visits + 1; }
}
WelcomePass pass = new WelcomePass("Zara");
System.out.println(pass.owner + ": " + pass.visits);
````

Expected output:

```text
Zara: 0
```

The constructor stores Zara's name and explicitly sets this new object's count to zero. No visit has been recorded. The report reads both values from the same object. The animation separates the temporary constructor parameter from the fields that remain afterward.

### Direct the shared operation to one object

The constructor provides a starting count. We now need to change that count when a visit occurs. An **instance method** is a method called for a particular object. The object receiving the call is its **receiver**. Our pass defines this operation:

```java
void visit() {
    this.visits = this.visits + 1;
}
```

`visit` is the method's name. The empty parentheses mean this method has no explicit parameters. The Java keyword **`void`** means it returns no result. Its useful effect is an update to the receiving object's state.

The right side reads the receiver's current `visits` field and adds one. The assignment then stores that new total back in the same field. It uses the previous count instead of replacing the count with one on every call. The field remains part of the object after the method finishes.

Unlike the static calculations from Module 3, this method is not declared `static`. Call it through an object reference, such as `east.visit()`. The reference before the dot selects the receiver, so `this` refers to that object during the call. `WelcomePass.visit()` would not identify which pass this instance method should update.

To see why the receiver matters, create two passes with two separate new expressions:

```java
WelcomePass east = new WelcomePass("Uma");
WelcomePass west = new WelcomePass("Ravi");
east.visit();
east.visit();
west.visit();
```

Each constructor initializes a different pass. The first two calls select the object reached through `east`; the last selects the object reached through `west`. A method call uses an existing object. It does not run the constructor again or create another pass.

**Do not repeat yourself (DRY)** means keeping a shared rule in one definition instead of copying its implementation for each use. There is one `visit` method definition here. It supplies the increment rule for both passes, while each object keeps its own field values. Sharing a rule does not merge the records.


````java
class WelcomePass {
    String owner;
    int visits;
    WelcomePass(String owner) {
        this.owner = owner;
        this.visits = 0;
    }
    void visit() { this.visits = this.visits + 1; }
}
WelcomePass east = new WelcomePass("Uma");
WelcomePass west = new WelcomePass("Ravi");
east.visit();
east.visit();
west.visit();
System.out.println(east.owner + ": " + east.visits);
System.out.println(west.owner + ": " + west.visits);
````

Expected output:

```text
Uma: 2
Ravi: 1
```

Both counts begin at zero. Uma's pass receives two calls and changes from zero to one to two. Ravi's pass receives one call and changes from zero to one. The print statements read the two objects in that order. In the animation, follow the receiver label as each call selects the object to update.

### Keep a lasting field separate from a temporary input

A fixed increment works for visits. Other records need the amount supplied by each call. For example, a reader may record six pages at once. We will use a ReadingCounter with a `pages` field initialized to zero and compare two methods that receive a numeric parameter also named `pages`.

The first method changes only its temporary input:

```java
void adjustLocal(int pages) {
    pages = pages + 1;
}
```

Inside this method, the unqualified name `pages` selects its parameter. In `log.adjustLocal(6)`, that parameter starts with the supplied value six. The assignment changes the local value to seven. Nothing in this body assigns the object's field, so that field remains zero after the call. This is a worked comparison of two different effects; changing the local input does not record pages.

The recording method instead updates the object:

```java
void record(int pages) {
    this.pages = this.pages + pages;
}
```

`this.pages` on the right reads the stored total. The unqualified `pages` reads this call's supplied amount. Their sum becomes the new value assigned to the field on the left. For a stored zero and an argument of six, the field becomes six and remains six when the method finishes.

The complete class also has `ReadingCounter() { this.pages = 0; }`. The empty constructor parentheses mean no argument is needed; the body still explicitly sets the starting total. Calling `new ReadingCounter()` creates a fresh counter. Both methods use `void` because neither returns a value, but only `record` performs the lasting state change needed by a page log.

Use this distinction whenever an input has the same name as a field: the input describes this one call, and the field stores the ongoing record. Our page amounts are nonnegative and their totals fit in `int`; these simple methods do not validate other inputs.


````java
class ReadingCounter {
    int pages;
    ReadingCounter() { this.pages = 0; }
    void adjustLocal(int pages) { pages = pages + 1; }
    void record(int pages) { this.pages = this.pages + pages; }
}
ReadingCounter log = new ReadingCounter();
log.adjustLocal(6);
System.out.println("After local: " + log.pages);
log.record(6);
System.out.println("After field: " + log.pages);
````

Expected output:

```text
After local: 0
After field: 6
```

The first report reads a field that the parameter-only assignment never changed. The second report reads the field updated by `record(6)`. The animation follows the temporary parameter and stored total separately. This prepares you to diagnose a page-log method later without guessing which occurrence of a name needs `this`.

## Worked Example

### Build the welcome desk's two-pass report

The desk now needs records for Maya and Luis. Each new pass must start with zero visits. Staff record two visits for Maya and one for Luis, then print each owner's name and count. We will assemble a DeskPass definition, create the two records, and apply the same update rule to the correct receiver.


### Define the record before creating its users

Start the class with `String owner;` and `int visits;`. These are the two fields that every DeskPass instance needs. In `DeskPass(String owner)`, assign `this.owner = owner;` and `this.visits = 0;`. The constructor connects the incoming name to the new object's field and establishes the starting count.

Next, define `void visit()` with `this.visits = this.visits + 1;`. This keeps the update rule in the class. The caller will select a pass and request a visit, rather than copying the increment arithmetic into every call site.

### Create the records, then apply the rule

```java
DeskPass first = new DeskPass("Maya");
DeskPass second = new DeskPass("Luis");
```

The two new expressions create separate objects. `first` reaches Maya's pass and `second` reaches Luis's. Their constructors use different owner arguments but apply the same zero-count starting rule.

```java
first.visit();
first.visit();
second.visit();
```

The first two lines update Maya's count. The third updates Luis's. The report can then combine each object's `owner`, a colon and space, and its `visits` value. The complete cell below keeps the definition, fresh objects, calls, and report together.


````java
class DeskPass {
    String owner;
    int visits;
    DeskPass(String owner) {
        this.owner = owner;
        this.visits = 0;
    }
    void visit() {
        this.visits = this.visits + 1;
    }
}
DeskPass first = new DeskPass("Maya");
DeskPass second = new DeskPass("Luis");
first.visit();
first.visit();
second.visit();
System.out.println(first.owner + ": " + first.visits);
System.out.println(second.owner + ": " + second.visits);
````

Expected output:

```text
Maya: 2
Luis: 1
```

Each new expression creates a separate pass and initializes its visits to zero. Calls through first update only Maya’s pass; the call through second updates Luis’s pass. The same method definition supplies both objects’ increment rule.

The report has one line per pass. It shows each owner's stored total, rather than a combined count across all visits. Rerunning this complete cell creates fresh passes before applying the same calls. Running only an extra call would update an existing pass instead.

## Summary

A class defines a kind of object's fields and operations. Each object created by a separate new expression has its own instance fields. A constructor establishes its starting state; an instance method acts on the receiver selected by the caller.

Inside a constructor or instance method, `this` identifies the current object. It makes the distinction between a field such as `this.minutes` and a same-named parameter clear. Assigning only a temporary parameter does not update the object's field.

One method definition can supply a shared rule for many objects. That applies DRY while preserving separate records, rather than giving every object one combined total.

A new expression creates an object and invokes its constructor; the expression yields the reference that the caller can store. The constructor's assignments establish starting state. Calling an instance method through a reference selects its receiver; the call does not itself create a new receiver. The method bodies in this lesson update the existing object.

A field remains stored in the object after a call. A parameter is an input for that particular call.
