# Arrays and Traversal

In this lesson, you will learn to store a fixed sequence in an array, update its elements, and traverse its values to produce a report.

A campus tour coordinator needs an ordered list of stops. You will replace a changed stop without disturbing the others, then print the route. Later, you will use the same idea of visiting each entry to summarize supplies in storage bins.

Use your earlier work with variables, String values, conditions, and `for` loops. The new part is storing several related values in one structure. The [Module 4 glossary](terms.md) collects the terms introduced across these lessons.


## Learning Goals

- Create and update a fixed-length array using valid indices.
- Trace indexed and enhanced `for` loops, including an empty array.
- Compute a total and a count from array values, and explain what each result measures.


## Why This Matters

Programs often receive a fixed batch of related data: daily measurements, a route's stops, or quantities counted during a stock check. Separate variables for every entry would require separate processing statements. That makes it harder to apply the same rule consistently and easier to leave an entry out.

An array groups the values so one loop can process the whole batch. Its position numbers also let a program replace a particular entry. These skills prepare you to use growing collections, process records from files, and choose a structure that fits the data. Choosing a structure well helps keep both updates and reports correct as a program develops.

## Concept

### Store a Fixed Sequence

The tour coordinator has three stops in visiting order: the desk, the lab, and the hall. The lab visit moves to the studio. The rule is to replace that stop in the same position; the route must still have three stops. A successful report shows the new route in order and ends with its entry count.

A **data structure** organizes values for the operations a program needs. An **array** is one such structure: a sequence with a fixed number of positions. Each stored value is an **element**. Here, every element is a String naming a stop. The order matters because it tells the guide where to go next.


### Create the positions and their starting values

This declaration creates the route:

```java
String[] locations = {"desk", "lab", "hall"};
```

`String[]` means an array whose elements hold String values. The brackets belong to the type, so `locations` names the whole array rather than a single stop. As in an ordinary variable declaration, `=` supplies its starting value.

The **array initializer**, `{"desk", "lab", "hall"}`, lists the three starting elements in order. Commas separate the elements; quotation marks make each stop name a String literal. These braces enclose starting data, rather than a loop or method body. The semicolon finishes the declaration.

For this array, the indices are `0`, `1`, and `2`. An **index** identifies a position; it does not describe the length of the text stored there. The first position contains `"desk"`, the second contains `"lab"`, and the third contains `"hall"`.

The array's length is fixed when it is created. Replacing a stop changes the value in a position. It does not insert another position or remove one. We will use a different collection in the next lesson when the number of entries must change.


### Read and replace one stop

**Array element access** uses the array name followed by an index in brackets. `locations[1]` reads the second stop because indexing starts at zero. A print statement can display that returned String.

The same indexed expression can be an assignment target:

```java
locations[1] = "studio";
```

The right side supplies the replacement String. The left side selects the existing position that receives it. This does not edit the characters of `"lab"`; it changes which String value the array element holds.

The **array length** is its number of positions. Read it with `locations.length`, without parentheses. `length` is a field supplied by arrays, not a method or a Java keyword. The complete example below prints the selected stop before and after replacement, then prints the length.


````java
String[] locations = {"desk", "lab", "hall"};
System.out.println(locations[1]);
locations[1] = "studio";
System.out.println(locations[1]);
System.out.println(locations.length);
````

The output is `lab`, then `studio`, then `3`, on separate lines. Both reads select index `1`. The assignment between them changes that element, while `"desk"` and `"hall"` keep their positions. The final `3` confirms that replacing one value did not change the number of positions.

An index must already exist. Using `locations[3]` would attempt to access a position beyond this array. Java reports an invalid array access at that statement; it does not automatically grow the array.

### Visit every valid position

Reading one element is useful for a particular update. A route report needs all of them. **Traversal** means visiting the elements according to a rule. An indexed `for` loop can visit each position once, in order.

```java
for (int index = 0; index < places.length; index = index + 1) {
    System.out.println(index + ": " + places[index]);
}
```

The **`for` keyword** introduces the loop you studied earlier. `int index = 0` starts at the first valid position. The test `index < places.length` allows another visit only while an element exists at that index. The body reads that element and prints its index followed by its stop name. Afterward, `index = index + 1` advances one position.

The strict `<` is essential. With three elements, index `2` is valid, but index `3` is not. After printing the third element, the update reaches `3`; the next test is false, so the body does not try to read a fourth element. Run the complete example to see this order.


````java
String[] places = {"desk", "lab", "hall"};
for (int index = 0; index < places.length; index = index + 1) {
    System.out.println(index + ": " + places[index]);
}
````

The report prints `0: desk`, `1: lab`, and `2: hall`. These are three visits to three elements. The condition is checked a fourth time, when `index` is `3`, but there is no fourth body execution.

The same strict bound handles an empty array. An empty initializer, `{}`, creates zero positions. The first test would be `0 < 0`, which is false. There is no first element to read, and the loop correctly performs no element access.

### Get each value with an enhanced loop

Sometimes the work needs each value but does not need its index. An **enhanced for loop** supplies the array's successive values to a local variable:

```java
for (int amount : amounts) {
    System.out.println(amount);
}
```

Here `amounts` is an `int[]`. The declaration `int amount` describes one integer element, not the whole array. The colon separates that declaration from the array being traversed. Java supplies the next element value before each body execution; you do not write an index update in this form.

The element variable is **loop-local**: its name is available within that loop body, and each iteration receives the current element's value. For an integer array, changing this local integer does not write a value back into the array. The assignment target matters, just as it did when a method changed its numeric parameter without changing the caller's variable.

The complete comparison below starts with amounts `2` and `4`. Its loop adds one to each local value and prints it. After the loop, it reads the stored elements directly. This lets us compare the local results with the array that supplied them.


````java
int[] amounts = {2, 4};
for (int amount : amounts) {
    amount = amount + 1;
    System.out.println("Local: " + amount);
}
System.out.println("Array: " + amounts[0] + ", " + amounts[1]);
````

The two loop lines are `Local: 3` and `Local: 5`. The final line is `Array: 2, 4`. On each visit, `amount = amount + 1;` changes `amount`, the local integer. It does not target `amounts[0]` or `amounts[1]`, so those stored values remain unchanged.

Use an indexed assignment when the task is to replace an array element. Use the enhanced form when the task is to read each value and compute or report something from it. This distinction also applies to reassigning a String loop variable: assigning that name another String does not replace the source array entry.

### Summarize amounts instead of only printing them

A stockroom assistant records how many supply units are in each of three bins: `2`, `0`, and `4`. Each array entry represents one bin; its integer is the number of units in that bin. The assistant needs two results: the total number of units and the number of bins that contain a positive amount.

A **running total** keeps the sum processed so far. Declare `int total = 0;` before the loop. Inside the loop, `total = total + quantity;` combines the earlier sum with the current bin's amount. Putting the declaration inside the loop would restart the sum on every visit. Assigning only `quantity` would discard the earlier amounts.

A **conditional count** measures how many entries meet a rule. Start `int nonempty = 0;` before the loop. The **`if` keyword** introduces the test `quantity > 0`; only a positive quantity increases `nonempty` by one. We add one because we are counting bins, not their units.

The two updates serve different purposes. Every amount contributes to `total`, while the condition controls whether that bin contributes to `nonempty`. A zero contributes no units and is not positive. The example prints a trace after each visit in the form `quantity -> total, nonempty`; the arrow is just text in the report.


````java
int[] quantities = {2, 0, 4};
int total = 0;
int nonempty = 0;
for (int quantity : quantities) {
    total = total + quantity;
    if (quantity > 0) {
        nonempty = nonempty + 1;
    }
    System.out.println(quantity + " -> " + total + ", " + nonempty);
}
````

The trace is:

```text
2 -> 2, 1
0 -> 2, 1
4 -> 6, 2
```

The first bin adds two units and one qualifying bin. The second adds zero units; its positive test is false, so neither result changes. The third adds four units to the earlier two and adds one qualifying bin. The final results are six units in two nonempty bins, across three bins inspected.

Neither result is initialized by reading a first element. For an empty array, the enhanced loop supplies no values, and both variables retain zero. That is the correct starting result when no bins have been supplied.

### Guard a last-element update

Traversal already checks its boundary before reading an element. A single update outside a loop needs its own check. The last valid index is `length - 1` only when the array is nonempty. For length zero, that expression would be `-1`, which is not a valid index.

The example below begins with no labels. Its `if` statement checks whether an element exists before replacing the last one. The print after the guard still runs, whether the array is empty or not. The initializer `{}` is an empty array of Strings; it is different from `{""}`, which contains one empty String.


````java
String[] labels = {};
if (labels.length > 0) {
    labels[labels.length - 1] = "updated";
}
System.out.println("Entries: " + labels.length);
````

The output is `Entries: 0`. Because `labels.length > 0` is false, the body never attempts an indexed replacement. If the initializer contained one label, its length would be one and the guarded replacement index would be zero. Replacing it would keep that same length.

One more count distinction matters when array elements contain text. `labels.length` counts entries. For a nonempty array, `labels[0].length()` would first select a String, then call the String's `length()` method to count its character positions. Only the String call uses parentheses. An empty String has zero character positions, but it still occupies one array entry.


## Worked Example

### Update and Report the Tour

Now connect the three route operations into one complete program. First, create the coordinator's ordered route. Next, assign `"studio"` to index `1` to replace the lab visit. Finally, traverse the positions to print the revised route, then print the entry count after the loop.

The replacement must happen before traversal. Otherwise, the report would show the earlier lab stop. The final count belongs after the loop so the program prints one count for the complete route, rather than repeating it after every stop.


````java
String[] locations = {"desk", "lab", "hall"};
locations[1] = "studio";
for (int index = 0; index < locations.length; index = index + 1) {
    System.out.println(index + ": " + locations[index]);
}
System.out.println("Entries: " + locations.length);
````

The complete report is:

```text
0: desk
1: studio
2: hall
Entries: 3
```

Index `1` now contains the studio, while indices `0` and `2` retain their original stops. Each printed index labels the entry the loop actually reads. After the update reaches index `3`, the false condition ends traversal. The last statement reports three entries, meeting the coordinator's rule that this change must preserve the route's size.

## Summary

An array keeps a fixed sequence of elements. Square brackets select a position, and its `length` field counts the positions. Replacement changes a stored value while preserving that count. An indexed loop provides positions for reading or replacement; an enhanced loop supplies successive element values for processing.

A running total adds amounts, while a conditional count adds one for each qualifying entry. Initializing those results before traversal allows earlier work to accumulate and gives useful results even when the array is empty.

For a nonempty array, valid indices start at zero and end at `length - 1`. The length itself counts the positions and is just beyond the last valid one. An empty array has length zero and no valid element index.

With an empty array, an indexed traversal using `index < array.length` fails its first condition at `0 < 0`. An enhanced loop receives no element values. Neither executes its body.

The enhanced-loop integer variable receives the current element's numeric value. Assigning that local variable a different integer changes the local value only. An indexed assignment such as `amounts[0] = 9;` targets a stored element and requires that index to exist. Confusing those assignment targets leads to a program that prints adjusted local values while leaving the array unchanged.
