# Sets and Unique Values

In this lesson, you will learn to recognize duplicate values with a set and build a report in first-seen order.

A campus supply desk receives repeated requests. The complete request sequence still matters, but a preparation report should list each kind of supply once. You will give a set and a list different jobs so the program can remember what it has already encountered and preserve a useful report order.

Build on String equality, arrays, enhanced loops, and the ArrayList operations from the previous lesson. The [Module 4 glossary](terms.md) supports the terms explained here.


## Learning Goals

- Use HashSet membership and change results to identify duplicate requests.
- Build a unique result list in the order values first appear.


## Why This Matters

Applications often receive the same value more than once. A service may receive repeated search terms, a supply log may mention the same item on several lines, or a report may need one entry for each distinct category. Keeping the full input is useful, but counting every occurrence as a new category would give the wrong result.

A set gives a program a direct way to remember membership. An ordered list serves a different purpose: preserving a sequence for display. Choosing those roles deliberately prevents a report from depending on an order its storage structure never promised. This extends your work with filtered lists and prepares you to use a map when each distinct item needs associated information, such as a request count.

## Concept

### Remember which kinds of supplies have appeared

The supply desk receives a request for a map, another request for a map, and a request for a kit. Each String names a kind of supply, rather than a quantity or a particular physical object. For the preparation report, two equal item names represent one kind. The desired count is two kinds: map and kit.

A **set** stores at most one representative of each equal value. **Membership** means whether a value is present. A **unique value** here is a distinct String value counted once, even if the input mentions it several times. The rule is equality of text contents, using the String equality behavior you already know. It is not a rule about how many requests arrived.

**HashSet** is the Java library class used for this set. Unlike ArrayList, it does not give entries numbered positions that your program can read with `get(index)`. You ask about a value itself. The full source log can keep repeated requests while this separate set records which kinds have appeared.


### Create an empty set for String values

The desk needs a collection before it can remember an item. This setup follows the import and construction pattern used for ArrayList:

```java
import java.util.HashSet;
HashSet<String> items = new HashSet<String>();
```

The keyword **`import`** makes the short class name `HashSet` available from `java.util`. The keyword **`new`** creates an empty HashSet object; `items` refers to it. Importing a class does not itself create a set. A fresh construction starts with no members.

The explicit type argument `<String>` appears on both sides of the declaration. It says that this set's elements are Strings. These angle brackets supply type information; they do not compare values. `HashSet` and `String` are class names, not Java keywords. A call such as `items.add("map")` supplies an actual String argument after that element type has been chosen.

The method call `items.size()` returns the number of distinct members currently present. It returns zero just after this construction. Like a list's `size()`, it uses parentheses. An array's `length` is a field, while a String's `length()` counts character positions. None of those counts is the number of calls the program has made.


### Separate a membership change from its result

A **change result** is a boolean that reports whether an operation changed the set. The set's `add` method has two related effects: it stores the argument if no equal value is present, and it returns whether a new member was added.

```java
items.add("map");
```

This snippet continues the empty setup above. The map is absent, so this call stores it and returns `true`. A statement can call a method without displaying its returned result. In that form, the set changes but no output line appears.

A print statement can use the result of the call:

```java
System.out.println("Duplicate map: " + items.add("map"));
```

After the earlier add, an equal map is already present. This second call keeps the existing membership and returns `false`. Java joins that boolean result to the label and prints `Duplicate map: false`. The result is normal duplicate handling, not an error. Unlike appending to an ArrayList, adding an equal value to this set does not create another entry.

The **`contains` method** asks whether an equal value is present. It returns a boolean without adding or removing anything. After adding a kit, `items.contains("kit")` returns `true`. Asking about an absent pen returns `false`; the question does not make pen a member.

The complete program below starts with its own empty set. Its first print combines the first map add with its change result. The next print attempts the duplicate. It then adds kit without printing that call's result, queries kit's membership, and reports the distinct count.


````java
import java.util.HashSet;
HashSet<String> items = new HashSet<String>();
System.out.println("First map: " + items.add("map"));
System.out.println("Duplicate map: " + items.add("map"));
items.add("kit");
System.out.println("Contains kit: " + items.contains("kit"));
System.out.println("Unique: " + items.size());
````

The output is:

```text
First map: true
Duplicate map: false
Contains kit: true
Unique: 2
```

The first call adds map, so its result is true and the size becomes one. The duplicate call returns false and leaves the size at one. Adding kit creates the second member. The contains call reads that membership without changing it, and size reports two. These labels describe operation results; they are not positions or a traversal of the set.

### Remove a value and report whether it was present

A lending desk has recorded one borrowed lamp. When it removes that item from its set, it needs to distinguish a change from a repeated request to remove an item that is already absent.

The **`remove` method** accepts the value to remove from a HashSet. This differs from the indexed ArrayList removal you used earlier. `borrowed.remove("lamp")` selects an equal String value; it does not select a numbered position.

```java
System.out.println("First removal: " + borrowed.remove("lamp"));
System.out.println("Second removal: " + borrowed.remove("lamp"));
```

These two lines assume a set named borrowed that contains lamp. The first call removes its equal member and returns `true`. The second finds no equal member to remove and returns `false`. An absent-value removal is a normal operation. The second call does not remove a different value or produce a negative size.

The complete example first creates its own set and adds lamp. After the two removal attempts, it asks whether lamp remains and reports the size. Those later queries let you check the resulting state separately from the change results.


````java
import java.util.HashSet;
HashSet<String> borrowed = new HashSet<String>();
borrowed.add("lamp");
System.out.println("First removal: " + borrowed.remove("lamp"));
System.out.println("Second removal: " + borrowed.remove("lamp"));
System.out.println("Contains lamp: " + borrowed.contains("lamp"));
System.out.println("Unique: " + borrowed.size());
````

The four lines are `First removal: true`, `Second removal: false`, `Contains lamp: false`, and `Unique: 0`.

Only the first removal changes membership. The second removal and the later contains query both return false for different reasons: one cannot remove an absent value, while the other reports that the value is absent. Neither restores the lamp. If the set started empty, both removal results would be false and the final size would still be zero.

### Use a hash as a lookup aid, not an equality test

The class name HashSet refers to **hashing**, which uses a number derived from a value to help locate where to look for that value. A **hash code** is that integer lookup aid. For a String, the method `hashCode()` supplies the number; you do not have to calculate it or write the method.

A hash code is not a set index or a guaranteed unique identifier. Equal Strings have equal hashes, but different Strings can also share a hash. A **hash collision** occurs when unequal values have the same hash code. HashSet still uses equality to distinguish those values.

The next check uses two short String values, `"Aa"` and `"BB"`. Their contents differ, even though both hash codes are `2112`:

```java
String first = "Aa";
String second = "BB";
System.out.println(first.hashCode());
System.out.println(second.hashCode());
System.out.println(first.equals(second));
```

The first two print statements report the integer hashes. The third compares the actual text contents and prints false. Matching the two integers would therefore be the wrong test for equal text. The complete program below also adds both Strings to a set and reports its size.


````java
import java.util.HashSet;
String first = "Aa";
String second = "BB";
System.out.println(first.hashCode());
System.out.println(second.hashCode());
System.out.println(first.equals(second));
HashSet<String> values = new HashSet<String>();
values.add(first);
values.add(second);
System.out.println(values.size());
````

The program prints `2112`, `2112`, `false`, and `2`, on separate lines. The equal numbers demonstrate a collision. The false equality result shows that Aa and BB remain distinct String values. Both survive in the set, so its size is two.

Hashing helps lookup; it does not replace equality. It also does not protect secret text or assign a unique identity to every possible String. A matching hash gives no reason to discard a value whose contents differ from the existing member.

### Keep membership and report order in separate collections

The supply desk now needs an ordered preparation report from the request array `{"map", "kit", "map", "pass"}`. A pass is another kind of supply. The rule is to print each kind once, in the order its first request appeared. That report should be map, kit, and pass, followed by a distinct count of three.

**Iteration order** is the sequence in which a traversal visits stored values. HashSet has **unspecified iteration order**: it does not promise insertion order, alphabetical order, or another order your report can rely on. A particular run can look ordered without establishing a guarantee. Drawing the set's members beside one another is only a way to show which values are present.

**First-seen order** follows the first appearance of each distinct value in the input. Give the structures separate jobs: the array supplies the complete request sequence, a set named `seen` remembers membership, and a list named `firstSeen` records the accepted report sequence. Create both collections once before traversing the array:

```java
HashSet<String> seen = new HashSet<String>();
ArrayList<String> firstSeen = new ArrayList<String>();
```

These declarations require the two corresponding imports. Both collections start empty. The names last across the loop because the declarations precede it; recreating either collection inside the loop would discard earlier work.

The enhanced loop supplies each String request in array order. Use the set add's change result to control the list append:

```java
for (String request : requests) {
    if (seen.add(request)) {
        firstSeen.add(request);
    }
}
```

The keyword **`for`** introduces the traversal, and the colon separates the local element declaration from its source. The keyword **`if`** selects the append only when add returns true. The condition calls add once: it both remembers a first-time value and supplies the boolean decision. A repeated equal value returns false, so the body is skipped. Calling add before this condition and then calling it again would make even a first-time value look repeated on the second call.

For map, the condition is true and the list gains map. Kit also produces true and appends kit. The later map produces false and leaves the list alone. Pass produces true and appends pass. This keeps one representative of each repeated value, a process called **deduplication**.

After the input loop ends, a separate enhanced loop prints the list in its stored sequence. It does not traverse the set. `seen.size()` then reports how many kinds were accepted. The complete program puts the imports, input, two collections, filtering loop, and report together.


````java
import java.util.ArrayList;
import java.util.HashSet;
String[] requests = {"map", "kit", "map", "pass"};
HashSet<String> seen = new HashSet<String>();
ArrayList<String> firstSeen = new ArrayList<String>();
for (String request : requests) {
    if (seen.add(request)) {
        firstSeen.add(request);
    }
}
for (String request : firstSeen) {
    System.out.println(request);
}
System.out.println("Unique: " + seen.size());
````

The output is:

```text
map
kit
pass
Unique: 3
```

The four input requests produce set-add results true, true, false, and true. Those three true results cause three list appends. The repeated map remains part of the original array but does not create a second list entry. The list provides the report order; the set provides membership and the distinct count.

For an empty source array, neither enhanced loop runs. Both collections stay empty and the final line is `Unique: 0`. For several copies of one value, only the first add is true. Starting the collections before traversal handles both cases without reading an element that might not exist.

### Distinguish a new request from a repeated request

Sometimes a program should label every request as it arrives, rather than save a separate result list. A desk log with `"desk"`, `"lab"`, and `"desk"` can use the same single add result for that choice.

In the next complete program, a true condition prints a New line. The keyword **`else`** provides the alternative branch: when the add result is false, that branch prints Repeat. Exactly one branch prints for each input value. The set remembers membership, while the array traversal determines the order of these immediate lines.

Using `seen.contains(request)` alone would only read membership. Starting with an empty set and never adding anything would leave every request absent. The add call is needed because this task both remembers new values and reports whether they were new.


````java
import java.util.HashSet;
String[] requests = {"desk", "lab", "desk"};
HashSet<String> seen = new HashSet<String>();
for (String request : requests) {
    if (seen.add(request)) {
        System.out.println("New: " + request);
    } else {
        System.out.println("Repeat: " + request);
    }
}
System.out.println("Unique: " + seen.size());
````

The output is `New: desk`, `New: lab`, `Repeat: desk`, and `Unique: 2`, on separate lines. The third request finds an equal desk already present, so only the else branch runs for that visit. Two kinds are present after three requests. The New/Repeat report follows the input array and makes no claim about HashSet traversal order.
