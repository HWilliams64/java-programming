# ArrayList and Changing Size

In this lesson, you will learn to add, replace, remove, and filter list entries while keeping their order.

A campus volunteer coordinator needs to update a sign-up list as requests arrive. You will keep repeated names as separate entries, correct a name, and remove a canceled entry. You will also build a smaller result from selected values in an array.

Use the array indices and loop patterns from the previous lesson. The new structure can change its number of entries. The [Module 4 glossary](terms.md) supports the terms explained here.


## Learning Goals

- Use ArrayList methods to add, read, replace, and remove text elements.
- Convert selected array entries into a list while preserving order.


## Why This Matters

Applications often work with a sequence whose final size is not known when it starts. A sign-up form receives another request; a task list loses a completed item; a report keeps only records that meet a rule. The program needs to change the collection while keeping the relationship between entries and their positions clear.

An array gives you a fixed number of positions. An ArrayList supplies operations for growing and shrinking an ordered sequence. Without an appropriate collection, each change would require more manual storage and position bookkeeping. Building a separate filtered result also keeps the complete input available for another report. These choices prepare you to compare lists, sets, and maps according to the work an application requires.

## Concept

### Keep a sequence that can change size

The volunteer coordinator receives three sign-up entries: Maya, Luis, and Maya again. The program records submitted names, not verified identities, so repeated text remains as a separate entry. A correction changes the second entry to Nora. Then the first entry is canceled. The report must show the remaining entries in their current order and give their count.

An **ArrayList** is a Java library class for an **ordered collection**: a group whose elements occupy a sequence of positions. Its number of elements can grow or shrink. A **duplicate** is another entry equal to an earlier value; an ArrayList keeps it. Order does not mean alphabetical sorting. The sequence reflects the additions and later changes you request.

As with an array, the first element has index zero. A position identifies the current entry at that location. After removal, the same number can refer to a different entry because later entries shift toward the beginning.


### Create an empty list with an element type

The coordinator needs a list before any entries can be added. These two statements prepare that list:

```java
import java.util.ArrayList;
ArrayList<String> names = new ArrayList<String>();
```

The keyword **`import`** makes the library class name `ArrayList` available. As in your Scanner work, the keyword **`new`** creates an object. Here it creates an empty list, and `names` refers to that list. The empty parentheses do not supply any initial names. Each fresh construction begins with zero entries.

ArrayList is a **generic class**: its use can specify the kind of elements it holds. A **type argument** is a type written inside angle brackets. Both occurrences of `<String>` specify String elements in this declaration. We keep both explicit forms throughout this lesson. `String` is a class name; `ArrayList` and `String` are not Java keywords. The angle brackets here provide type information, rather than comparing two values.

This type argument differs from a method-call argument. In `names.add("Maya")`, `<String>` has already established the element type, while `"Maya"` supplies the actual value to store. Java checks the element type; an integer such as `7` is not a String name for this list.

The **list size** is the number of elements currently stored. `names.size()` returns that count, so it returns `0` just after construction. It is a method call with parentheses. An array uses `values.length` without parentheses; a String uses `text.length()` for its character-position count. Those expressions answer different questions. A list's size is not its last valid index.


### Append, read, replace, and remove

**List insertion** adds a new element. The form of `add` used here **appends** its argument: it places the new value after all current entries.

```java
names.add("Maya");
names.add("Luis");
names.add("Maya");
```

These lines continue the empty-list setup above. The first call creates entry 0. The next two create entries 1 and 2, so the size becomes three. Both Maya entries remain. No call sorts the names or merges equal text.

**Indexed list access** reads an element with `get`. The expression `names.get(1)` returns the second name, Luis, without changing the list. Use this method call rather than array brackets such as `names[1]`.

**List replacement** uses `set` to store a new value at an existing position:

```java
names.set(1, "Nora");
```

The first argument chooses index 1; the second supplies the replacement. Luis becomes Nora. The size stays three because no position was added. `set` cannot create the first entry of an empty list.

**List removal** deletes an existing entry. In this lesson, `remove` always receives an integer index:

```java
names.remove(0);
```

That call removes the entry currently at index 0, the first Maya. Nora shifts from 1 to 0, and the remaining Maya shifts from 2 to 1. The size becomes two. The call does not remove every equal name, and the old index 2 is no longer valid.

| Operation | Purpose | Effect on size |
| --- | --- | --- |
| `names.add("Maya")` | Append one new entry | Increase by one |
| `names.get(1)` | Read the current second entry | No change |
| `names.set(1, "Nora")` | Replace the current second entry | No change |
| `names.remove(0)` | Delete the current first entry | Decrease by one |
| `names.size()` | Read the current count | No change |

For `get`, `set`, and the indexed `remove`, the index must already exist: at least zero and less than the current size. These method names come from the library; they are not language keywords. Run the complete example below to follow the same changes with printed checks.


````java
import java.util.ArrayList;
ArrayList<String> names = new ArrayList<String>();
names.add("Maya");
names.add("Luis");
names.add("Maya");
System.out.println(names.size());
names.set(1, "Nora");
System.out.println(names.get(1) + ": " + names.size());
names.remove(0);
System.out.println(names.get(0) + ", " + names.get(1) + ": " + names.size());
````

Expected output:

```text
3
Nora: 3
Nora, Maya: 2
```

The first line confirms three appended entries. The second reads index 1 after replacement and pairs Nora with the unchanged count. The last line reads the two current positions after removal. It shows both the shift and the smaller size. Every printed check follows the operation it describes.

### Read a sequence and keep selected entries

Before selecting values, connect the two traversal patterns to a list. The following complete examples use the same two names. The indexed form includes position labels:

```java
import java.util.ArrayList;
ArrayList<String> readers = new ArrayList<String>();
readers.add("Maya");
readers.add("Nora");
for (int index = 0; index < readers.size(); index = index + 1) {
    System.out.println(index + ": " + readers.get(index));
}
```

It prints `0: Maya` and `1: Nora`. Initialize the index once, test it against the current size, read that position, and then advance. When index 2 fails `2 < 2`, no `get` call follows. This uses the same strict-bound reasoning as array traversal, with list methods for the count and access.

When only values matter, the enhanced loop is shorter:

```java
import java.util.ArrayList;
ArrayList<String> readers = new ArrayList<String>();
readers.add("Maya");
readers.add("Nora");
for (String reader : readers) {
    System.out.println(reader);
}
```

It prints Maya and Nora without position labels. The keyword **`for`** introduces the loop. `String reader` declares the current element variable, and the colon separates that declaration from the source list. Each iteration receives the next String value in list order. The source list is not being added to or removed from during either report.

Now a tour coordinator has an array of proposed places: desk, lab, desk, and studio. Desk entries represent a check-in point that should be omitted from the tour itself. The program must keep the other places in order while retaining the complete original array.

**Filtering a sequence** means building a result from entries that meet a condition. Create an empty result once, visit every source value, and append only values that pass the rule:

```java
ArrayList<String> selected = new ArrayList<String>();
for (String place : places) {
    if (!place.equals("desk")) {
        selected.add(place);
    }
}
```

This snippet uses the source array `places`, declared in the complete example below. The enhanced loop receives one source String at a time. Its **`if`** statement checks the keep rule: `equals` compares text contents, and the logical NOT operator `!` changes equal-to-desk into not-equal-to-desk. A desk value makes the keep condition false, so no append occurs. Lab and studio each make it true.

The source and result have separate jobs. The loop reads the array and changes only the result list. Appending accepted values as they are encountered preserves their relative order. If an accepted value appears twice, both occurrences would be appended. No matching source values are deleted from the array. This lesson filters into a separate result rather than adding to or removing from the sequence currently being traversed.


````java
import java.util.ArrayList;
String[] places = {"desk", "lab", "desk", "studio"};
ArrayList<String> selected = new ArrayList<String>();
for (String place : places) {
    if (!place.equals("desk")) {
        selected.add(place);
    }
}
for (String place : selected) {
    System.out.println(place);
}
System.out.println("Source first: " + places[0]);
````

Expected output:

```text
lab
studio
Source first: desk
```

The first desk is skipped, lab is appended, the second desk is skipped, and studio is appended. The second enhanced loop prints the selected list in that order. The last statement reads the first source element, which is still desk. Filtering selected values for the report has left the original four-element array available.

### Check whether a position exists before replacing it

Filtering does not guarantee a nonempty result. The source might be empty, or every entry might fail the keep rule. An empty list has size zero and no element at index 0. A program must not call `get(0)` or `set(0, ...)` just because zero is normally the first index.

The coordinator may want to correct the first selected stop. A **guard** is a condition that allows an operation only when its required state is present. The familiar condition `stops.size() > 0` checks whether at least one entry exists:

```java
if (stops.size() > 0) {
    stops.set(0, "studio");
}
```

For an empty list, `0 > 0` is false and replacement is skipped. For a one-entry list, `1 > 0` is true and index 0 exists. Replacement then keeps the size at one.

The next complete example compares those two states. It first checks an empty list. It then appends lab, repeats the same guard, and reads the corrected value. Its last `get(0)` is valid because the earlier append established that entry and no removal followed.


````java
import java.util.ArrayList;
ArrayList<String> stops = new ArrayList<String>();
if (stops.size() > 0) {
    stops.set(0, "studio");
}
System.out.println("Empty count: " + stops.size());
stops.add("lab");
if (stops.size() > 0) {
    stops.set(0, "studio");
}
System.out.println(stops.get(0));
````

Expected output:

```text
Empty count: 0
studio
```

The first guard skips its body, so the count stays zero. Appending lab creates index 0 and changes the size to one. The second guard allows `set` to replace lab with studio. The final read returns studio; replacement has not added a second entry.

### Keep repeated matches and recognize an empty result

A campus service records requests for map, kit, and map. It needs a list of map requests so it can keep the two separate requests in their received order. This time the keep rule is positive: `request.equals("map")`, without `!`.

Start the result before traversal. A new list inside the loop would discard the result from earlier iterations. In the complete example below, each matching request calls `add`, while kit makes the condition false. Printing the result size counts retained requests, not distinct request names.


````java
import java.util.ArrayList;
String[] requests = {"map", "kit", "map"};
ArrayList<String> maps = new ArrayList<String>();
for (String request : requests) {
    if (request.equals("map")) {
        maps.add(request);
    }
}
System.out.println(maps.size());
````

Expected output is `2`. The first map is appended, kit is skipped, and the second map is appended. Both entries count even though their text is equal.

With an empty source initializer `{}`, the source loop would have no iterations. With a source containing only kit, it would visit that entry but append nothing. Both paths leave the result empty. A later guarded replacement must check the result's size, because even a nonempty source can produce no retained values.


## Worked Example

### Update the volunteer list and report its current order

Return to the volunteer coordinator's complete change request: receive Maya, Luis, and Maya; correct the second entry to Nora; then cancel the first entry. The duplicate names remain separate submitted entries. The final report needs current index labels and the number of entries still present.

**Subgoal 1: collect the entries.** Import ArrayList, construct an empty String list, and append the three supplied names. The order is Maya at 0, Luis at 1, and Maya at 2.

**Subgoal 2: apply the changes in order.** `set(1, "Nora")` replaces Luis without growing the list. Then `remove(0)` deletes the first Maya and shifts the later entries. Performing those operations in their stated order matters because removal changes positions.

**Subgoal 3: report the current positions.** The indexed loop uses `names.size()` as its exclusive upper bound and `names.get(index)` to read the current entry. The print after the loop reports the final count. Run this full example to connect the individual operations.


````java
import java.util.ArrayList;
ArrayList<String> names = new ArrayList<String>();
names.add("Maya");
names.add("Luis");
names.add("Maya");
names.set(1, "Nora");
names.remove(0);
for (int index = 0; index < names.size(); index = index + 1) {
    System.out.println(index + ": " + names.get(index));
}
System.out.println("Names: " + names.size());
````

Expected output:

```text
0: Nora
1: Maya
Names: 2
```

After insertion, the names are Maya, Luis, Maya. Replacement changes only Luis to Nora. Removing index 0 deletes the first Maya, so Nora moves to 0 and the remaining Maya moves to 1. Two entries remain, in that order.

The print loop accepts indices 0 and 1. Its update then makes index 2, and `2 < 2` is false. No read occurs at index 2. The final size line counts the remaining entries, while the first two lines identify their positions and values.

## Summary

An ArrayList stores a sequence whose size can change. Its type argument identifies the element type. Appending grows the sequence; replacement changes an existing value; indexed removal deletes one entry and shifts later positions. Reading and traversal must use current valid indices. A separate filtered result preserves the source and the relative order of retained entries.

<details>
<summary>Show answer</summary>

`add` appends one entry and increases size; `remove` with an integer index deletes one entry and reduces size. `set` replaces an existing value, while `get` and `size` read information. Those three do not change the number of entries.

After indexed removal, each later entry moves one position toward the start. For example, removing index 0 from Maya, Nora leaves Nora at index 0 and size 1. Reusing the old index 1 would no longer be valid.

A filter appends only entries that pass its condition. If every source entry fails, the result stays empty even though the source loop ran. Checking the result size before indexed access handles this case as well as an empty input.
