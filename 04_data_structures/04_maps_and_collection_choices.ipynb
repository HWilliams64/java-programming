# Maps and Collection Choices

In this lesson, you will learn to associate keys with values, count repeated requests, and combine collections for an ordered report.

A campus supply desk needs both a quantity for each kind of supply and a summary of repeated requests. A set can answer whether a kind has appeared, but the report also needs its count. You will give a map and an ordered list different jobs while keeping the original request sequence.

Build on arrays, enhanced loops, String equality, ArrayList, and the set-based first-seen report from the previous lesson. The [Module 4 glossary](terms.md) supports the terms explained here.


## Learning Goals

- Read and update key/value entries without assuming missing keys have numeric values.
- Combine a list and map to count requests and report them in first-seen order.
- Explain a choice among an array, list, set, and map for an extraction task.


## Why This Matters

Applications often need information identified by a meaningful name: a quantity for a supply, a total for a category, or the number of requests for a service. Separate variables become awkward when the names come from the input. A map lets one lookup rule work for each key.

A useful report may also need to preserve the order in which categories appeared. Counting and ordering are separate requirements. Combining structures lets each handle a clear job, rather than relying on an order the map never promised. This completes your comparison of arrays, lists, sets, and maps, and prepares you to design classes whose state supports particular operations.

## Concept

### Associate each supply name with its quantity

The supply desk records two maps and one kit. A later count corrects the map quantity to three. Here `"map"` names a kind of printed supply; it is also one key inside the map data structure. The result should report three maps, one kit, and two kinds of supply.

A **map** associates each distinct **key** with a **value**. The key identifies an entry, such as the String `"map"`. The value is its associated information, such as quantity 2. A **map entry** is that key/value pair considered together. The key is not an integer position. Two different keys may have equal values.

**HashMap** is the Java library class used here. It uses hashing and equality to organize the keys, as HashSet does for membership. Equal String keys identify the same association even when supplied through different variables. Matching hashes alone do not make unequal keys equal. HashMap also makes no promise of first-seen traversal order.


### Create a map for names and whole-number values

The desk needs an empty map before it can store quantities:

```java
import java.util.HashMap;
HashMap<String, Integer> stock = new HashMap<String, Integer>();
```

The keyword **`import`** makes the short class name available from `java.util`. The keyword **`new`** constructs an empty HashMap. The variable `stock` holds the reference used to call its methods. Importing the class does not create an object.

The two **type arguments** describe the key type first and the value type second. `String` names the key type. **Integer** is Java's object form for a primitive `int` number. It is a **wrapper type**: a library class whose objects represent values of a corresponding primitive type. Java's generic type arguments use reference types, so write `Integer`, not the keyword `int`, inside these angle brackets.

A **reference value** refers to an object; it is different from directly storing a primitive number. The class name `Integer` is capitalized, while the primitive keyword `int` is lowercase. You do not define Integer yourself. When the following examples pass an int quantity to this map, Java supplies its Integer form for storage. We will trace that conversion explicitly after the first map example.

The map starts empty. `stock.size()` counts distinct keys currently recorded. It does not add their quantities or count the number of method calls.


### Add an association, then replace its recorded value

The **`put` method** stores a value under a key:

```java
stock.put("map", 2);
stock.put("kit", 1);
stock.put("map", 3);
```

These snippets continue the empty stock setup. The first statement creates the map association with quantity 2. The second adds a distinct kit key with quantity 1. There are now two keys.

The third statement performs **map replacement**. An equal map key already exists, so its associated quantity becomes 3. The previous value 2 is replaced; it is not added to 3, and no third key is created. These calls do not print anything. A call can change stored data without displaying its returned result.

A **lookup** retrieves information by its key. The **`get` method** performs that lookup:

```java
System.out.println("Maps: " + stock.get("map"));
System.out.println("Kits: " + stock.get("kit"));
System.out.println("Types: " + stock.size());
```

Each get asks for the current associated value. Java joins the displayed result to its label, then prints the line. The final size is a key count. These explicit print statements determine report order; they do not traverse the HashMap.

The complete example below includes its own import and empty construction before the three updates.


````java
import java.util.HashMap;
HashMap<String, Integer> stock = new HashMap<String, Integer>();
stock.put("map", 2);
stock.put("kit", 1);
stock.put("map", 3);
System.out.println("Maps: " + stock.get("map"));
System.out.println("Kits: " + stock.get("kit"));
System.out.println("Types: " + stock.size());
````

The report is:

```text
Maps: 3
Kits: 1
Types: 2
```

The map quantity changed from 2 to 3. The kit quantity stayed 1, and both keys remain present. Three puts produced two entries because the third put replaced an existing value. The sum of the quantities would be four; size reports two because it answers a different question.

### Follow the number between primitive and object forms

A count update needs arithmetic, so it matters what a map lookup returns. The next complete program starts with primitive `int amount = 5` and stores it under the map key.

**Boxing** converts a primitive value to its corresponding object form. In `stock.put("map", amount)`, Java boxes the number for an Integer-valued entry. The variable amount still has declared type int; it does not become an Integer variable.

```java
Integer saved = stock.get("map");
```

The get returns an Integer reference, which saved holds. Its displayed text is 5, but saved is not a primitive int variable. It refers to an Integer object representing that number.

**Unboxing** obtains a primitive value from an object such as an Integer. The expression `saved + 1` unboxes the stored five for addition, producing primitive int six. Assigning that result to `int nextAmount` stores the primitive result. The final put boxes six for the replacement value.

This example first stores a numeric value under the requested key, so saved refers to an Integer and can supply a number. The next section examines what happens when the requested key has no entry.


````java
import java.util.HashMap;
HashMap<String, Integer> stock = new HashMap<String, Integer>();
int amount = 5;
stock.put("map", amount);
Integer saved = stock.get("map");
System.out.println("Saved: " + saved);
int nextAmount = saved + 1;
stock.put("map", nextAmount);
System.out.println("Next: " + stock.get("map"));
System.out.println("Types: " + stock.size());
````

The three lines are `Saved: 5`, `Next: 6`, and `Types: 1`. The first print displays saved before the replacement. Unboxing supplies 5 for the addition; boxing supplies the new Integer value 6 for storage. Replacing that value keeps the map's one key. No operation turns saved itself into an int variable or changes the declared type of amount.


### Check presence before using a missing lookup as a number

The next map records the map supply with quantity zero. That is a real entry: the desk has counted that kind and recorded no items available. Kit has no entry yet. These situations need different treatment.

A **key-presence check** uses `containsKey` to ask whether an entry exists. It returns a boolean without adding a key or changing its value. `containsKey("map")` is true even when the recorded quantity is zero.

When get finds no entry for a key, it returns the **null reference**, written with the Java literal `null`. Null refers to no object; it is not numeric zero and is not the String `"null"`. Displaying a null reference in a joined message prints the text null. That display does not obtain a primitive int from it. Unboxing null cannot supply a number and would fail.

Our examples store only numeric values and never deliberately store null. Under that rule, a present key has an Integer that can be used for arithmetic. HashMap can store null values in general, so presence alone would not prove a usable number in every possible map.

```java
if (counts.containsKey("kit")) {
    counts.put("kit", counts.get("kit") + 1);
} else {
    counts.put("kit", 1);
}
```

The keyword **`if`** selects the existing-key branch when the presence check is true. In that branch, get supplies a recorded Integer; Java unboxes it for addition and boxes the new result for put. The keyword **`else`** selects initialization when no kit entry exists. It stores one without first trying to add to a missing value.

This example does not change the map between checking a key and using its value. The complete cell first shows the difference between present zero and absent kit, then applies the guard.


````java
import java.util.HashMap;
HashMap<String, Integer> counts = new HashMap<String, Integer>();
counts.put("map", 0);
System.out.println(counts.containsKey("map"));
System.out.println(counts.get("map"));
System.out.println(counts.containsKey("kit"));
System.out.println(counts.get("kit"));
if (counts.containsKey("kit")) {
    counts.put("kit", counts.get("kit") + 1);
} else {
    counts.put("kit", 1);
}
System.out.println(counts.get("kit"));
````

The first four lines are true, 0, false, and null: map is present with recorded zero, while kit is absent. The false kit check chooses the else branch, which creates kit with count 1. The last line prints 1.

The present-zero case and the missing case have different presence results. A condition based only on the numeric value would lose that distinction. In these no-null-values examples, the guard prevents arithmetic on an absent lookup.

### Count occurrences while recording first-seen order

The desk now counts requests rather than available stock. The array `{"map", "kit", "map"}` contains three requests in arrival order. Each String represents one request for that kind of supply. The report needs a frequency for each kind and must print kinds in the order they first appeared.

A **frequency table** associates each observed value with its number of occurrences. A HashMap named counts keeps those numbers. An ArrayList named firstSeen keeps the distinct report sequence. The array retains the full request log, including repetition. Create both collections before traversing the array so earlier work survives each loop iteration.

```java
if (counts.containsKey(request)) {
    counts.put(request, counts.get(request) + 1);
} else {
    counts.put(request, 1);
    firstSeen.add(request);
}
```

For the first map request, no entry exists. The else branch starts its count at one and appends map to the list. Kit is also new, so it gets count one and the next list position.

For the repeated map request, the if branch looks up Integer one, unboxes it, adds one to get primitive two, then boxes two for replacement. The existing key remains present. This branch does not append to firstSeen, so the list still contains map and kit once each.

After the input loop, traverse the ordered list:

```java
for (String item : firstSeen) {
    System.out.println(item + ": " + counts.get(item));
}
```

Each list entry supplies a key whose final count is fetched from the map. The list gives report order, and the map gives each frequency. A set alone could remember that map occurred, but it would not store the number two beside it. The complete program below puts these steps together.


````java
import java.util.ArrayList;
import java.util.HashMap;
String[] requests = {"map", "kit", "map"};
HashMap<String, Integer> counts = new HashMap<String, Integer>();
ArrayList<String> firstSeen = new ArrayList<String>();
for (String request : requests) {
    if (counts.containsKey(request)) {
        counts.put(request, counts.get(request) + 1);
    } else {
        counts.put(request, 1);
        firstSeen.add(request);
    }
}
for (String request : firstSeen) {
    System.out.println(request + ": " + counts.get(request));
}
````

The program prints `map: 2` and then `kit: 1`. The source has three requests, while the frequency table has two keys. Their recorded counts add up to three; this example does not print a separate key-count line.

The list contains map and kit in first-seen order. It never receives the repeated map request, so the report has no duplicate row. If input is empty, neither loop body runs and this example prints no lines. Nothing needs a first key to exist before traversal starts. The independent task later adds an explicit key-count report.

### Choose a structure for the information you need

Different structures preserve different information. Choosing one does not make the others unnecessary.

| Requirement | Useful representation | Reason |
|---|---|---|
| Keep a fixed supplied sequence, including repetition | Array | It retains a fixed number of positions in source order. |
| Append values in a sequence that changes size | ArrayList | It keeps ordered positions and permits duplicate values. |
| Remember distinct values or ask about membership | HashSet | Equality determines whether a value is already represented. |
| Retrieve associated information using a meaningful key | HashMap | Each distinct key identifies a recorded value. |

A map's values need not be unique. The complete check below stores quantity 2 under both map and kit. Both entries remain because the keys differ. The ordered frequency report needs the source array, a map for numbers, and a list for report order; no single count tells you all of that information.


````java
import java.util.HashMap;
HashMap<String, Integer> counts = new HashMap<String, Integer>();
counts.put("map", 2);
counts.put("kit", 2);
System.out.println("Map: " + counts.get("map"));
System.out.println("Kit: " + counts.get("kit"));
System.out.println("Types: " + counts.size());
````

The lines are `Map: 2`, `Kit: 2`, and `Types: 2`. Equal quantities do not merge distinct keys. Size counts the two entries, rather than the one distinct numeric value or the sum four.
